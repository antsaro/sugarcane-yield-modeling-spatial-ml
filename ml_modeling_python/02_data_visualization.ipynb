```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns
from matplotlib import patheffects
from scipy import stats

# ============================================================================
# STYLE CONFIGURATION
# ============================================================================
plt.rcParams.update({
    # Font settings - Nature standard
    'font.family': 'sans-serif',
    'font.sans-serif': ['Arial', 'DejaVu Sans', 'Liberation Sans'],
    'font.size': 7,
    'axes.labelsize': 8,
    'axes.titlesize': 9,
    'xtick.labelsize': 6,
    'ytick.labelsize': 6,
    'legend.fontsize': 6,
    
    # Figure quality
    'figure.dpi': 600,
    'savefig.dpi': 600,
    'savefig.format': 'pdf',
    
    # Line weights - Nature prefers thinner lines
    'axes.linewidth': 0.5,
    'grid.linewidth': 0.3,
    'xtick.major.width': 0.5,
    'ytick.major.width': 0.5,
    'xtick.major.size': 2.5,
    'ytick.major.size': 2.5,
    
    # Colors - Nature prefers black axes on white
    'axes.edgecolor': 'black',
    'axes.labelcolor': 'black',
    'xtick.color': 'black',
    'ytick.color': 'black',
    'text.color': 'black',
    
    # Background
    'figure.facecolor': 'white',
    'axes.facecolor': 'white',
    'savefig.facecolor': 'white',
    
    # No spines on top/right (Nature style)
    'axes.spines.top': False,
    'axes.spines.right': False,
})

# Load the dataset
df = pd.read_csv('C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/final_dataset_for_analysis_with_quality_metrics.csv', 
                 sep=';')

print("="*80)
print("CREATING NATURE JOURNAL STYLE VISUALIZATIONS WITH OUTLIER DETECTION")
print("="*80)

# Rename columns
columns_to_rename = {
    "SC_pol_juice_percent": "pol_juice_percent",
    "SC_purity_percent": "purity_percent",
    "SC_fiber_percent": "fiber_percent",
    "SC_pol_cane_percent": "pol_cane_percent",
    "SC_percent_cane_percent": "percent_cane_percent",
    "SC_glucose_cane_percent": "glucose_cane_percent",
    "SC_hugot_sugar": "hugot_sugar",
    "SC_Pur_Tc": "pur_tc",
    "SC_fiber_content_percent": "fiber_content_percent",
    "SC_juice_tons": "juice_tons",
    "SC_tmss": "tmss",
    "SC_tpj": "tpj",
    "SC_tpc": "tpc",
    "SC_ts": "ts",
    "SC_tsr": "tsr",
    "SC_brix_percent": "brix_percent",
    "SC_tc_per_ha_actual": "tc_per_ha_actual",
    "SC_tc_per_ha_per_week": "tc_per_ha_per_week",
    "SC_tc_per_ha_52_weeks" : "tc_per_ha_52_weeks"
}

actual_renames = {old: new for old, new in columns_to_rename.items() if old in df.columns}
df_renamed = df.rename(columns=actual_renames)

# Define metrics with Nature-style labels
sugar_content_metrics = {
    "brix_percent": "Brix (%)",
    "pol_juice_percent": "Pol juice (%)",
    "pol_cane_percent": "Pol cane (%)",
    "purity_percent": "Purity (%)",
    "glucose_cane_percent": "Glucose (%)",
    "hugot_sugar": "Hugot sugar (%)"
}

yield_parameters = {
    "tc_per_ha_actual": "Cane yield (t ha⁻¹)",
    "tc_per_ha_per_week" : "Cane yield per week (t ha⁻¹/week)",
    "tc_per_ha_52_weeks" : "Cane yield 52 weeks (t ha⁻¹)",
    "fiber_percent": "Fiber in Cane (%)",

}

# Filter available metrics
available_sugar = {k: v for k, v in sugar_content_metrics.items() if k in df_renamed.columns}
available_yield = {k: v for k, v in yield_parameters.items() if k in df_renamed.columns}

# Nature color palette
NATURE_COLORS = {
    'blue': '#0173B2',
    'orange': '#DE8F05',
    'green': '#029E73',
    'red': '#CC78BC',
    'cyan': '#56B4E9',
    'yellow': '#ECE133',
    'outlier_red': '#D55E00'  # Red for outliers
}

def detect_outliers_zscore(data, threshold=3):
    """
    Detect outliers using z-score method.
    Returns boolean mask where True indicates outlier.
    """
    z_scores = np.abs(stats.zscore(data))
    return z_scores > threshold

def create_enhanced_boxplot(ax, data, color, label):
    """
    Create boxplot with outlier detection and comprehensive statistics.
    """
    # Detect outliers using z-score (threshold = 3)
    outlier_mask = detect_outliers_zscore(data)
    outliers = data[outlier_mask]
    normal_data = data[~outlier_mask]
    n_outliers = len(outliers)
    
    # Calculate statistics
    n_total = len(data)
    mean_val = data.mean()
    std_val = data.std()
    median_val = data.median()
    min_val = data.min()
    max_val = data.max()
    
    # Create boxplot without outliers (we'll plot them separately)
    bp = ax.boxplot([normal_data], 
                   positions=[0],
                   widths=0.15,
                   patch_artist=True,
                   showfliers=False,  # Don't show default outliers
                   notch=False,
                   boxprops=dict(facecolor=color, 
                               edgecolor='black', 
                               linewidth=0.8,
                               alpha=0.7),
                   whiskerprops=dict(color='black', linewidth=0.8),
                   capprops=dict(color='black', linewidth=0.8),
                   medianprops=dict(color='black', linewidth=1.2))
    
    # Plot outliers in red
    if n_outliers > 0:
        outlier_positions = np.zeros(n_outliers)
        # Add small jitter to outliers for visibility
        jitter = np.random.normal(0, 0.02, n_outliers)
        ax.scatter(outlier_positions + jitter, outliers, 
                  color=NATURE_COLORS['outlier_red'], 
                  marker='o', 
                  s=12,
                  alpha=0.6,
                  edgecolors='darkred',
                  linewidths=0.5,
                  zorder=3,
                  label='Outliers (|z| > 3)')
    
    # Styling
    ax.set_ylabel(label, fontsize=8, fontweight='normal')
    ax.set_xlim(-0.5, 0.5)
    ax.set_xticks([])
    ax.tick_params(axis='y', labelsize=6, length=2.5, width=0.5)
    
    # Subtle grid
    ax.yaxis.grid(True, linestyle='-', linewidth=0.3, 
                 color='gray', alpha=0.3, zorder=0)
    ax.set_axisbelow(True)
    
    # Enhanced statistics panel
    stats_text = (
        f'n = {n_total}\n'
        f'Mean: {mean_val:.2f} ± {std_val:.2f}\n'
        f'Median: {median_val:.2f}\n'
        f'Range: [{min_val:.2f}, {max_val:.2f}]\n'
    )
    
    # Add outlier info in italic
    if n_outliers > 0:
        outlier_text = f'Outliers: {n_outliers} ({n_outliers/n_total*100:.1f}%)'
    else:
        outlier_text = 'Outliers: 0'
    
    # Main statistics box
    ax.text(0.98, 0.98, stats_text, 
           transform=ax.transAxes, 
           ha='right', va='top',
           fontsize=5,
           color='black',
           linespacing=1.5,
           bbox=dict(boxstyle='round,pad=0.4', 
                   facecolor='white', 
                   edgecolor='gray', 
                   alpha=0.9, 
                   linewidth=0.5))
    
    # Outlier info box (italic, at bottom)
    ax.text(0.98, 0.02, outlier_text, 
           transform=ax.transAxes, 
           ha='right', va='bottom',
           fontsize=5,
           color=NATURE_COLORS['outlier_red'] if n_outliers > 0 else 'gray',
           fontstyle='italic',
           bbox=dict(boxstyle='round,pad=0.3', 
                   facecolor='white', 
                   edgecolor=NATURE_COLORS['outlier_red'] if n_outliers > 0 else 'gray', 
                   alpha=0.9, 
                   linewidth=0.5))
    
    return n_outliers

# ============================================================================
# SUGAR CONTENT METRICS
# ============================================================================
print("\nCreating Sugar Content Metrics (Enhanced)...")

if available_sugar:
    n_metrics = len(available_sugar)
    ncols = 3
    nrows = int(np.ceil(n_metrics / ncols))
    
    fig, axes = plt.subplots(nrows, ncols, figsize=(5.5, nrows * 2.5))
    
    if n_metrics == 1:
        axes = [axes]
    else:
        axes = axes.flatten() if nrows > 1 else axes
    
    total_outliers_sugar = 0
    
    for idx, (metric, label) in enumerate(available_sugar.items()):
        ax = axes[idx]
        data = df_renamed[metric].dropna()
        
        if len(data) > 0:
            n_outliers = create_enhanced_boxplot(ax, data, NATURE_COLORS['blue'], label)
            total_outliers_sugar += n_outliers
    
    # Hide extra subplots
    for idx in range(n_metrics, len(axes)):
        axes[idx].axis('off')
    
    plt.tight_layout()
    
    # Add figure label
    fig.text(0.02, 0.98, 'a', fontsize=12, fontweight='bold', 
             va='top', ha='left')
    
    # Save
    output_path = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Fig1_sugar_metrics_nature_enhanced.pdf'
    plt.savefig(output_path, dpi=600, bbox_inches='tight', 
                facecolor='white', edgecolor='none',
                format='pdf', backend='pdf')
    print(f"Saved: {output_path}")
    print(f"Total outliers detected (Sugar metrics): {total_outliers_sugar}")
    
    output_path_png = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Fig1_sugar_metrics_nature_enhanced.png'
    plt.savefig(output_path_png, dpi=600, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    print(f"Saved: {output_path_png}")
    
    plt.show()

# ============================================================================
# YIELD PARAMETERS - ENHANCED
# ============================================================================
print("\nCreating Yield Parameters (Enhanced)...")

if available_yield:
    n_metrics = len(available_yield)
    ncols = 3
    nrows = int(np.ceil(n_metrics / ncols))
    
    fig, axes = plt.subplots(nrows, ncols, figsize=(5.5, nrows * 2.5))
    
    if n_metrics == 1:
        axes = [axes]
    else:
        axes = axes.flatten() if nrows > 1 else axes
    
    total_outliers_yield = 0
    
    for idx, (metric, label) in enumerate(available_yield.items()):
        ax = axes[idx]
        data = df_renamed[metric].dropna()
        
        if len(data) > 0:
            n_outliers = create_enhanced_boxplot(ax, data, NATURE_COLORS['orange'], label)
            total_outliers_yield += n_outliers
    
    # Hide extra subplots
    for idx in range(n_metrics, len(axes)):
        axes[idx].axis('off')
    
    plt.tight_layout()
    
    # Add figure label
    fig.text(0.02, 0.98, 'b', fontsize=12, fontweight='bold', 
             va='top', ha='left')
    
    # Save
    output_path = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Fig2_yield_params_nature_enhanced.pdf'
    plt.savefig(output_path, dpi=600, bbox_inches='tight', 
                facecolor='white', edgecolor='none',
                format='pdf', backend='pdf')
    print(f"Saved: {output_path}")
    print(f"Total outliers detected (Yield parameters): {total_outliers_yield}")
    
    output_path_png = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Fig2_yield_params_nature_enhanced.png'
    plt.savefig(output_path_png, dpi=600, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    print(f"Saved: {output_path_png}")
    
    plt.show()
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

# ============================================================================
# STYLE CONFIGURATION
# ============================================================================

# Professional Nature journal style parameters
plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 7
plt.rcParams['axes.linewidth'] = 0.6
plt.rcParams['xtick.major.width'] = 0.6
plt.rcParams['ytick.major.width'] = 0.6
plt.rcParams['xtick.major.size'] = 2.5
plt.rcParams['ytick.major.size'] = 2.5
plt.rcParams['xtick.minor.size'] = 1.5
plt.rcParams['ytick.minor.size'] = 1.5
plt.rcParams['figure.dpi'] = 600
plt.rcParams['savefig.dpi'] = 600
plt.rcParams['savefig.bbox'] = 'tight'
plt.rcParams['savefig.pad_inches'] = 0.02
plt.rcParams['axes.labelpad'] = 2
plt.rcParams['axes.titlepad'] = 4

# Response variable display names with units
RESPONSE_LABELS = {
    "brix_percent": "Brix (%)",
    "pol_juice_percent": "Pol juice (%)",
    "pol_cane_percent": "Pol cane (%)",
    "purity_percent": "Purity (%)",
    "glucose_cane_percent": "Glucose (%)",
    "hugot_sugar": "Hugot sugar (%)",
    "tc_per_ha_actual": "Cane yield (t/ha)",
    "tc_per_ha_per_week": "Yield/week (t/ha/wk)",
    "tc_per_ha_52_weeks": "Yield 52-wk (t/ha)",
    "fiber_percent": "Fiber (%)"
}

# Model name abbreviations for compact display
MODEL_ABBREV = {
    'LinearRegression': 'LR',
    'KNN': 'KNN',
    'DecisionTree': 'DT',
    'RandomForest': 'RF',
    'ExtraTrees': 'ET',
    'GradientBoosting': 'GB',
    'BaggingRegressor': 'Bag',
    'XGBoost': 'XGB',
    'LightGBM': 'LGBM',
    'CatBoost': 'CB',
    'GAM': 'GAM'  # Added GAM
}

# Model color mapping using viridis-like palette
MODEL_COLORS = {
    'LinearRegression': '#440154',  # Deep purple
    'KNN': '#31688e',               # Blue
    'DecisionTree': '#35b779',      # Green
    'RandomForest': '#fde724',      # Yellow
    'ExtraTrees': '#21918c',        # Teal
    'GradientBoosting': '#443983',  # Purple-blue
    'BaggingRegressor': '#90d743',  # Light green
    'XGBoost': '#feb72d',           # Orange
    'LightGBM': '#1f9e89',          # Cyan
    'CatBoost': '#cc4778',          # Pink
    'GAM': '#e63946'                # Red (distinctive for GAM)
}

# ============================================================================
# GAM RESULTS CONFIGURATION
# ============================================================================

# Define your GAM model results here
# Structure: {response_variable: {metric: value}}
GAM_RESULTS = {
    'tc_per_ha_actual': {
        'Overall_R2': 0.856,        # Replace with your actual value
        'Overall_RMSE': 12.34,      # Replace with your actual value
        'Overall_MAE': 9.87,        # Replace with your actual value
        'Overall_RPD': 2.54,        # Replace with your actual value
        'Overall_RPIQ': 3.21,       # Replace with your actual value
        'Mean_CV_R2': 0.845,        # Optional: CV metrics
        'Std_CV_R2': 0.023
    },
    'brix_percent': {
        'Overall_R2': 0.782,        # Replace with your actual value
        'Overall_RMSE': 0.98,       # Replace with your actual value
        'Overall_MAE': 0.76,        # Replace with your actual value
        'Overall_RPD': 2.13,        # Replace with your actual value
        'Overall_RPIQ': 2.87        # Replace with your actual value
    },
    'glucose_cane_percent': {
        'Overall_R2': 0.723,        # Replace with your actual value
        'Overall_RMSE': 0.45,       # Replace with your actual value
        'Overall_MAE': 0.34,        # Replace with your actual value
        'Overall_RPD': 1.89,        # Replace with your actual value
        'Overall_RPIQ': 2.45        # Replace with your actual value
    },
    'purity_percent': {
        'Overall_R2': 0.698,        # Replace with your actual value
        'Overall_RMSE': 1.23,       # Replace with your actual value
        'Overall_MAE': 0.98,        # Replace with your actual value
        'Overall_RPD': 1.76,        # Replace with your actual value
        'Overall_RPIQ': 2.12        # Replace with your actual value
    }
    # Add more response variables as needed
}

# ============================================================================
# DATA LOADING FUNCTIONS
# ============================================================================

def find_result_folders(base_dir, model_type='all', response_variables=None):
    """Find all result folders containing kfold_cv_summary files"""
    result_folders = {}
    base_path = Path(base_dir)
    
    if not base_path.exists():
        print(f"Warning: Base directory not found: {base_dir}")
        return result_folders
    
    for folder in base_path.glob('kfold_models_*'):
        if folder.is_dir():
            folder_name = folder.name
            
            if model_type == 'custom' and '_custom' not in folder_name:
                continue
            elif model_type == 'with_categorical' and '_with_categorical' not in folder_name:
                continue
            
            response_var = folder_name.replace('kfold_models_', '')
            
            if '_custom' in response_var:
                response_var = response_var.replace('_custom', '')
                suffix = '_custom'
            elif '_with_categorical' in response_var:
                response_var = response_var.replace('_with_categorical', '')
                suffix = '_with_categorical'
            else:
                suffix = ''
            
            if response_variables is not None and response_var not in response_variables:
                continue
            
            summary_files = list(folder.glob('kfold_cv_summary_*.csv'))
            if summary_files:
                result_folders[response_var] = {
                    'folder': str(folder),
                    'summary_file': str(summary_files[0]),
                    'type': suffix.replace('_', '') if suffix else 'standard'
                }
    
    return result_folders

def add_gam_results_to_dataframe(df, response_var, gam_results_dict):
    """Add GAM results as a new row in the dataframe"""
    if response_var not in gam_results_dict:
        return df
    
    gam_metrics = gam_results_dict[response_var]
    
    # Create GAM row with same structure as other models
    gam_row = {
        'Model': 'GAM',
        'Complexity': 'Complex',  # You can adjust this
    }
    
    # Add all metrics from GAM_RESULTS
    gam_row.update(gam_metrics)
    
    # Append to dataframe
    gam_df = pd.DataFrame([gam_row])
    df = pd.concat([df, gam_df], ignore_index=True)
    
    return df

def load_all_results(base_dir, model_type='all', models_to_include=None, 
                    response_variables=None, include_gam=True, gam_results_dict=None):
    """Load all model results from different response variables, optionally including GAM"""
    result_folders = find_result_folders(base_dir, model_type, response_variables)
    
    if not result_folders and not include_gam:
        return None
    
    all_results = {}
    
    # Load regular model results
    for response_var, paths in result_folders.items():
        try:
            df = pd.read_csv(paths['summary_file'])
            
            if models_to_include is not None:
                df = df[df['Model'].isin(models_to_include)].copy()
            
            # Add GAM results if requested
            if include_gam and gam_results_dict is not None:
                df = add_gam_results_to_dataframe(df, response_var, gam_results_dict)
            
            if len(df) > 0:
                all_results[response_var] = df
                
        except Exception as e:
            print(f"Error loading {response_var}: {e}")
    
    # If only GAM results requested (no other models)
    if include_gam and gam_results_dict is not None and not result_folders:
        for response_var, gam_metrics in gam_results_dict.items():
            if response_variables is None or response_var in response_variables:
                gam_row = {'Model': 'GAM', 'Complexity': 'Complex'}
                gam_row.update(gam_metrics)
                all_results[response_var] = pd.DataFrame([gam_row])
    
    return all_results

# ============================================================================
# COMPACT PLOTTING FUNCTIONS
# ============================================================================

def create_metric_comparison_grid(all_results, metric='Overall_R2', 
                                   metric_label='R²', 
                                   output_path='metric_comparison.png',
                                   figsize_per_plot=(1.6, 1.3),
                                   use_abbrev=True,
                                   highlight_gam=True):
    """
    Create ultra-compact grid of bar plots with Nature styling
    GAM bars are highlighted with special styling
    """
    n_responses = len(all_results)
    
    if n_responses == 0:
        print(f"No data to plot for {metric}")
        return
    
    # Optimal grid dimensions for space efficiency
    n_cols = min(4, n_responses)  # Max 4 columns for compactness
    n_rows = int(np.ceil(n_responses / n_cols))
    
    fig_width = figsize_per_plot[0] * n_cols
    fig_height = figsize_per_plot[1] * n_rows
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(fig_width, fig_height))
    
    if n_responses == 1:
        axes = np.array([axes])
    axes = axes.flatten() if n_responses > 1 else [axes]
    
    for idx, (response_var, df) in enumerate(all_results.items()):
        ax = axes[idx]
        
        if metric not in df.columns:
            ax.axis('off')
            continue
        
        df_sorted = df.sort_values(metric, ascending=True)
        
        # Use abbreviated model names if requested
        model_names = [MODEL_ABBREV.get(m, m) if use_abbrev else m for m in df_sorted['Model']]
        
        colors = [MODEL_COLORS.get(model, '#666666') for model in df_sorted['Model']]
        
        # Highlight GAM with special edge color
        edgecolors = ['red' if model == 'GAM' else 'white' for model in df_sorted['Model']]
        linewidths = [1.2 if model == 'GAM' else 0.3 for model in df_sorted['Model']]
        
        # Horizontal bars with thin borders - SLIMMER bars
        bars = ax.barh(range(len(model_names)), df_sorted[metric], 
                       color=colors, edgecolor=edgecolors, linewidth=linewidths, height=0.5)
        
        # Minimal formatting
        ax.set_yticks(range(len(model_names)))
        ax.set_yticklabels(model_names, fontsize=6.5, fontweight='normal')
        
        # Make GAM label bold
        for i, (label, model) in enumerate(zip(ax.get_yticklabels(), df_sorted['Model'])):
            if model == 'GAM':
                label.set_fontweight('bold')
                label.set_color('#e63946')
        
        ax.set_xlabel(metric_label, fontsize=7, fontweight='bold')
        
        # Compact title
        title = RESPONSE_LABELS.get(response_var, response_var)
        ax.set_title(title, fontsize=7.5, fontweight='bold', pad=3)
        
        ax.tick_params(axis='x', labelsize=6)
        ax.tick_params(axis='y', length=0)
        
        # Minimal grid
        ax.grid(axis='x', alpha=0.2, linewidth=0.3, linestyle='--')
        ax.set_axisbelow(True)
        
        # Clean spines
        for spine in ['top', 'right', 'left']:
            ax.spines[spine].set_visible(False)
        ax.spines['bottom'].set_linewidth(0.6)
        
        # Value labels - compact
        for i, (bar, val, model) in enumerate(zip(bars, df_sorted[metric], df_sorted['Model'])):
            if val > 0:
                # Bold white text for GAM
                fontweight = 'bold' if model == 'GAM' else 'bold'
                ax.text(val * 0.98, bar.get_y() + bar.get_height()/2, 
                       f'{val:.2f}', 
                       ha='right', va='center', fontsize=5.5, 
                       fontweight=fontweight, color='white')
    
    # Hide unused subplots
    for idx in range(n_responses, len(axes)):
        axes[idx].axis('off')
    
    plt.subplots_adjust(hspace=0.4, wspace=0.35)
    plt.savefig(output_path, dpi=600, bbox_inches='tight', 
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

def create_comprehensive_comparison(all_results, metric='Overall_R2',
                                   metric_label='R²',
                                   output_path='comprehensive_comparison.png',
                                   figsize=(7.08, 3.5),
                                   use_abbrev=True,
                                   highlight_gam=True):
    """
    Compact grouped bar chart with Nature styling
    GAM bars are highlighted
    """
    plot_data = []
    for response_var, df in all_results.items():
        if metric not in df.columns:
            continue
        for _, row in df.iterrows():
            plot_data.append({
                'Response': RESPONSE_LABELS.get(response_var, response_var),
                'Model': row['Model'],
                'Value': row[metric]
            })
    
    if not plot_data:
        return
    
    plot_df = pd.DataFrame(plot_data)
    
    fig, ax = plt.subplots(figsize=figsize)
    
    models = sorted(plot_df['Model'].unique())
    responses = sorted(plot_df['Response'].unique())
    
    x = np.arange(len(responses))
    width = 0.6 / len(models)  # SLIMMER bars
    
    for i, model in enumerate(models):
        model_data = plot_df[plot_df['Model'] == model]
        values = [model_data[model_data['Response'] == resp]['Value'].values[0] 
                 if len(model_data[model_data['Response'] == resp]) > 0 else 0 
                 for resp in responses]
        
        offset = (i - len(models)/2) * width + width/2
        color = MODEL_COLORS.get(model, '#666666')
        label = MODEL_ABBREV.get(model, model) if use_abbrev else model
        
        # Special styling for GAM
        if model == 'GAM':
            edgecolor = 'red'
            linewidth = 1.2
        else:
            edgecolor = 'white'
            linewidth = 0.3
        
        ax.bar(x + offset, values, width, label=label, 
              color=color, edgecolor=edgecolor, linewidth=linewidth)
    
    # Formatting
    ax.set_ylabel(metric_label, fontsize=8, fontweight='bold')
    ax.set_title(f'Model performance comparison', 
                fontsize=9, fontweight='bold', pad=6)
    
    ax.set_xticks(x)
    ax.set_xticklabels(responses, rotation=45, ha='right', fontsize=7)
    ax.tick_params(axis='y', labelsize=7)
    ax.tick_params(axis='x', length=2)
    
    # Minimal grid
    ax.grid(axis='y', alpha=0.2, linewidth=0.3, linestyle='--')
    ax.set_axisbelow(True)
    
    # Clean spines
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    for spine in ['left', 'bottom']:
        ax.spines[spine].set_linewidth(0.6)
    
    # Compact legend - highlight GAM
    legend = ax.legend(loc='upper left', bbox_to_anchor=(1, 1), 
                      fontsize=6, frameon=True, fancybox=False, 
                      shadow=False, ncol=1, borderaxespad=0,
                      edgecolor='#CCCCCC', framealpha=0.95)
    
    # Make GAM legend text bold and red
    for text in legend.get_texts():
        if text.get_text() == 'GAM':
            text.set_fontweight('bold')
            text.set_color('#e63946')
    
    plt.tight_layout()
    plt.savefig(output_path, dpi=600, bbox_inches='tight',
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

def create_best_model_summary(all_results, output_path='best_models_summary.png',
                              figsize=(7.08, 2.5)):
    """
    Compact multi-metric comparison of best models
    """
    best_models = []
    
    for response_var, df in all_results.items():
        if 'Overall_R2' not in df.columns:
            continue
            
        best_idx = df['Overall_R2'].idxmax()
        best_row = df.loc[best_idx]
        
        best_models.append({
            'Response': RESPONSE_LABELS.get(response_var, response_var),
            'Model': MODEL_ABBREV.get(best_row['Model'], best_row['Model']),
            'R²': best_row['Overall_R2'],
            'RMSE': best_row.get('Overall_RMSE', np.nan),
            'RPD': best_row.get('Overall_RPD', np.nan),
            'RPIQ': best_row.get('Overall_RPIQ', np.nan)
        })
    
    if not best_models:
        return
    
    best_df = pd.DataFrame(best_models)
    
    fig, axes = plt.subplots(1, 4, figsize=figsize)
    
    metrics = [
        ('R²', 'R²', 'viridis'),
        ('RMSE', 'RMSE', 'plasma'),
        ('RPD', 'RPD', 'cividis'),
        ('RPIQ', 'RPIQ', 'viridis')
    ]
    
    for ax, (metric, label, cmap) in zip(axes, metrics):
        if metric in best_df.columns and not best_df[metric].isna().all():
            values = best_df[metric].values
            norm_values = (values - values.min()) / (values.max() - values.min() + 1e-10)
            colors = plt.cm.get_cmap(cmap)(norm_values)
            
            bars = ax.barh(range(len(best_df)), best_df[metric], 
                          color=colors, edgecolor='white', linewidth=0.4, height=0.5)
            
            ax.set_yticks(range(len(best_df)))
            if ax == axes[0]:
                ax.set_yticklabels(best_df['Response'], fontsize=6.5)
            else:
                ax.set_yticklabels([])
            
            ax.set_xlabel(label, fontsize=7.5, fontweight='bold')
            ax.set_title(f'{label}', fontsize=8, fontweight='bold', pad=3)
            ax.tick_params(axis='both', labelsize=6, length=2)
            ax.tick_params(axis='y', length=0)
            
            ax.grid(axis='x', alpha=0.2, linewidth=0.3, linestyle='--')
            ax.set_axisbelow(True)
            
            for spine in ['top', 'right', 'left']:
                ax.spines[spine].set_visible(False)
            ax.spines['bottom'].set_linewidth(0.6)
            
            # Value labels
            for i, (bar, val) in enumerate(zip(bars, best_df[metric])):
                if not np.isnan(val):
                    ax.text(val * 0.95, bar.get_y() + bar.get_height()/2,
                           f'{val:.2f}',
                           ha='right', va='center', fontsize=5.5,
                           fontweight='bold', color='white')
    
    plt.subplots_adjust(wspace=0.25)
    plt.savefig(output_path, dpi=600, bbox_inches='tight',
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

def create_heatmap_comparison(all_results, metric='Overall_R2',
                              metric_label='R²',
                              output_path='heatmap_comparison.png',
                              figsize=(7.08, 3.5),
                              use_abbrev=True):
    """
    Compact heatmap with viridis colormap, white gridlines, and larger fonts
    GAM row is highlighted
    """
    all_models = set()
    for df in all_results.values():
        all_models.update(df['Model'].tolist())
    models_unique = sorted(list(all_models))
    
    responses = sorted([RESPONSE_LABELS.get(rv, rv) for rv in all_results.keys()])
    response_var_keys = sorted(list(all_results.keys()))
    
    data_matrix = []
    for model in models_unique:
        row = []
        for response_var in response_var_keys:
            df = all_results[response_var]
            if metric not in df.columns:
                row.append(np.nan)
                continue
            
            model_data = df[df['Model'] == model]
            if len(model_data) > 0:
                row.append(model_data[metric].values[0])
            else:
                row.append(np.nan)
        data_matrix.append(row)
    
    model_labels = [MODEL_ABBREV.get(m, m) if use_abbrev else m for m in models_unique]
    heatmap_df = pd.DataFrame(data_matrix, columns=responses, index=model_labels)
    
    fig, ax = plt.subplots(figsize=figsize)
    
    # Use viridis colormap
    im = ax.imshow(heatmap_df.values, cmap='viridis', aspect='auto',
                   vmin=np.nanmin(heatmap_df.values), 
                   vmax=np.nanmax(heatmap_df.values))
    
    ax.set_xticks(np.arange(len(responses)))
    ax.set_yticks(np.arange(len(model_labels)))
    ax.set_xticklabels(responses, rotation=45, ha='right', fontsize=8)
    ax.set_yticklabels(model_labels, fontsize=8)
    
    # Highlight GAM row
    for i, (label, model) in enumerate(zip(ax.get_yticklabels(), models_unique)):
        if model == 'GAM':
            label.set_fontweight('bold')
            label.set_color('#e63946')
    
    ax.set_xlabel('Response variable', fontsize=9, fontweight='bold')
    ax.set_ylabel('Model', fontsize=9, fontweight='bold')
    ax.set_title(f'Performance heatmap: {metric_label}', 
                fontsize=10, fontweight='bold', pad=6)
    
    ax.tick_params(axis='both', length=2)
    
    # Add WHITE GRIDLINES between cells
    for i in range(len(model_labels) + 1):
        ax.axhline(i - 0.5, color='white', linewidth=1.5, zorder=10)
    for j in range(len(responses) + 1):
        ax.axvline(j - 0.5, color='white', linewidth=1.5, zorder=10)
    
    # Add special highlight for GAM row
    gam_idx = None
    for i, model in enumerate(models_unique):
        if model == 'GAM':
            gam_idx = i
            break
    
    if gam_idx is not None:
        # Add red border around GAM row
        rect = plt.Rectangle((-0.5, gam_idx - 0.5), len(responses), 1,
                            fill=False, edgecolor='red', linewidth=2, zorder=11)
        ax.add_patch(rect)
    
    # Add values in cells with BIGGER FONT
    for i in range(len(model_labels)):
        for j in range(len(responses)):
            val = heatmap_df.iloc[i, j]
            if not np.isnan(val):
                # White text for dark cells, black for light cells
                luminance = (val - np.nanmin(heatmap_df.values)) / (np.nanmax(heatmap_df.values) - np.nanmin(heatmap_df.values))
                text_color = 'white' if luminance < 0.5 else 'black'
                
                # Bold for GAM
                fontweight = 'bold' if models_unique[i] == 'GAM' else 'bold'
                
                ax.text(j, i, f'{val:.2f}',
                       ha='center', va='center', 
                       color=text_color,
                       fontsize=7, fontweight=fontweight)
    
    # Compact colorbar
    cbar = plt.colorbar(im, ax=ax, fraction=0.035, pad=0.02)
    cbar.set_label(metric_label, rotation=270, labelpad=12, 
                  fontsize=8, fontweight='bold')
    cbar.ax.tick_params(labelsize=7)
    cbar.outline.set_linewidth(0.6)
    
    for spine in ax.spines.values():
        spine.set_linewidth(0.6)
    
    plt.tight_layout()
    plt.savefig(output_path, dpi=600, bbox_inches='tight',
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

# Keep all other plotting functions the same...
# (create_model_ranking_plot, create_metric_correlation_plot, create_performance_distribution)
# Just add highlight_gam parameter where needed

def create_model_ranking_plot(all_results, metric='Overall_R2',
                              metric_label='R²',
                              output_path='model_ranking.png',
                              figsize=(7.08, 3.5),
                              use_abbrev=True):
    """
    Create a compact ranking visualization showing model performance across responses
    GAM highlighted if present
    """
    # Calculate average rank for each model
    model_ranks = {}
    
    for response_var, df in all_results.items():
        if metric not in df.columns:
            continue
        
        df_sorted = df.sort_values(metric, ascending=False)
        for rank, row in enumerate(df_sorted.itertuples(), start=1):
            model = row.Model
            if model not in model_ranks:
                model_ranks[model] = []
            model_ranks[model].append(rank)
    
    # Calculate statistics
    rank_stats = []
    for model, ranks in model_ranks.items():
        rank_stats.append({
            'Model': MODEL_ABBREV.get(model, model) if use_abbrev else model,
            'Mean_Rank': np.mean(ranks),
            'Median_Rank': np.median(ranks),
            'Best_Rank': np.min(ranks),
            'Worst_Rank': np.max(ranks),
            'Full_Model': model
        })
    
    rank_df = pd.DataFrame(rank_stats).sort_values('Mean_Rank')
    
    fig, ax = plt.subplots(figsize=figsize)
    
    y_pos = np.arange(len(rank_df))
    
    # Plot mean rank with error bars showing range
    colors = [MODEL_COLORS.get(m, '#666666') for m in rank_df['Full_Model']]
    
    # Special styling for GAM
    alphas = [0.9 if m == 'GAM' else 0.7 for m in rank_df['Full_Model']]
    edgecolors = ['red' if m == 'GAM' else 'white' for m in rank_df['Full_Model']]
    linewidths = [1.2 if m == 'GAM' else 0.4 for m in rank_df['Full_Model']]
    
    for i, (color, alpha, edge, lw) in enumerate(zip(colors, alphas, edgecolors, linewidths)):
        ax.barh([i], [rank_df.iloc[i]['Mean_Rank']], 
               color=color, alpha=alpha, edgecolor=edge, linewidth=lw, height=0.5)
    
    # Add range indicators
    for i, row in rank_df.iterrows():
        ax.plot([row['Best_Rank'], row['Worst_Rank']], [i, i],
               color='black', linewidth=1.5, alpha=0.8, zorder=10)
        ax.plot([row['Best_Rank'], row['Best_Rank']], [i-0.15, i+0.15],
               color='black', linewidth=1.5, alpha=0.8, zorder=10)
        ax.plot([row['Worst_Rank'], row['Worst_Rank']], [i-0.15, i+0.15],
               color='black', linewidth=1.5, alpha=0.8, zorder=10)
    
    ax.set_yticks(y_pos)
    ax.set_yticklabels(rank_df['Model'], fontsize=7)
    
    # Highlight GAM label
    for i, (label, model) in enumerate(zip(ax.get_yticklabels(), rank_df['Full_Model'])):
        if model == 'GAM':
            label.set_fontweight('bold')
            label.set_color('#e63946')
    
    ax.set_xlabel('Ranking (lower is better)', fontsize=8, fontweight='bold')
    ax.set_title('Model ranking across all response variables', 
                fontsize=9, fontweight='bold', pad=6)
    
    ax.tick_params(axis='both', labelsize=6.5, length=2)
    ax.tick_params(axis='y', length=0)
    
    ax.invert_xaxis()  # Lower ranks on right
    ax.grid(axis='x', alpha=0.2, linewidth=0.3, linestyle='--')
    ax.set_axisbelow(True)
    
    for spine in ['top', 'right', 'left']:
        ax.spines[spine].set_visible(False)
    ax.spines['bottom'].set_linewidth(0.6)
    
    # Add legend
    from matplotlib.patches import Rectangle
    legend_elements = [
        Rectangle((0, 0), 1, 1, fc='gray', alpha=0.7, label='Mean rank'),
        plt.Line2D([0], [0], color='black', linewidth=1.5, label='Best-worst range')
    ]
    ax.legend(handles=legend_elements, loc='lower right', fontsize=6, 
             frameon=True, edgecolor='#CCCCCC', framealpha=0.95)
    
    plt.tight_layout()
    plt.savefig(output_path, dpi=600, bbox_inches='tight',
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

def create_metric_correlation_plot(all_results, 
                                   output_path='metric_correlation.png',
                                   figsize=(3.54, 3.54)):
    """
    Compact correlation plot between different metrics
    GAM highlighted with special marker
    """
    metrics_data = []
    
    for response_var, df in all_results.items():
        for _, row in df.iterrows():
            metrics_data.append({
                'Model': row['Model'],
                'Response': response_var,
                'R²': row.get('Overall_R2', np.nan),
                'RMSE': row.get('Overall_RMSE', np.nan),
                'RPD': row.get('Overall_RPD', np.nan),
                'RPIQ': row.get('Overall_RPIQ', np.nan)
            })
    
    if not metrics_data:
        return
    
    metrics_df = pd.DataFrame(metrics_data).dropna()
    
    fig, ax = plt.subplots(figsize=figsize)
    
    # Scatter plot R² vs RPD colored by model
    models = metrics_df['Model'].unique()
    
    for model in models:
        model_data = metrics_df[metrics_df['Model'] == model]
        color = MODEL_COLORS.get(model, '#666666')
        
        # Special marker for GAM
        if model == 'GAM':
            marker = 's'  # Square
            s = 50  # Larger
            edgecolor = 'red'
            linewidth = 1.5
            alpha = 0.9
        else:
            marker = 'o'  # Circle
            s = 30
            edgecolor = 'white'
            linewidth = 0.5
            alpha = 0.7
        
        ax.scatter(model_data['R²'], model_data['RPD'], 
                  c=[color], alpha=alpha, s=s, marker=marker,
                  edgecolors=edgecolor, linewidth=linewidth,
                  label=MODEL_ABBREV.get(model, model))
    
    ax.set_xlabel('R²', fontsize=8, fontweight='bold')
    ax.set_ylabel('RPD', fontsize=8, fontweight='bold')
    ax.set_title('Metric correlation: R² vs RPD', 
                fontsize=9, fontweight='bold', pad=6)
    
    ax.tick_params(axis='both', labelsize=7, length=2)
    ax.grid(alpha=0.2, linewidth=0.3, linestyle='--')
    ax.set_axisbelow(True)
    
    for spine in ax.spines.values():
        spine.set_linewidth(0.6)
    
    # Compact legend with GAM highlighted
    legend = ax.legend(loc='best', fontsize=5.5, frameon=True,
                      edgecolor='#CCCCCC', framealpha=0.95, ncol=2)
    
    for text in legend.get_texts():
        if text.get_text() == 'GAM':
            text.set_fontweight('bold')
            text.set_color('#e63946')
    
    plt.tight_layout()
    plt.savefig(output_path, dpi=600, bbox_inches='tight',
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

def create_performance_distribution(all_results, metric='Overall_R2',
                                   metric_label='R²',
                                   output_path='performance_distribution.png',
                                   figsize=(7.08, 2.5),
                                   use_abbrev=True):
    """
    Compact violin/box plot showing distribution of performance
    """
    plot_data = []
    for response_var, df in all_results.items():
        if metric not in df.columns:
            continue
        for _, row in df.iterrows():
            plot_data.append({
                'Model': row['Model'],
                'Value': row[metric]
            })
    
    if not plot_data:
        return
    
    plot_df = pd.DataFrame(plot_data)
    
    fig, ax = plt.subplots(figsize=figsize)
    
    models = sorted(plot_df['Model'].unique(), 
                   key=lambda x: plot_df[plot_df['Model']==x]['Value'].median(),
                   reverse=True)
    
    positions = np.arange(len(models))
    
    # Create violin plots
    parts = ax.violinplot([plot_df[plot_df['Model']==m]['Value'].values for m in models],
                          positions=positions, widths=0.6,
                          showmeans=False, showmedians=True, showextrema=True)
    
    # Color by model
    for i, (pc, model) in enumerate(zip(parts['bodies'], models)):
        pc.set_facecolor(MODEL_COLORS.get(model, '#666666'))
        
        # Special styling for GAM
        if model == 'GAM':
            pc.set_alpha(0.9)
            pc.set_edgecolor('red')
            pc.set_linewidth(1.5)
        else:
            pc.set_alpha(0.7)
            pc.set_edgecolor('white')
            pc.set_linewidth(0.5)
    
    # Style the median, min, max lines
    for partname in ['cmedians', 'cmins', 'cmaxes', 'cbars']:
        if partname in parts:
            parts[partname].set_edgecolor('black')
            parts[partname].set_linewidth(1)
    
    model_labels = [MODEL_ABBREV.get(m, m) if use_abbrev else m for m in models]
    ax.set_xticks(positions)
    ax.set_xticklabels(model_labels, rotation=45, ha='right', fontsize=7)
    
    # Highlight GAM label
    for i, (label, model) in enumerate(zip(ax.get_xticklabels(), models)):
        if model == 'GAM':
            label.set_fontweight('bold')
            label.set_color('#e63946')
    
    ax.set_ylabel(metric_label, fontsize=8, fontweight='bold')
    ax.set_title(f'Performance distribution: {metric_label}', 
                fontsize=9, fontweight='bold', pad=6)
    
    ax.tick_params(axis='both', labelsize=7, length=2)
    ax.grid(axis='y', alpha=0.2, linewidth=0.3, linestyle='--')
    ax.set_axisbelow(True)
    
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    for spine in ['left', 'bottom']:
        ax.spines[spine].set_linewidth(0.6)
    
    plt.tight_layout()
    plt.savefig(output_path, dpi=600, bbox_inches='tight',
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

# ============================================================================
# MAIN EXECUTION
# ============================================================================

def main(base_results_dir, output_dir='plots', model_type='all',
         models_to_include=None, response_variables=None, use_abbrev=True,
         include_gam=True, gam_results_dict=None):
    """
    Main function to generate all compact Nature-style plots
    
    Parameters:
    -----------
    base_results_dir : str
        Base directory containing result folders
    output_dir : str
        Directory to save output plots
    model_type : str
        Type of models to plot: 'custom', 'with_categorical', or 'all'
    models_to_include : list or None
        List of model names to include
    response_variables : list or None
        List of response variables to include
    use_abbrev : bool
        Use abbreviated model names for compact display
    include_gam : bool
        Include GAM results
    gam_results_dict : dict or None
        GAM results dictionary
    """
    print("="*80)
    print("CREATING NATURE-STYLE COMPACT MODEL COMPARISON PLOTS")
    if include_gam:
        print("Including GAM results")
    print("="*80)
    
    if models_to_include:
        print(f"\n📊 Models: {', '.join(models_to_include)}")
    else:
        print(f"\n📊 Models: All available")
    
    if include_gam:
        print(f"📊 + GAM (Generalized Additive Models)")
    
    if response_variables:
        print(f"📈 Responses: {', '.join(response_variables)}")
    else:
        print(f"📈 Responses: All available")
    
    if model_type != 'all':
        output_dir = f"{output_dir}_{model_type}"
    
    if include_gam:
        output_dir = f"{output_dir}_with_gam"
    
    os.makedirs(output_dir, exist_ok=True)
    print(f"\n💾 Output: {output_dir}")
    
    print(f"\n🔍 Loading results from: {base_results_dir}")
    all_results = load_all_results(base_results_dir, model_type, 
                                   models_to_include, response_variables,
                                   include_gam=include_gam,
                                   gam_results_dict=gam_results_dict)
    
    if not all_results:
        print(f"\n❌ Error: No results found!")
        return
    
    print(f"\n✓ Loaded {len(all_results)} response variables")
    for var, df in all_results.items():
        print(f"  • {var}: {len(df)} models")
        if include_gam and 'GAM' in df['Model'].values:
            print(f"    → GAM included")
    
    print("\n" + "="*80)
    print("GENERATING PLOTS")
    print("="*80)
    
    # 1. R² comparison grid
    print("\n1/9 R² comparison grid...")
    create_metric_comparison_grid(
        all_results, 
        metric='Overall_R2',
        metric_label='R²',
        output_path=os.path.join(output_dir, '01_r2_grid.png'),
        figsize_per_plot=(1.6, 1.3),
        use_abbrev=use_abbrev,
        highlight_gam=include_gam
    )
    
    # 2. RMSE comparison grid
    print("2/9 RMSE comparison grid...")
    create_metric_comparison_grid(
        all_results,
        metric='Overall_RMSE',
        metric_label='RMSE',
        output_path=os.path.join(output_dir, '02_rmse_grid.png'),
        figsize_per_plot=(1.6, 1.3),
        use_abbrev=use_abbrev,
        highlight_gam=include_gam
    )
    
    # 3. RPD comparison grid
    print("3/9 RPD comparison grid...")
    create_metric_comparison_grid(
        all_results,
        metric='Overall_RPD',
        metric_label='RPD',
        output_path=os.path.join(output_dir, '03_rpd_grid.png'),
        figsize_per_plot=(1.6, 1.3),
        use_abbrev=use_abbrev,
        highlight_gam=include_gam
    )
    
    # 4. Comprehensive comparison
    print("4/9 Comprehensive comparison...")
    create_comprehensive_comparison(
        all_results,
        metric='Overall_R2',
        metric_label='R²',
        output_path=os.path.join(output_dir, '04_comprehensive.png'),
        figsize=(7.08, 3.5),
        use_abbrev=use_abbrev,
        highlight_gam=include_gam
    )
    
    # 5. Best models summary
    print("5/9 Best models summary...")
    create_best_model_summary(
        all_results,
        output_path=os.path.join(output_dir, '05_best_models.png'),
        figsize=(7.08, 2.5)
    )
    
    # 6. Heatmap comparison
    print("6/9 Heatmap comparison...")
    create_heatmap_comparison(
        all_results,
        metric='Overall_R2',
        metric_label='R²',
        output_path=os.path.join(output_dir, '06_heatmap.png'),
        figsize=(7.08, 3.5),
        use_abbrev=use_abbrev
    )
    
    # 7. Model ranking
    print("7/9 Model ranking...")
    create_model_ranking_plot(
        all_results,
        metric='Overall_R2',
        metric_label='R²',
        output_path=os.path.join(output_dir, '07_ranking.png'),
        figsize=(7.08, 3.5),
        use_abbrev=use_abbrev
    )
    
    # 8. Performance distribution
    print("8/9 Performance distribution...")
    create_performance_distribution(
        all_results,
        metric='Overall_R2',
        metric_label='R²',
        output_path=os.path.join(output_dir, '08_distribution.png'),
        figsize=(7.08, 2.5),
        use_abbrev=use_abbrev
    )
    
    # 9. Metric correlation
    print("9/9 Metric correlation...")
    create_metric_correlation_plot(
        all_results,
        output_path=os.path.join(output_dir, '09_correlation.png'),
        figsize=(3.54, 3.54)
    )
    
    print("\n" + "="*80)
    print("✓ ALL PLOTS CREATED SUCCESSFULLY!")
    print("="*80)
    print(f"\n📁 Plots saved to: {output_dir}")
    print(f"\n📊 Generated {9} publication-ready plots")
    if include_gam:
        print("   • GAM results highlighted in red")
        print("   • GAM bars have red edges")
        print("   • GAM labels are bold")
    print("\n📐 All plots optimized for Nature journal submission")
    print("   • DPI: 600 (publication quality)")
    print("   • Sizes: Single (3.54\") / Double (7.08\") column")
    print("   • Style: Minimal, high-contrast, viridis colors")
    print("="*80)

if __name__ == "__main__":
    # =========================================================================
    # CONFIGURATION
    # =========================================================================
    
    base_results_dir = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis'
    output_dir = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/nature_plots'
    
    # Model type: 'custom', 'with_categorical', or 'all'
    model_type = 'with_categorical'
    
    # =========================================================================
    # GAM RESULTS - UPDATE THESE VALUES WITH YOUR ACTUAL RESULTS
    # =========================================================================
    
    GAM_RESULTS = {
        'tc_per_ha_actual': {
            'Overall_R2': 0.856,        # ← Update with your actual R² value
            'Overall_RMSE': 12.34,      # ← Update with your actual RMSE value
            'Overall_MAE': 9.87,        # Optional
            'Overall_RPD': 2.54,        # Optional
            'Overall_RPIQ': 3.21        # Optional
        },
        'brix_percent': {
            'Overall_R2': 0.782,        # ← Update with your actual R² value
            'Overall_RMSE': 0.98,       # ← Update with your actual RMSE value
            'Overall_MAE': 0.76,        # Optional
            'Overall_RPD': 2.13,        # Optional
            'Overall_RPIQ': 2.87        # Optional
        },
        'glucose_cane_percent': {
            'Overall_R2': 0.723,        # ← Update with your actual values
            'Overall_RMSE': 0.45,
            'Overall_MAE': 0.34,
            'Overall_RPD': 1.89,
            'Overall_RPIQ': 2.45
        },
        'purity_percent': {
            'Overall_R2': 0.698,        # ← Update with your actual values
            'Overall_RMSE': 1.23,
            'Overall_MAE': 0.98,
            'Overall_RPD': 1.76,
            'Overall_RPIQ': 2.12
        }
        # Add more response variables as needed
    }
    
    # =========================================================================
    # FILTERS (set to None to include all)
    # =========================================================================
    
    # Select specific models
    models_to_include = ['RandomForest', 'XGBoost', 'LinearRegression']
    # models_to_include = None  # For all models
    
    # Select specific response variables
    response_variables = ['brix_percent', 'tc_per_ha_actual']
    # response_variables = None  # For all responses
    
    # Use abbreviated model names (RF, GB, XGB, GAM, etc.) for compact plots
    use_abbrev = True
    
    # Include GAM results
    include_gam = False
    
    # =========================================================================
    # RUN
    # =========================================================================
    
    main(base_results_dir, output_dir, model_type, 
         models_to_include=models_to_include,
         response_variables=response_variables,
         use_abbrev=use_abbrev,
         include_gam=include_gam,
         gam_results_dict=GAM_RESULTS)
```

# Métriques en %

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
from pathlib import Path
import warnings
warnings.filterwarnings('ignore')

# ============================================================================
# STYLE CONFIGURATION
# ============================================================================

plt.rcParams['font.family'] = 'Arial'
plt.rcParams['font.size'] = 7
plt.rcParams['axes.linewidth'] = 0.6
plt.rcParams['xtick.major.width'] = 0.6
plt.rcParams['ytick.major.width'] = 0.6
plt.rcParams['xtick.major.size'] = 2.5
plt.rcParams['ytick.major.size'] = 2.5
plt.rcParams['xtick.minor.size'] = 1.5
plt.rcParams['ytick.minor.size'] = 1.5
plt.rcParams['figure.dpi'] = 600
plt.rcParams['savefig.dpi'] = 600
plt.rcParams['savefig.bbox'] = 'tight'
plt.rcParams['savefig.pad_inches'] = 0.02
plt.rcParams['axes.labelpad'] = 2
plt.rcParams['axes.titlepad'] = 4

RESPONSE_LABELS = {
    "brix_percent": "Brix (%)",
    "pol_juice_percent": "Pol juice (%)",
    "pol_cane_percent": "Pol cane (%)",
    "purity_percent": "Purity (%)",
    "glucose_cane_percent": "Glucose (%)",
    "hugot_sugar": "Sucrose (%)",
    "tc_per_ha_actual": "Cane yield (t/ha)",
    "tc_per_ha_per_week": "Yield/week (t/ha/wk)",
    "tc_per_ha_52_weeks": "Yield 52-wk (t/ha)",
    "fiber_percent": "Fiber (%)",
    "total_recoverable_sugar_t_per_ha": "TPER (t/ha)"
}

MODEL_ABBREV = {
    'LinearRegression': 'LR',
    'KNN': 'KNN',
    'DecisionTree': 'DT',
    'RandomForest': 'RF',
    'ExtraTrees': 'ET',
    'GradientBoosting': 'GB',
    'BaggingRegressor': 'Bag',
    'XGBoost': 'XGB',
    'LightGBM': 'LGBM',
    'CatBoost': 'CB',
    'GAM': 'GAM'
}

MODEL_COLORS = {
    'LinearRegression': '#440154',
    'KNN': '#31688e',
    'DecisionTree': '#35b779',
    'RandomForest': '#fde724',
    'ExtraTrees': '#21918c',
    'GradientBoosting': '#443983',
    'BaggingRegressor': '#90d743',
    'XGBoost': '#feb72d',
    'LightGBM': '#1f9e89',
    'CatBoost': '#cc4778',
    'GAM': '#e63946'
}

# ============================================================================
# GAM RESULTS CONFIGURATION
# ============================================================================

GAM_RESULTS = {
    'tc_per_ha_actual': {
        'Overall_R2': 0.856,
        'Overall_RMSE': 12.34,
        'Overall_MAE': 9.87,
        'Overall_RPD': 2.54,
        'Overall_RPIQ': 3.21,
        'Mean_CV_R2': 0.845,
        'Std_CV_R2': 0.023
    },
    'brix_percent': {
        'Overall_R2': 0.782,
        'Overall_RMSE': 0.98,
        'Overall_MAE': 0.76,
        'Overall_RPD': 2.13,
        'Overall_RPIQ': 2.87
    },
    'glucose_cane_percent': {
        'Overall_R2': 0.723,
        'Overall_RMSE': 0.45,
        'Overall_MAE': 0.34,
        'Overall_RPD': 1.89,
        'Overall_RPIQ': 2.45
    },
    'purity_percent': {
        'Overall_R2': 0.698,
        'Overall_RMSE': 1.23,
        'Overall_MAE': 0.98,
        'Overall_RPD': 1.76,
        'Overall_RPIQ': 2.12
    }
}

# ============================================================================
# DATA LOADING FUNCTIONS
# ============================================================================

def find_result_folders(base_dir, model_type='all', response_variables=None):
    """Find all result folders containing kfold_cv_summary files"""
    result_folders = {}
    base_path = Path(base_dir)
    
    if not base_path.exists():
        print(f"Warning: Base directory not found: {base_dir}")
        return result_folders
    
    for folder in base_path.glob('kfold_models_*'):
        if folder.is_dir():
            folder_name = folder.name
            
            if model_type == 'custom' and '_custom' not in folder_name:
                continue
            elif model_type == 'with_categorical' and '_with_categorical' not in folder_name:
                continue
            
            response_var = folder_name.replace('kfold_models_', '')
            
            if '_custom' in response_var:
                response_var = response_var.replace('_custom', '')
                suffix = '_custom'
            elif '_with_categorical' in response_var:
                response_var = response_var.replace('_with_categorical', '')
                suffix = '_with_categorical'
            else:
                suffix = ''
            
            if response_variables is not None and response_var not in response_variables:
                continue
            
            summary_files = list(folder.glob('kfold_cv_summary_*.csv'))
            if summary_files:
                result_folders[response_var] = {
                    'folder': str(folder),
                    'summary_file': str(summary_files[0]),
                    'type': suffix.replace('_', '') if suffix else 'standard'
                }
    
    return result_folders

def add_gam_results_to_dataframe(df, response_var, gam_results_dict):
    """Add GAM results as a new row in the dataframe"""
    if response_var not in gam_results_dict:
        return df
    
    gam_metrics = gam_results_dict[response_var]
    gam_row = {'Model': 'GAM', 'Complexity': 'Complex'}
    gam_row.update(gam_metrics)
    
    gam_df = pd.DataFrame([gam_row])
    df = pd.concat([df, gam_df], ignore_index=True)
    
    return df

def load_all_results(base_dir, model_type='all', models_to_include=None, 
                    response_variables=None, include_gam=True, gam_results_dict=None):
    """Load all model results from different response variables, optionally including GAM"""
    result_folders = find_result_folders(base_dir, model_type, response_variables)
    
    if not result_folders and not include_gam:
        return None
    
    all_results = {}
    
    for response_var, paths in result_folders.items():
        try:
            df = pd.read_csv(paths['summary_file'])
            
            if models_to_include is not None:
                df = df[df['Model'].isin(models_to_include)].copy()
            
            if include_gam and gam_results_dict is not None:
                df = add_gam_results_to_dataframe(df, response_var, gam_results_dict)
            
            if len(df) > 0:
                all_results[response_var] = df
                
        except Exception as e:
            print(f"Error loading {response_var}: {e}")
    
    if include_gam and gam_results_dict is not None and not result_folders:
        for response_var, gam_metrics in gam_results_dict.items():
            if response_variables is None or response_var in response_variables:
                gam_row = {'Model': 'GAM', 'Complexity': 'Complex'}
                gam_row.update(gam_metrics)
                all_results[response_var] = pd.DataFrame([gam_row])
    
    return all_results

# ============================================================================
# COMPACT PLOTTING FUNCTIONS WITH 3x2 GRID
# ============================================================================

def format_value_with_unit(value, metric):
    """Format value based on metric type - only R² gets percentage"""
    if 'R2' in metric or 'r2' in metric.lower():
        # R² as percentage
        return f"{int(round(value * 100))}"
    else:
        # Other metrics keep their original units
        if isinstance(value, (int, np.integer)):
            return f"{value}"
        elif value >= 100:
            return f"{int(round(value))}"
        elif value >= 10:
            return f"{value:.1f}"
        else:
            return f"{value:.2f}"

def get_metric_label(metric):
    """Get appropriate label for metric - only R² gets %"""
    if 'R2' in metric or 'r2' in metric.lower():
        return 'R² (%)'
    elif 'RMSE' in metric:
        return 'RMSE'
    elif 'MAE' in metric:
        return 'MAE'
    elif 'RPD' in metric:
        return 'RPD'
    elif 'RPIQ' in metric:
        return 'RPIQ'
    else:
        return metric

def create_metric_comparison_grid_3x2(all_results, metric='Overall_R2', 
                                       output_path='metric_comparison_3x2.png',
                                       figsize_per_plot=(2.0, 1.5),
                                       use_abbrev=True,
                                       highlight_gam=True,
                                       response_order=None):
    """Create 3x2 grid of bar plots with Nature styling, following response variable order"""
    
    # Order results according to response_order
    if response_order is not None:
        ordered_results = {}
        for resp_var in response_order:
            if resp_var in all_results:
                ordered_results[resp_var] = all_results[resp_var]
    else:
        ordered_results = all_results
    
    n_responses = len(ordered_results)
    
    if n_responses == 0:
        print(f"No data to plot for {metric}")
        return
    
    # Fixed 3x2 grid
    n_cols = 3
    n_rows = 2
    
    fig_width = figsize_per_plot[0] * n_cols
    fig_height = figsize_per_plot[1] * n_rows
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(fig_width, fig_height))
    axes = axes.flatten()
    
    metric_label = get_metric_label(metric)
    is_r2_metric = 'R2' in metric or 'r2' in metric.lower()
    
    for idx, (response_var, df) in enumerate(ordered_results.items()):
        if idx >= 6:  # Only show first 6 plots in 3x2 grid
            break
            
        ax = axes[idx]
        
        if metric not in df.columns:
            ax.axis('off')
            continue
        
        df_sorted = df.sort_values(metric, ascending=True)
        
        # Format values based on metric type
        if is_r2_metric:
            values = df_sorted[metric].values * 100  # Convert to percentage
            display_values = [int(round(v)) for v in values]
        else:
            values = df_sorted[metric].values
            display_values = values
        
        model_names = [MODEL_ABBREV.get(m, m) if use_abbrev else m for m in df_sorted['Model']]
        colors = [MODEL_COLORS.get(model, '#666666') for model in df_sorted['Model']]
        edgecolors = ['red' if model == 'GAM' else 'white' for model in df_sorted['Model']]
        linewidths = [1.2 if model == 'GAM' else 0.3 for model in df_sorted['Model']]
        
        bars = ax.barh(range(len(model_names)), values, 
                       color=colors, edgecolor=edgecolors, linewidth=linewidths, height=0.5)
        
        ax.set_yticks(range(len(model_names)))
        ax.set_yticklabels(model_names, fontsize=6.5, fontweight='normal')
        
        for i, (label, model) in enumerate(zip(ax.get_yticklabels(), df_sorted['Model'])):
            if model == 'GAM':
                label.set_fontweight('bold')
                label.set_color('#e63946')
        
        ax.set_xlabel(metric_label, fontsize=7, fontweight='bold')
        
        title = RESPONSE_LABELS.get(response_var, response_var)
        ax.set_title(title, fontsize=7.5, fontweight='bold', pad=3)
        
        ax.tick_params(axis='x', labelsize=6)
        ax.tick_params(axis='y', length=0)
        
        ax.grid(axis='x', alpha=0.2, linewidth=0.3, linestyle='--')
        ax.set_axisbelow(True)
        
        for spine in ['top', 'right', 'left']:
            ax.spines[spine].set_visible(False)
        ax.spines['bottom'].set_linewidth(0.6)
        
        # Add value labels
        for i, (bar, val, disp_val, model) in enumerate(zip(bars, values, display_values, df_sorted['Model'])):
            if val > 0:
                fontweight = 'bold' if model == 'GAM' else 'bold'
                if is_r2_metric:
                    label_text = f'{int(disp_val)}'
                else:
                    label_text = format_value_with_unit(disp_val, metric)
                
                ax.text(val * 0.98, bar.get_y() + bar.get_height()/2, 
                       label_text, 
                       ha='right', va='center', fontsize=5.5, 
                       fontweight=fontweight, color='white')
    
    # Hide unused subplots
    for idx in range(n_responses, 6):
        axes[idx].axis('off')
    
    plt.subplots_adjust(hspace=0.5, wspace=0.4)
    plt.savefig(output_path, dpi=600, bbox_inches='tight', 
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

def create_comprehensive_comparison(all_results, metric='Overall_R2',
                                   output_path='comprehensive_comparison.png',
                                   figsize=(7.08, 3.5),
                                   use_abbrev=True,
                                   highlight_gam=True,
                                   response_order=None):
    """Compact grouped bar chart with Nature styling"""
    
    # Order according to response_order
    if response_order is not None:
        ordered_results = {k: all_results[k] for k in response_order if k in all_results}
    else:
        ordered_results = all_results
    
    plot_data = []
    metric_label = get_metric_label(metric)
    is_r2_metric = 'R2' in metric or 'r2' in metric.lower()
    
    for response_var, df in ordered_results.items():
        if metric not in df.columns:
            continue
        for _, row in df.iterrows():
            value = row[metric]
            if is_r2_metric:
                value = value * 100  # Convert to percentage for R²
            
            plot_data.append({
                'Response': RESPONSE_LABELS.get(response_var, response_var),
                'Model': row['Model'],
                'Value': value
            })
    
    if not plot_data:
        return
    
    plot_df = pd.DataFrame(plot_data)
    
    fig, ax = plt.subplots(figsize=figsize)
    
    models = sorted(plot_df['Model'].unique())
    # Maintain response order
    responses = [RESPONSE_LABELS.get(rv, rv) for rv in ordered_results.keys()]
    
    x = np.arange(len(responses))
    width = 0.6 / len(models)
    
    for i, model in enumerate(models):
        model_data = plot_df[plot_df['Model'] == model]
        values = [model_data[model_data['Response'] == resp]['Value'].values[0] 
                 if len(model_data[model_data['Response'] == resp]) > 0 else 0 
                 for resp in responses]
        
        offset = (i - len(models)/2) * width + width/2
        color = MODEL_COLORS.get(model, '#666666')
        label = MODEL_ABBREV.get(model, model) if use_abbrev else model
        
        if model == 'GAM':
            edgecolor = 'red'
            linewidth = 1.2
        else:
            edgecolor = 'white'
            linewidth = 0.3
        
        ax.bar(x + offset, values, width, label=label, 
              color=color, edgecolor=edgecolor, linewidth=linewidth)
    
    ax.set_ylabel(metric_label, fontsize=8, fontweight='bold')
    ax.set_title(f'Model performance comparison', 
                fontsize=9, fontweight='bold', pad=6)
    
    ax.set_xticks(x)
    ax.set_xticklabels(responses, rotation=45, ha='right', fontsize=7)
    ax.tick_params(axis='y', labelsize=7)
    ax.tick_params(axis='x', length=2)
    
    ax.grid(axis='y', alpha=0.2, linewidth=0.3, linestyle='--')
    ax.set_axisbelow(True)
    
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    for spine in ['left', 'bottom']:
        ax.spines[spine].set_linewidth(0.6)
    
    legend = ax.legend(loc='upper left', bbox_to_anchor=(1, 1), 
                      fontsize=6, frameon=True, fancybox=False, 
                      shadow=False, ncol=1, borderaxespad=0,
                      edgecolor='#CCCCCC', framealpha=0.95)
    
    for text in legend.get_texts():
        if text.get_text() == 'GAM':
            text.set_fontweight('bold')
            text.set_color('#e63946')
    
    plt.tight_layout()
    plt.savefig(output_path, dpi=600, bbox_inches='tight',
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

def create_best_model_summary(all_results, output_path='best_models_summary.png',
                              figsize=(7.08, 2.5), response_order=None):
    """Compact multi-metric comparison of best models"""
    
    # Order according to response_order
    if response_order is not None:
        ordered_results = {k: all_results[k] for k in response_order if k in all_results}
    else:
        ordered_results = all_results
    
    best_models = []
    
    for response_var, df in ordered_results.items():
        if 'Overall_R2' not in df.columns:
            continue
            
        best_idx = df['Overall_R2'].idxmax()
        best_row = df.loc[best_idx]
        
        best_models.append({
            'Response': RESPONSE_LABELS.get(response_var, response_var),
            'Model': MODEL_ABBREV.get(best_row['Model'], best_row['Model']),
            'R²': best_row['Overall_R2'] * 100,  # Convert to percentage
            'RMSE': best_row.get('Overall_RMSE', np.nan),
            'RPD': best_row.get('Overall_RPD', np.nan),
            'RPIQ': best_row.get('Overall_RPIQ', np.nan)
        })
    
    if not best_models:
        return
    
    best_df = pd.DataFrame(best_models)
    
    fig, axes = plt.subplots(1, 4, figsize=figsize)
    
    metrics = [
        ('R²', 'R² (%)', 'viridis', True),
        ('RMSE', 'RMSE', 'plasma', False),
        ('RPD', 'RPD', 'cividis', False),
        ('RPIQ', 'RPIQ', 'viridis', False)
    ]
    
    for ax, (metric, label, cmap, is_r2) in zip(axes, metrics):
        if metric in best_df.columns and not best_df[metric].isna().all():
            values = best_df[metric].values
            norm_values = (values - values.min()) / (values.max() - values.min() + 1e-10)
            colors = plt.cm.get_cmap(cmap)(norm_values)
            
            bars = ax.barh(range(len(best_df)), best_df[metric], 
                          color=colors, edgecolor='white', linewidth=0.4, height=0.5)
            
            ax.set_yticks(range(len(best_df)))
            if ax == axes[0]:
                ax.set_yticklabels(best_df['Response'], fontsize=6.5)
            else:
                ax.set_yticklabels([])
            
            ax.set_xlabel(label, fontsize=7.5, fontweight='bold')
            ax.set_title(f'{label}', fontsize=8, fontweight='bold', pad=3)
            ax.tick_params(axis='both', labelsize=6, length=2)
            ax.tick_params(axis='y', length=0)
            
            ax.grid(axis='x', alpha=0.2, linewidth=0.3, linestyle='--')
            ax.set_axisbelow(True)
            
            for spine in ['top', 'right', 'left']:
                ax.spines[spine].set_visible(False)
            ax.spines['bottom'].set_linewidth(0.6)
            
            # Add value labels with proper formatting
            for i, (bar, val) in enumerate(zip(bars, best_df[metric])):
                if not np.isnan(val):
                    if is_r2:
                        label_text = f'{int(round(val))}'
                    else:
                        label_text = format_value_with_unit(val, metric)
                    
                    ax.text(val * 0.95, bar.get_y() + bar.get_height()/2,
                           label_text,
                           ha='right', va='center', fontsize=5.5,
                           fontweight='bold', color='white')
    
    plt.subplots_adjust(wspace=0.25)
    plt.savefig(output_path, dpi=600, bbox_inches='tight',
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

def create_heatmap_comparison(all_results, metric='Overall_R2',
                              output_path='heatmap_comparison.png',
                              figsize=(7.08, 3.5),
                              use_abbrev=True,
                              response_order=None):
    """Compact heatmap with viridis colormap and white gridlines"""
    
    # Order according to response_order
    if response_order is not None:
        ordered_results = {k: all_results[k] for k in response_order if k in all_results}
    else:
        ordered_results = all_results
    
    all_models = set()
    for df in ordered_results.values():
        all_models.update(df['Model'].tolist())
    models_unique = sorted(list(all_models))
    
    # Maintain response order
    responses = [RESPONSE_LABELS.get(rv, rv) for rv in ordered_results.keys()]
    response_var_keys = list(ordered_results.keys())
    
    metric_label = get_metric_label(metric)
    is_r2_metric = 'R2' in metric or 'r2' in metric.lower()
    
    data_matrix = []
    for model in models_unique:
        row = []
        for response_var in response_var_keys:
            df = ordered_results[response_var]
            if metric not in df.columns:
                row.append(np.nan)
                continue
            
            model_data = df[df['Model'] == model]
            if len(model_data) > 0:
                value = model_data[metric].values[0]
                if is_r2_metric:
                    value = value * 100  # Convert to percentage
                row.append(value)
            else:
                row.append(np.nan)
        data_matrix.append(row)
    
    model_labels = [MODEL_ABBREV.get(m, m) if use_abbrev else m for m in models_unique]
    heatmap_df = pd.DataFrame(data_matrix, columns=responses, index=model_labels)
    
    fig, ax = plt.subplots(figsize=figsize)
    
    im = ax.imshow(heatmap_df.values, cmap='viridis', aspect='auto',
                   vmin=np.nanmin(heatmap_df.values), 
                   vmax=np.nanmax(heatmap_df.values))
    
    ax.set_xticks(np.arange(len(responses)))
    ax.set_yticks(np.arange(len(model_labels)))
    ax.set_xticklabels(responses, rotation=45, ha='right', fontsize=8)
    ax.set_yticklabels(model_labels, fontsize=8)
    
    for i, (label, model) in enumerate(zip(ax.get_yticklabels(), models_unique)):
        if model == 'GAM':
            label.set_fontweight('bold')
            label.set_color('#e63946')
    
    ax.set_xlabel('Response variable', fontsize=9, fontweight='bold')
    ax.set_ylabel('Model', fontsize=9, fontweight='bold')
    ax.set_title(f'Performance heatmap: {metric_label}', 
                fontsize=10, fontweight='bold', pad=6)
    
    ax.tick_params(axis='both', length=2)
    
    # White gridlines
    for i in range(len(model_labels) + 1):
        ax.axhline(i - 0.5, color='white', linewidth=1.5, zorder=10)
    for j in range(len(responses) + 1):
        ax.axvline(j - 0.5, color='white', linewidth=1.5, zorder=10)
    
    # Highlight GAM row
    gam_idx = None
    for i, model in enumerate(models_unique):
        if model == 'GAM':
            gam_idx = i
            break
    
    if gam_idx is not None:
        rect = plt.Rectangle((-0.5, gam_idx - 0.5), len(responses), 1,
                            fill=False, edgecolor='red', linewidth=2, zorder=11)
        ax.add_patch(rect)
    
    # Add value labels with proper formatting
    for i in range(len(model_labels)):
        for j in range(len(responses)):
            val = heatmap_df.iloc[i, j]
            if not np.isnan(val):
                luminance = (val - np.nanmin(heatmap_df.values)) / (np.nanmax(heatmap_df.values) - np.nanmin(heatmap_df.values))
                text_color = 'white' if luminance < 0.5 else 'black'
                fontweight = 'bold' if models_unique[i] == 'GAM' else 'bold'
                
                if is_r2_metric:
                    label_text = f'{int(round(val))}'
                else:
                    label_text = format_value_with_unit(val, metric)
                
                ax.text(j, i, label_text,
                       ha='center', va='center', 
                       color=text_color,
                       fontsize=7, fontweight=fontweight)
    
    cbar = plt.colorbar(im, ax=ax, fraction=0.035, pad=0.02)
    cbar.set_label(metric_label, rotation=270, labelpad=12, 
                  fontsize=8, fontweight='bold')
    cbar.ax.tick_params(labelsize=7)
    cbar.outline.set_linewidth(0.6)
    
    for spine in ax.spines.values():
        spine.set_linewidth(0.6)
    
    plt.tight_layout()
    plt.savefig(output_path, dpi=600, bbox_inches='tight',
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

def create_model_ranking_plot(all_results, metric='Overall_R2',
                              output_path='model_ranking.png',
                              figsize=(7.08, 3.5),
                              use_abbrev=True,
                              response_order=None):
    """Create a compact ranking visualization showing model performance across responses"""
    
    # Order according to response_order
    if response_order is not None:
        ordered_results = {k: all_results[k] for k in response_order if k in all_results}
    else:
        ordered_results = all_results
    
    model_ranks = {}
    
    for response_var, df in ordered_results.items():
        if metric not in df.columns:
            continue
        
        df_sorted = df.sort_values(metric, ascending=False)
        for rank, row in enumerate(df_sorted.itertuples(), start=1):
            model = row.Model
            if model not in model_ranks:
                model_ranks[model] = []
            model_ranks[model].append(rank)
    
    rank_stats = []
    for model, ranks in model_ranks.items():
        rank_stats.append({
            'Model': MODEL_ABBREV.get(model, model) if use_abbrev else model,
            'Mean_Rank': np.mean(ranks),
            'Median_Rank': np.median(ranks),
            'Best_Rank': np.min(ranks),
            'Worst_Rank': np.max(ranks),
            'Full_Model': model
        })
    
    rank_df = pd.DataFrame(rank_stats).sort_values('Mean_Rank')
    
    fig, ax = plt.subplots(figsize=figsize)
    
    y_pos = np.arange(len(rank_df))
    
    colors = [MODEL_COLORS.get(m, '#666666') for m in rank_df['Full_Model']]
    alphas = [0.9 if m == 'GAM' else 0.7 for m in rank_df['Full_Model']]
    edgecolors = ['red' if m == 'GAM' else 'white' for m in rank_df['Full_Model']]
    linewidths = [1.2 if m == 'GAM' else 0.4 for m in rank_df['Full_Model']]
    
    for i, (color, alpha, edge, lw) in enumerate(zip(colors, alphas, edgecolors, linewidths)):
        ax.barh([i], [rank_df.iloc[i]['Mean_Rank']], 
               color=color, alpha=alpha, edgecolor=edge, linewidth=lw, height=0.5)
    
    for i, row in rank_df.iterrows():
        ax.plot([row['Best_Rank'], row['Worst_Rank']], [i, i],
               color='black', linewidth=1.5, alpha=0.8, zorder=10)
        ax.plot([row['Best_Rank'], row['Best_Rank']], [i-0.15, i+0.15],
               color='black', linewidth=1.5, alpha=0.8, zorder=10)
        ax.plot([row['Worst_Rank'], row['Worst_Rank']], [i-0.15, i+0.15],
               color='black', linewidth=1.5, alpha=0.8, zorder=10)
    
    ax.set_yticks(y_pos)
    ax.set_yticklabels(rank_df['Model'], fontsize=7)
    
    for i, (label, model) in enumerate(zip(ax.get_yticklabels(), rank_df['Full_Model'])):
        if model == 'GAM':
            label.set_fontweight('bold')
            label.set_color('#e63946')
    
    ax.set_xlabel('Ranking (lower is better)', fontsize=8, fontweight='bold')
    ax.set_title('Model ranking across all response variables', 
                fontsize=9, fontweight='bold', pad=6)
    
    ax.tick_params(axis='both', labelsize=6.5, length=2)
    ax.tick_params(axis='y', length=0)
    
    ax.invert_xaxis()
    ax.grid(axis='x', alpha=0.2, linewidth=0.3, linestyle='--')
    ax.set_axisbelow(True)
    
    for spine in ['top', 'right', 'left']:
        ax.spines[spine].set_visible(False)
    ax.spines['bottom'].set_linewidth(0.6)
    
    from matplotlib.patches import Rectangle
    legend_elements = [
        Rectangle((0, 0), 1, 1, fc='gray', alpha=0.7, label='Mean rank'),
        plt.Line2D([0], [0], color='black', linewidth=1.5, label='Best-worst range')
    ]
    ax.legend(handles=legend_elements, loc='lower right', fontsize=6, 
             frameon=True, edgecolor='#CCCCCC', framealpha=0.95)
    
    plt.tight_layout()
    plt.savefig(output_path, dpi=600, bbox_inches='tight',
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

def create_metric_correlation_plot(all_results, 
                                   output_path='metric_correlation.png',
                                   figsize=(3.54, 3.54)):
    """Compact correlation plot between different metrics"""
    metrics_data = []
    
    for response_var, df in all_results.items():
        for _, row in df.iterrows():
            metrics_data.append({
                'Model': row['Model'],
                'Response': response_var,
                'R²': row.get('Overall_R2', np.nan) * 100,  # Convert to percentage
                'RMSE': row.get('Overall_RMSE', np.nan),
                'RPD': row.get('Overall_RPD', np.nan),
                'RPIQ': row.get('Overall_RPIQ', np.nan)
            })
    
    if not metrics_data:
        return
    
    metrics_df = pd.DataFrame(metrics_data).dropna()
    
    fig, ax = plt.subplots(figsize=figsize)
    
    models = metrics_df['Model'].unique()
    
    for model in models:
        model_data = metrics_df[metrics_df['Model'] == model]
        color = MODEL_COLORS.get(model, '#666666')
        
        if model == 'GAM':
            marker = 's'
            s = 50
            edgecolor = 'red'
            linewidth = 1.5
            alpha = 0.9
        else:
            marker = 'o'
            s = 30
            edgecolor = 'white'
            linewidth = 0.5
            alpha = 0.7
        
        ax.scatter(model_data['R²'], model_data['RPD'], 
                  c=[color], alpha=alpha, s=s, marker=marker,
                  edgecolors=edgecolor, linewidth=linewidth,
                  label=MODEL_ABBREV.get(model, model))
    
    ax.set_xlabel('R² (%)', fontsize=8, fontweight='bold')
    ax.set_ylabel('RPD', fontsize=8, fontweight='bold')
    ax.set_title('Metric correlation: R² vs RPD', 
                fontsize=9, fontweight='bold', pad=6)
    
    ax.tick_params(axis='both', labelsize=7, length=2)
    ax.grid(alpha=0.2, linewidth=0.3, linestyle='--')
    ax.set_axisbelow(True)
    
    for spine in ax.spines.values():
        spine.set_linewidth(0.6)
    
    legend = ax.legend(loc='best', fontsize=5.5, frameon=True,
                      edgecolor='#CCCCCC', framealpha=0.95, ncol=2)
    
    for text in legend.get_texts():
        if text.get_text() == 'GAM':
            text.set_fontweight('bold')
            text.set_color('#e63946')
    
    plt.tight_layout()
    plt.savefig(output_path, dpi=600, bbox_inches='tight',
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

def create_performance_distribution(all_results, metric='Overall_R2',
                                   output_path='performance_distribution.png',
                                   figsize=(7.08, 2.5),
                                   use_abbrev=True):
    """Compact violin/box plot showing distribution of performance"""
    plot_data = []
    metric_label = get_metric_label(metric)
    is_r2_metric = 'R2' in metric or 'r2' in metric.lower()
    
    for response_var, df in all_results.items():
        if metric not in df.columns:
            continue
        for _, row in df.iterrows():
            value = row[metric]
            if is_r2_metric:
                value = value * 100  # Convert to percentage
            
            plot_data.append({
                'Model': row['Model'],
                'Value': value
            })
    
    if not plot_data:
        return
    
    plot_df = pd.DataFrame(plot_data)
    
    fig, ax = plt.subplots(figsize=figsize)
    
    models = sorted(plot_df['Model'].unique(), 
                   key=lambda x: plot_df[plot_df['Model']==x]['Value'].median(),
                   reverse=True)
    
    positions = np.arange(len(models))
    
    parts = ax.violinplot([plot_df[plot_df['Model']==m]['Value'].values for m in models],
                          positions=positions, widths=0.6,
                          showmeans=False, showmedians=True, showextrema=True)
    
    for i, (pc, model) in enumerate(zip(parts['bodies'], models)):
        pc.set_facecolor(MODEL_COLORS.get(model, '#666666'))
        
        if model == 'GAM':
            pc.set_alpha(0.9)
            pc.set_edgecolor('red')
            pc.set_linewidth(1.5)
        else:
            pc.set_alpha(0.7)
            pc.set_edgecolor('white')
            pc.set_linewidth(0.5)
    
    for partname in ['cmedians', 'cmins', 'cmaxes', 'cbars']:
        if partname in parts:
            parts[partname].set_edgecolor('black')
            parts[partname].set_linewidth(1)
    
    model_labels = [MODEL_ABBREV.get(m, m) if use_abbrev else m for m in models]
    ax.set_xticks(positions)
    ax.set_xticklabels(model_labels, rotation=45, ha='right', fontsize=7)
    
    for i, (label, model) in enumerate(zip(ax.get_xticklabels(), models)):
        if model == 'GAM':
            label.set_fontweight('bold')
            label.set_color('#e63946')
    
    ax.set_ylabel(metric_label, fontsize=8, fontweight='bold')
    ax.set_title(f'Performance distribution: {metric_label}', 
                fontsize=9, fontweight='bold', pad=6)
    
    ax.tick_params(axis='both', labelsize=7, length=2)
    ax.grid(axis='y', alpha=0.2, linewidth=0.3, linestyle='--')
    ax.set_axisbelow(True)
    
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    for spine in ['left', 'bottom']:
        ax.spines[spine].set_linewidth(0.6)
    
    plt.tight_layout()
    plt.savefig(output_path, dpi=600, bbox_inches='tight',
               facecolor='white', edgecolor='none')
    print(f"✓ Saved: {output_path}")
    plt.close()

# ============================================================================
# MAIN EXECUTION
# ============================================================================

def main(base_results_dir, output_dir='plots', model_type='all',
         models_to_include=None, response_variables=None, use_abbrev=True,
         include_gam=True, gam_results_dict=None):
    """Main function to generate all compact Nature-style plots with 3x2 grid"""
    print("="*80)
    print("CREATING NATURE-STYLE COMPACT MODEL COMPARISON PLOTS (3x2 GRID)")
    if include_gam:
        print("Including GAM results")
    print("="*80)
    
    if models_to_include:
        print(f"\n📊 Models: {', '.join(models_to_include)}")
    else:
        print(f"\n📊 Models: All available")
    
    if include_gam:
        print(f"📊 + GAM (Generalized Additive Models)")
    
    if response_variables:
        print(f"📈 Responses: {', '.join(response_variables)}")
    else:
        print(f"📈 Responses: All available")
    
    if model_type != 'all':
        output_dir = f"{output_dir}_{model_type}"
    
    if include_gam:
        output_dir = f"{output_dir}_with_gam"
    
    os.makedirs(output_dir, exist_ok=True)
    print(f"\n💾 Output: {output_dir}")
    
    print(f"\n🔍 Loading results from: {base_results_dir}")
    all_results = load_all_results(base_results_dir, model_type, 
                                   models_to_include, response_variables,
                                   include_gam=include_gam,
                                   gam_results_dict=gam_results_dict)
    
    if not all_results:
        print(f"\n❌ Error: No results found!")
        return
    
    print(f"\n✓ Loaded {len(all_results)} response variables")
    for var, df in all_results.items():
        print(f"  • {var}: {len(df)} models")
        if include_gam and 'GAM' in df['Model'].values:
            print(f"    → GAM included")
    
    print("\n" + "="*80)
    print("GENERATING PLOTS")
    print("="*80)
    
    # 1. R² comparison grid (3x2)
    print("\n1/9 R² comparison grid (3x2)...")
    create_metric_comparison_grid_3x2(
        all_results, 
        metric='Overall_R2',
        output_path=os.path.join(output_dir, '01_r2_grid_3x2.png'),
        figsize_per_plot=(2.0, 1.5),
        use_abbrev=use_abbrev,
        highlight_gam=include_gam,
        response_order=response_variables
    )
    
    # 2. RMSE comparison grid (3x2)
    print("2/9 RMSE comparison grid (3x2)...")
    create_metric_comparison_grid_3x2(
        all_results,
        metric='Overall_RMSE',
        output_path=os.path.join(output_dir, '02_rmse_grid_3x2.png'),
        figsize_per_plot=(2.0, 1.5),
        use_abbrev=use_abbrev,
        highlight_gam=include_gam,
        response_order=response_variables
    )
    
    # 3. RPD comparison grid (3x2)
    print("3/9 RPD comparison grid (3x2)...")
    create_metric_comparison_grid_3x2(
        all_results,
        metric='Overall_RPD',
        output_path=os.path.join(output_dir, '03_rpd_grid_3x2.png'),
        figsize_per_plot=(2.0, 1.5),
        use_abbrev=use_abbrev,
        highlight_gam=include_gam,
        response_order=response_variables
    )
    
    # 4. Comprehensive comparison
    print("4/9 Comprehensive comparison...")
    create_comprehensive_comparison(
        all_results,
        metric='Overall_R2',
        output_path=os.path.join(output_dir, '04_comprehensive.png'),
        figsize=(7.08, 3.5),
        use_abbrev=use_abbrev,
        highlight_gam=include_gam,
        response_order=response_variables
    )
    
    # 5. Best models summary
    print("5/9 Best models summary...")
    create_best_model_summary(
        all_results,
        output_path=os.path.join(output_dir, '05_best_models.png'),
        figsize=(7.08, 2.5),
        response_order=response_variables
    )
    
    # 6. Heatmap comparison
    print("6/9 Heatmap comparison...")
    create_heatmap_comparison(
        all_results,
        metric='Overall_R2',
        output_path=os.path.join(output_dir, '06_heatmap.png'),
        figsize=(7.08, 3.5),
        use_abbrev=use_abbrev,
        response_order=response_variables
    )
    
    # 7. Model ranking
    print("7/9 Model ranking...")
    create_model_ranking_plot(
        all_results,
        metric='Overall_R2',
        output_path=os.path.join(output_dir, '07_ranking.png'),
        figsize=(7.08, 3.5),
        use_abbrev=use_abbrev,
        response_order=response_variables
    )
    
    # 8. Performance distribution
    print("8/9 Performance distribution...")
    create_performance_distribution(
        all_results,
        metric='Overall_R2',
        output_path=os.path.join(output_dir, '08_distribution.png'),
        figsize=(7.08, 2.5),
        use_abbrev=use_abbrev
    )
    
    # 9. Metric correlation
    print("9/9 Metric correlation...")
    create_metric_correlation_plot(
        all_results,
        output_path=os.path.join(output_dir, '09_correlation.png'),
        figsize=(3.54, 3.54)
    )
    
    print("\n" + "="*80)
    print("✓ ALL PLOTS CREATED SUCCESSFULLY!")
    print("="*80)
    print(f"\n📁 Plots saved to: {output_dir}")
    print(f"\n📊 Generated 9 publication-ready plots")
    print("   • Grid plots: 3 columns × 2 rows layout")
    print("   • Response order: follows input list order")
    if include_gam:
        print("   • GAM results highlighted in red")
        print("   • GAM bars have red edges")
        print("   • GAM labels are bold")
    print("\n📐 All plots optimized for Nature journal submission")
    print("   • DPI: 600 (publication quality)")
    print("   • Sizes: Single (3.54\") / Double (7.08\") column")
    print("   • Style: Minimal, high-contrast, viridis colors")
    print("   • R² displayed as percentage (%)")
    print("   • Other metrics use original units")
    print("="*80)

if __name__ == "__main__":
    # Configuration
    base_results_dir = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis'
    output_dir = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/nature_plots_3x2'
    
    model_type = 'with_categorical'
    
    GAM_RESULTS = {
        'tc_per_ha_actual': {
            'Overall_R2': 0.856,
            'Overall_RMSE': 12.34,
            'Overall_MAE': 9.87,
            'Overall_RPD': 2.54,
            'Overall_RPIQ': 3.21,
            'Mean_CV_R2': 0.845,
            'Std_CV_R2': 0.023
        },
        'brix_percent': {
            'Overall_R2': 0.782,
            'Overall_RMSE': 0.98,
            'Overall_MAE': 0.76,
            'Overall_RPD': 2.13,
            'Overall_RPIQ': 2.87
        },
        'glucose_cane_percent': {
            'Overall_R2': 0.723,
            'Overall_RMSE': 0.45,
            'Overall_MAE': 0.34,
            'Overall_RPD': 1.89,
            'Overall_RPIQ': 2.45
        },
        'purity_percent': {
            'Overall_R2': 0.698,
            'Overall_RMSE': 1.23,
            'Overall_MAE': 0.98,
            'Overall_RPD': 1.76,
            'Overall_RPIQ': 2.12
        }
    }
    
    models_to_include = ['RandomForest', 'XGBoost', 'LinearRegression']
    
    # Response variables in desired order (will be displayed in 3x2 grid following this order)
    response_variables = [
        'tc_per_ha_actual',
        'brix_percent', 
        'hugot_sugar',
        'purity_percent', 
        'total_recoverable_sugar_t_per_ha'
    ]
    
    use_abbrev = True
    include_gam = False
    
    main(base_results_dir, output_dir, model_type, 
         models_to_include=models_to_include,
         response_variables=response_variables,
         use_abbrev=use_abbrev,
         include_gam=include_gam,
         gam_results_dict=GAM_RESULTS)
```

```python
"""
SUGARCANE KEY VARIABLES TEMPORAL TRENDS 
===============================================================

"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
import warnings
import os
warnings.filterwarnings('ignore')

# Enable LaTeX-like rendering
plt.rcParams['text.usetex'] = False
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['mathtext.default'] = 'regular'

# ============================================================================
# CONFIGURATION
# ============================================================================

DATA_FILE = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/final_dataset_regrouped_categorical.csv'

OUTPUT_DIR_BASE_EN = 'sugarcane_trends_nature_english'
OUTPUT_DIR_BASE_FR = 'sugarcane_trends_nature_french'
OUTPUT_DPI = 600

# Language-specific labels
LABELS = {
    'en': {
        'year': 'Year',
        'main_title': '',
        'subplot_title': ''
    },
    'fr': {
        'year': 'Année',
        'main_title': '',
        'subplot_title': ''
    }
}

# BILINGUAL Response variables with proper LaTeX formatting
RESPONSE_VARIABLES = {
    'en': {
        'tc_per_ha_actual': {'name': r'Cane Yield (t $ha^{-1}$)', 'unit': r't $ha^{-1}$ $year^{-1}$'},
        'brix_percent': {'name': 'Brix (%)', 'unit': r'% $year^{-1}$'},
        'hugot_sugar': {'name': 'Sucrose (%)', 'unit': r'% $year^{-1}$'},
        'purity_percent': {'name': 'Purity (%)', 'unit': r'% $year^{-1}$'},
        'estimated_recoverable_pol': {'name': 'PER (%)', 'unit': r'% $year^{-1}$'},
        'total_recoverable_sugar_t_per_ha': {'name': r'TPER (t $ha^{-1}$)', 'unit': r't $ha^{-1}$ $year^{-1}$'}
    },
    'fr': {
        'tc_per_ha_actual': {'name': r'Rendement Canne (t $ha^{-1}$)', 'unit': r't $ha^{-1}$ $an^{-1}$'},
        'brix_percent': {'name': 'Brix (%)', 'unit': r'% $an^{-1}$'},
        'hugot_sugar': {'name': 'Saccharose (%)', 'unit': r'% $an^{-1}$'},
        'purity_percent': {'name': 'Pureté (%)', 'unit': r'% $an^{-1}$'},
        'estimated_recoverable_pol': {'name': 'PER (%)', 'unit': r'% $an^{-1}$'},
        'total_recoverable_sugar_t_per_ha': {'name': r'TPER (t $ha^{-1}$)', 'unit': r't $ha^{-1}$ $an^{-1}$'}
    }
}

COLORS = {
    'GAG': {
        'data_points': '#2E86AB',
        'trend_line': '#1E5F8C',
        'error_bars': '#2E86AB',
    },
    'GRV': {
        'data_points': '#E63946',
        'trend_line': '#C1121F',
        'error_bars': '#E63946',
    }
}
NATURE_SETTINGS = {
    'subplot_width': 4.5,
    'subplot_height': 3.2,
    'title_fontsize': 16,
    'axis_label_fontsize': 15,
    'tick_fontsize': 14,
    'stats_fontsize': 10,
    'legend_fontsize': 14,
    'suptitle_fontsize': 18,
    'marker_size': 6,
    'line_width': 3.0,
    'error_bar_width': 2.0,
    'cap_size': 4,
    'h_pad': 1.0,
    'w_pad': 2.0,
}

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def calculate_temporal_statistics(df, response_var, year_column, irrigation_type):
    df_filtered = df[df['irrigation_type'] == irrigation_type].copy()
    df_clean = df_filtered[[year_column, response_var]].dropna()
    
    if len(df_clean) < 3:
        return None
    
    yearly_stats = df_clean.groupby(year_column)[response_var].agg([
        ('mean', 'mean'),
        ('std', 'std'),
        ('count', 'count')
    ]).reset_index()
    
    years = yearly_stats[year_column].values
    means = yearly_stats['mean'].values
    stds = yearly_stats['std'].values
    
    if len(years) >= 3:
        slope, intercept, r_value, p_value, std_err = stats.linregress(years, means)
        
        first_value = means[0]
        last_value = means[-1]
        
        if abs(first_value) > 1e-10:
            percent_change = ((last_value - first_value) / abs(first_value)) * 100
        else:
            percent_change = np.nan
        
        absolute_change = last_value - first_value
        
        trend_stats = {
            'slope': slope,
            'intercept': intercept,
            'r_squared': r_value**2,
            'p_value': p_value,
            'significant': p_value < 0.05,
            'percent_change': percent_change,
            'absolute_change': absolute_change,
            'years_span': years[-1] - years[0],
            'first_value': first_value,
            'last_value': last_value
        }
    else:
        trend_stats = None
    
    return {
        'years': years,
        'means': means,
        'stds': stds,
        'yearly_stats': yearly_stats,
        'trend_stats': trend_stats,
        'irrigation_type': irrigation_type
    }

def format_pvalue(p):
    if p < 0.001:
        return 'p<0.001'
    elif p < 0.01:
        return 'p<0.01'
    elif p < 0.05:
        return f'p={p:.3f}'
    else:
        return f'p={p:.2f}'

def get_year_tick_labels(years, max_labels=4):
    unique_years = sorted(set(years))
    n_years = len(unique_years)
    
    if n_years <= max_labels:
        return unique_years, unique_years
    
    # Always include first and last
    selected_years = [unique_years[0]]
    
    # Calculate how many intermediate years to show
    n_intermediate = max_labels - 2
    
    if n_intermediate > 0:
        # Calculate step to evenly space intermediate years
        step = (n_years - 1) / (n_intermediate + 1)
        
        for i in range(1, n_intermediate + 1):
            idx = int(round(i * step))
            if idx > 0 and idx < n_years - 1:
                selected_years.append(unique_years[idx])
    
    # Always add last year
    selected_years.append(unique_years[-1])
    
    # Remove duplicates and sort
    selected_years = sorted(list(set(selected_years)))
    
    return unique_years, selected_years

def plot_dual_trends_nature(ax, stats_gag, stats_grv, var_name, var_info, colors, 
                            settings, language='en', show_legend=False):
    display_name = var_info['name']
    year_label = LABELS[language]['year']
    
    if stats_gag is not None:
        years_gag = stats_gag['years']
        means_gag = stats_gag['means']
        stds_gag = stats_gag['stds']
        
        ax.errorbar(years_gag, means_gag, yerr=stds_gag, 
                    fmt='o', markersize=settings['marker_size'], 
                    color=colors['GAG']['data_points'],
                    ecolor=colors['GAG']['error_bars'],
                    elinewidth=settings['error_bar_width'],
                    capsize=settings['cap_size'],
                    capthick=settings['error_bar_width'],
                    alpha=0.85,
                    zorder=3,
                    label='GAG')
        
        ax.plot(years_gag, means_gag, '-', 
                color=colors['GAG']['data_points'], 
                alpha=0.25, 
                linewidth=settings['line_width'] * 0.5,
                zorder=2)
        
        if stats_gag['trend_stats'] is not None:
            trend_line_gag = stats_gag['trend_stats']['slope'] * years_gag + stats_gag['trend_stats']['intercept']
            ax.plot(years_gag, trend_line_gag, '--', 
                    color=colors['GAG']['trend_line'],
                    linewidth=settings['line_width'],
                    alpha=0.9,
                    zorder=4)
    
    if stats_grv is not None:
        years_grv = stats_grv['years']
        means_grv = stats_grv['means']
        stds_grv = stats_grv['stds']
        
        ax.errorbar(years_grv, means_grv, yerr=stds_grv, 
                    fmt='s', markersize=settings['marker_size'], 
                    color=colors['GRV']['data_points'],
                    ecolor=colors['GRV']['error_bars'],
                    elinewidth=settings['error_bar_width'],
                    capsize=settings['cap_size'],
                    capthick=settings['error_bar_width'],
                    alpha=0.85,
                    zorder=3,
                    label='GRV')
        
        ax.plot(years_grv, means_grv, '-', 
                color=colors['GRV']['data_points'], 
                alpha=0.25, 
                linewidth=settings['line_width'] * 0.5,
                zorder=2)
        
        if stats_grv['trend_stats'] is not None:
            trend_line_grv = stats_grv['trend_stats']['slope'] * years_grv + stats_grv['trend_stats']['intercept']
            ax.plot(years_grv, trend_line_grv, '--', 
                    color=colors['GRV']['trend_line'],
                    linewidth=settings['line_width'],
                    alpha=0.9,
                    zorder=4)
    
    if stats_gag is not None and stats_gag['trend_stats'] is not None:
        ts = stats_gag['trend_stats']
        r2_str = f"R²={ts['r_squared']:.2f}"
        p_val_str = format_pvalue(ts['p_value'])
        pct_change = ts['percent_change']
        
        if not np.isnan(pct_change):
            pct_sign = '+' if pct_change > 0 else ''
            pct_str = f"Δ{pct_sign}{pct_change:.1f}%"
        else:
            pct_str = "Δ=N/A"
        
        stats_text_gag = f"GAG\n{r2_str}\n{p_val_str}\n{pct_str}"
        
        bbox_props_gag = dict(
            boxstyle='round,pad=0.4',
            facecolor='white',
            edgecolor=colors['GAG']['trend_line'],
            linewidth=2.5,
            alpha=0.95
        )
        
        ax.text(0.04, 0.96, stats_text_gag,
                transform=ax.transAxes,
                fontsize=settings['stats_fontsize'],
                verticalalignment='top',
                horizontalalignment='left',
                bbox=bbox_props_gag,
                fontweight='bold',
                family='monospace',
                linespacing=1.2,
                zorder=100)
    
    if stats_grv is not None and stats_grv['trend_stats'] is not None:
        ts = stats_grv['trend_stats']
        r2_str = f"R²={ts['r_squared']:.2f}"
        p_val_str = format_pvalue(ts['p_value'])
        pct_change = ts['percent_change']
        
        if not np.isnan(pct_change):
            pct_sign = '+' if pct_change > 0 else ''
            pct_str = f"Δ{pct_sign}{pct_change:.1f}%"
        else:
            pct_str = "Δ=N/A"
        
        stats_text_grv = f"GRV\n{r2_str}\n{p_val_str}\n{pct_str}"
        
        bbox_props_grv = dict(
            boxstyle='round,pad=0.4',
            facecolor='white',
            edgecolor=colors['GRV']['trend_line'],
            linewidth=2.5,
            alpha=0.95
        )
        
        ax.text(0.96, 0.96, stats_text_grv,
                transform=ax.transAxes,
                fontsize=settings['stats_fontsize'],
                verticalalignment='top',
                horizontalalignment='right',
                bbox=bbox_props_grv,
                fontweight='bold',
                family='monospace',
                linespacing=1.2,
                zorder=100)
    
    ax.set_xlabel(year_label, fontsize=settings['axis_label_fontsize'], fontweight='bold')
    ax.set_ylabel(display_name, fontsize=settings['axis_label_fontsize'], fontweight='bold')
    ax.set_title(display_name, fontsize=settings['title_fontsize'], fontweight='bold', pad=10)
    
    ax.grid(True, alpha=0.2, linestyle=':', linewidth=0.6, zorder=1)
    ax.set_axisbelow(True)
    
    if show_legend:
        ax.legend(loc='lower left', frameon=True, framealpha=0.95, 
                  edgecolor='black', fontsize=settings['legend_fontsize'], 
                  ncol=2, handletextpad=0.4, columnspacing=0.8,
                  borderpad=0.4, labelspacing=0.3)
    
    ax.tick_params(axis='both', which='major', labelsize=settings['tick_fontsize'], 
                   width=2, length=6, pad=5)
    
    all_years = []
    if stats_gag is not None:
        all_years.extend(stats_gag['years'])
    if stats_grv is not None:
        all_years.extend(stats_grv['years'])
    
    if len(all_years) > 0:
        all_tick_years, display_years = get_year_tick_labels(all_years, max_labels=4)
        ax.set_xticks(display_years)
        ax.set_xticklabels([int(y) for y in display_years], rotation=0, ha='center')
    
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    
    ax.spines['left'].set_linewidth(2)
    ax.spines['left'].set_color('black')
    ax.spines['bottom'].set_linewidth(2)
    ax.spines['bottom'].set_color('black')

# ============================================================================
# SAVE INDIVIDUAL PLOTS
# ============================================================================

def save_individual_plot(stats_gag, stats_grv, var_name, var_info, 
                        output_dir, colors, settings, language, dpi=600):
    plt.style.use('seaborn-v0_8-whitegrid')
    
    fig, ax = plt.subplots(1, 1, figsize=(8, 6.5))
    
    plot_dual_trends_nature(ax, stats_gag, stats_grv, var_name, var_info, 
                            colors, settings, language, show_legend=True)
    
    plt.tight_layout()
    
    safe_filename = var_name.replace('_', '-').replace('/', '-')
    
    png_path = os.path.join(output_dir, f'{safe_filename}_GAG-vs-GRV.png')
    plt.savefig(png_path, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    
    pdf_path = os.path.join(output_dir, f'{safe_filename}_GAG-vs-GRV.pdf')
    plt.savefig(pdf_path, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    
    plt.close()
    
    return png_path, pdf_path

# ============================================================================
# CREATE COMBINED PLOT (3 COLUMNS × 2 ROWS)
# ============================================================================

def create_sugarcane_trends_plot(data_file, response_variables_all, output_dir_base, 
                                 language='en', dpi=600, colors=COLORS, 
                                 settings=NATURE_SETTINGS):
    
    lang_label = "ENGLISH" if language == 'en' else "FRANÇAIS"
    print("="*80)
    print(f"SUGARCANE KEY VARIABLES - IMPROVED NATURE STYLE ({lang_label})")
    print("="*80)
    
    print(f"\nLoading data from: {data_file}")
    df = pd.read_csv(data_file)
    
    year_column = None
    for col in ['campaign_year', 'Harvest_Year', 'year', 'Campaign_Year']:
        if col in df.columns:
            year_column = col
            break
    
    if year_column is None:
        raise ValueError("No year column found!")
    
    if 'irrigation_type' not in df.columns:
        raise ValueError("irrigation_type column not found!")
    
    irrigation_types = df['irrigation_type'].unique()
    print(f"Irrigation types: {irrigation_types}")
    print(f"Year column: {year_column}")
    print(f"Years: {sorted(df[year_column].unique())}")
    
    os.makedirs(output_dir_base, exist_ok=True)
    
    response_variables = response_variables_all[language]
    
    output_files = {
        'combined_plot': [],
        'individual_plots': [],
        'summaries': []
    }
    
    individual_dir = os.path.join(output_dir_base, 'individual_plots')
    os.makedirs(individual_dir, exist_ok=True)
    
    print("\n" + "▓"*80)
    print("CALCULATING STATISTICS FOR 6 KEY VARIABLES")
    print("▓"*80)
    
    all_stats = {}
    
    for var_name, var_info in response_variables.items():
        if var_name not in df.columns:
            print(f"  ⚠ {var_name} not found, skipping...")
            continue
        
        stats_gag = calculate_temporal_statistics(df, var_name, year_column, 'GAG')
        stats_grv = calculate_temporal_statistics(df, var_name, year_column, 'GRV')
        
        all_stats[var_name] = {'GAG': stats_gag, 'GRV': stats_grv}
        
        print(f"\n  • {var_info['name']}:")
        
        if stats_gag is not None and stats_gag['trend_stats'] is not None:
            ts = stats_gag['trend_stats']
            sig = "✓" if ts['significant'] else "✗"
            print(f"      GAG: {sig} (R²={ts['r_squared']:.3f}, Δ={ts['percent_change']:+.1f}%)")
        else:
            print(f"      GAG: Insufficient data")
        
        if stats_grv is not None and stats_grv['trend_stats'] is not None:
            ts = stats_grv['trend_stats']
            sig = "✓" if ts['significant'] else "✗"
            print(f"      GRV: {sig} (R²={ts['r_squared']:.3f}, Δ={ts['percent_change']:+.1f}%)")
        else:
            print(f"      GRV: Insufficient data")
    
    print(f"\nCreating NATURE 3×2 grid plot...")
    
    plt.style.use('seaborn-v0_8-whitegrid')
    
    n_rows = 2
    n_cols = 3
    
    fig_width = n_cols * settings['subplot_width']
    fig_height = n_rows * settings['subplot_height']
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(fig_width, fig_height))
    axes_flat = axes.flatten()
    
    for idx, (var_name, var_info) in enumerate(response_variables.items()):
        if var_name not in all_stats:
            continue
        
        ax = axes_flat[idx]
        stats_gag = all_stats[var_name]['GAG']
        stats_grv = all_stats[var_name]['GRV']
        
        plot_dual_trends_nature(ax, stats_gag, stats_grv, var_name, var_info, 
                               colors, settings, language, show_legend=False)
    
    plot_title = LABELS[language]['subplot_title']
    fig.suptitle(plot_title, fontsize=settings['suptitle_fontsize'], 
                 fontweight='bold', y=0.998)
    
    from matplotlib.lines import Line2D
    legend_elements = [
        Line2D([0], [0], marker='o', color='w', 
               markerfacecolor=colors['GAG']['data_points'], 
               markersize=9, label='GAG',
               markeredgecolor=colors['GAG']['data_points'], markeredgewidth=2),
        Line2D([0], [0], marker='s', color='w', 
               markerfacecolor=colors['GRV']['data_points'], 
               markersize=9, label='GRV',
               markeredgecolor=colors['GRV']['data_points'], markeredgewidth=2)
    ]
    
    fig.legend(handles=legend_elements, 
               loc='lower center', 
               ncol=2, 
               fontsize=settings['legend_fontsize'],
               frameon=False,
               bbox_to_anchor=(0.5, -0.01),
               columnspacing=2.5,
               handletextpad=0.6)
    
    plt.tight_layout(rect=[0, 0.02, 1, 0.996], 
                     h_pad=settings['h_pad'],
                     w_pad=settings['w_pad'])
    
    filename = 'Fig_Sugarcane_key_variables_3x2_improved.png' if language == 'en' else 'Fig_Variables_cles_canne_3x2_ameliore.png'
    output_file = os.path.join(output_dir_base, filename)
    
    print(f"\nSaving combined plot: {os.path.basename(output_file)}")
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    print(f"✓ PNG saved at {dpi} DPI")
    
    pdf_file = output_file.replace('.png', '.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    print(f"✓ PDF saved")
    
    plt.close()
    
    output_files['combined_plot'] = [output_file, pdf_file]
    
    print(f"\nSaving individual plots to: individual_plots/")
    
    for var_name, var_info in response_variables.items():
        if var_name not in all_stats:
            continue
        
        stats_gag = all_stats[var_name]['GAG']
        stats_grv = all_stats[var_name]['GRV']
        
        png_path, pdf_path = save_individual_plot(
            stats_gag, stats_grv, var_name, var_info, 
            individual_dir, colors, settings, language, dpi
        )
        
        output_files['individual_plots'].extend([png_path, pdf_path])
        print(f"  ✓ {var_info['name']}")
    
    summary_data = []
    for var_name, var_info in response_variables.items():
        if var_name not in all_stats:
            continue
        
        for irr_type in ['GAG', 'GRV']:
            stats_dict = all_stats[var_name][irr_type]
            
            if stats_dict is not None and stats_dict['trend_stats'] is not None:
                trend_stats = stats_dict['trend_stats']
                
                summary_data.append({
                    'Variable': var_info['name'],
                    'Irrigation': irr_type,
                    'Slope': f"{trend_stats['slope']:.4f}",
                    'R²': f"{trend_stats['r_squared']:.3f}",
                    'P-value': f"{trend_stats['p_value']:.4e}",
                    'Significant': 'Yes' if trend_stats['significant'] else 'No',
                    'Change (%)': f"{trend_stats['percent_change']:+.1f}" if not np.isnan(trend_stats['percent_change']) else 'N/A',
                })
    
    if len(summary_data) > 0:
        summary_df = pd.DataFrame(summary_data)
        summary_file = output_file.replace('.png', '_summary.csv')
        summary_df.to_csv(summary_file, index=False)
        output_files['summaries'].append(summary_file)
        print(f"✓ Summary saved")
    
    print("\n" + "="*80)
    print(f"✓ {lang_label} VERSION COMPLETE!")
    print("="*80)
    
    return output_files

# ============================================================================
# MAIN EXECUTION
# ============================================================================

if __name__ == "__main__":
    

    
    output_files_en = create_sugarcane_trends_plot(
        data_file=DATA_FILE,
        response_variables_all=RESPONSE_VARIABLES,
        output_dir_base=OUTPUT_DIR_BASE_EN,
        language='en',
        dpi=OUTPUT_DPI,
        colors=COLORS,
        settings=NATURE_SETTINGS
    )
    
    
    output_files_fr = create_sugarcane_trends_plot(
        data_file=DATA_FILE,
        response_variables_all=RESPONSE_VARIABLES,
        output_dir_base=OUTPUT_DIR_BASE_FR,
        language='fr',
        dpi=OUTPUT_DPI,
        colors=COLORS,
        settings=NATURE_SETTINGS
    )
    
    print("\n" + "="*80)
    print("BILINGUAL PLOTS COMPLETE!")
    print("="*80)
```

# SUGARCANE 3 KEY VARIABLES (1×3 LAYOUT)

```python
"""
SUGARCANE 3 KEY VARIABLES (1×3 LAYOUT)
======================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
import warnings
import os
warnings.filterwarnings('ignore')

# Enable LaTeX-like rendering
plt.rcParams['text.usetex'] = False
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['mathtext.default'] = 'regular'

# ============================================================================
# CONFIGURATION
# ============================================================================

# Path to your data (ADJUST THIS)
DATA_FILE = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/final_dataset_regrouped_categorical.csv'

# Output configuration
OUTPUT_DIR_BASE_EN = 'sugarcane_3vars_nature_english'
OUTPUT_DIR_BASE_FR = 'sugarcane_3vars_nature_french'
OUTPUT_DPI = 600

# Language-specific labels
LABELS = {
    'en': {
        'year': 'Year',
        'main_title': '',
    },
    'fr': {
        'year': 'Année',
        'main_title': '',
    }
}

# BILINGUAL Response variables with proper LaTeX formatting
# Only 3 variables: Yield, Saccharose, TPER
RESPONSE_VARIABLES = {
    'en': {
        'tc_per_ha_actual': {'name': r'Cane Yield (t $ha^{-1}$)', 'unit': r't $ha^{-1}$ $year^{-1}$'},
        'hugot_sugar': {'name': 'Sucrose (%)', 'unit': r'% $year^{-1}$'},
        'total_recoverable_sugar_t_per_ha': {'name': r'TPER (t $ha^{-1}$)', 'unit': r't $ha^{-1}$ $year^{-1}$'}
    },
    'fr': {
        'tc_per_ha_actual': {'name': r'Rendement canne (t $ha^{-1}$)', 'unit': r't $ha^{-1}$ $an^{-1}$'},
        'hugot_sugar': {'name': 'Saccharose (%)', 'unit': r'% $an^{-1}$'},
        'total_recoverable_sugar_t_per_ha': {'name': r'TPER (t $ha^{-1}$)', 'unit': r't $ha^{-1}$ $an^{-1}$'}
    }
}

# Nature journal color scheme
COLORS = {
    'GAG': {
        'data_points': '#2E86AB',
        'trend_line': '#1E5F8C',
        'error_bars': '#2E86AB',
    },
    'GRV': {
        'data_points': '#E63946',
        'trend_line': '#C1121F',
        'error_bars': '#E63946',
    }
}

# IMPROVED NATURE JOURNAL SETTINGS - LARGER FONTS
NATURE_SETTINGS = {
    'subplot_width': 5.0,        # Wider for 1×3 layout
    'subplot_height': 3.8,       # Taller for better proportions
    'title_fontsize': 16,        
    'axis_label_fontsize': 15,   
    'tick_fontsize': 14,         
    'stats_fontsize': 10,        
    'legend_fontsize': 14,       
    'suptitle_fontsize': 18,     
    'marker_size': 6,            
    'line_width': 3.0,           
    'error_bar_width': 2.0,      
    'cap_size': 4,               
    'h_pad': 1.5,                
    'w_pad': 2.5,                
}

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def calculate_temporal_statistics(df, response_var, year_column, irrigation_type):
    """Calculate temporal statistics for a response variable filtered by irrigation type"""
    df_filtered = df[df['irrigation_type'] == irrigation_type].copy()
    df_clean = df_filtered[[year_column, response_var]].dropna()
    
    if len(df_clean) < 3:
        return None
    
    yearly_stats = df_clean.groupby(year_column)[response_var].agg([
        ('mean', 'mean'),
        ('std', 'std'),
        ('count', 'count')
    ]).reset_index()
    
    years = yearly_stats[year_column].values
    means = yearly_stats['mean'].values
    stds = yearly_stats['std'].values
    
    if len(years) >= 3:
        slope, intercept, r_value, p_value, std_err = stats.linregress(years, means)
        
        first_value = means[0]
        last_value = means[-1]
        
        if abs(first_value) > 1e-10:
            percent_change = ((last_value - first_value) / abs(first_value)) * 100
        else:
            percent_change = np.nan
        
        absolute_change = last_value - first_value
        
        trend_stats = {
            'slope': slope,
            'intercept': intercept,
            'r_squared': r_value**2,
            'p_value': p_value,
            'significant': p_value < 0.05,
            'percent_change': percent_change,
            'absolute_change': absolute_change,
            'years_span': years[-1] - years[0],
            'first_value': first_value,
            'last_value': last_value
        }
    else:
        trend_stats = None
    
    return {
        'years': years,
        'means': means,
        'stds': stds,
        'yearly_stats': yearly_stats,
        'trend_stats': trend_stats,
        'irrigation_type': irrigation_type
    }

def format_pvalue(p):
    """Format p-value for display - compact Nature style"""
    if p < 0.001:
        return 'p<0.001'
    elif p < 0.01:
        return 'p<0.01'
    elif p < 0.05:
        return f'p={p:.3f}'
    else:
        return f'p={p:.2f}'

def get_year_tick_labels(years, max_labels=4):
    """Intelligently select year labels to prevent overlap"""
    unique_years = sorted(set(years))
    n_years = len(unique_years)
    
    if n_years <= max_labels:
        return unique_years, unique_years
    
    # Always include first and last
    selected_years = [unique_years[0]]
    
    # Calculate how many intermediate years to show
    n_intermediate = max_labels - 2
    
    if n_intermediate > 0:
        # Calculate step to evenly space intermediate years
        step = (n_years - 1) / (n_intermediate + 1)
        
        for i in range(1, n_intermediate + 1):
            idx = int(round(i * step))
            if idx > 0 and idx < n_years - 1:
                selected_years.append(unique_years[idx])
    
    # Always add last year
    selected_years.append(unique_years[-1])
    
    # Remove duplicates and sort
    selected_years = sorted(list(set(selected_years)))
    
    return unique_years, selected_years

def plot_dual_trends_nature(ax, stats_gag, stats_grv, var_name, var_info, colors, 
                            settings, language='en', show_legend=False):
    """
    Plot temporal trends for both GAG and GRV - IMPROVED NATURE STYLE
    Features: L-shaped axes, larger fonts, cleaner appearance
    """
    display_name = var_info['name']
    year_label = LABELS[language]['year']
    
    # Plot GAG data (Blue)
    if stats_gag is not None:
        years_gag = stats_gag['years']
        means_gag = stats_gag['means']
        stds_gag = stats_gag['stds']
        
        ax.errorbar(years_gag, means_gag, yerr=stds_gag, 
                    fmt='o', markersize=settings['marker_size'], 
                    color=colors['GAG']['data_points'],
                    ecolor=colors['GAG']['error_bars'],
                    elinewidth=settings['error_bar_width'],
                    capsize=settings['cap_size'],
                    capthick=settings['error_bar_width'],
                    alpha=0.85,
                    zorder=3,
                    label='GAG')
        
        ax.plot(years_gag, means_gag, '-', 
                color=colors['GAG']['data_points'], 
                alpha=0.25, 
                linewidth=settings['line_width'] * 0.5,
                zorder=2)
        
        if stats_gag['trend_stats'] is not None:
            trend_line_gag = stats_gag['trend_stats']['slope'] * years_gag + stats_gag['trend_stats']['intercept']
            ax.plot(years_gag, trend_line_gag, '--', 
                    color=colors['GAG']['trend_line'],
                    linewidth=settings['line_width'],
                    alpha=0.9,
                    zorder=4)
    
    # Plot GRV data (Red)
    if stats_grv is not None:
        years_grv = stats_grv['years']
        means_grv = stats_grv['means']
        stds_grv = stats_grv['stds']
        
        ax.errorbar(years_grv, means_grv, yerr=stds_grv, 
                    fmt='s', markersize=settings['marker_size'], 
                    color=colors['GRV']['data_points'],
                    ecolor=colors['GRV']['error_bars'],
                    elinewidth=settings['error_bar_width'],
                    capsize=settings['cap_size'],
                    capthick=settings['error_bar_width'],
                    alpha=0.85,
                    zorder=3,
                    label='GRV')
        
        ax.plot(years_grv, means_grv, '-', 
                color=colors['GRV']['data_points'], 
                alpha=0.25, 
                linewidth=settings['line_width'] * 0.5,
                zorder=2)
        
        if stats_grv['trend_stats'] is not None:
            trend_line_grv = stats_grv['trend_stats']['slope'] * years_grv + stats_grv['trend_stats']['intercept']
            ax.plot(years_grv, trend_line_grv, '--', 
                    color=colors['GRV']['trend_line'],
                    linewidth=settings['line_width'],
                    alpha=0.9,
                    zorder=4)
    
    # Statistics boxes - LARGER FONTS
    # GAG statistics (top-left)
    if stats_gag is not None and stats_gag['trend_stats'] is not None:
        ts = stats_gag['trend_stats']
        r2_str = f"R²={ts['r_squared']:.2f}"
        p_val_str = format_pvalue(ts['p_value'])
        pct_change = ts['percent_change']
        
        if not np.isnan(pct_change):
            pct_sign = '+' if pct_change > 0 else ''
            pct_str = f"Δ{pct_sign}{pct_change:.1f}%"
        else:
            pct_str = "Δ=N/A"
        
        stats_text_gag = f"GAG\n{r2_str}\n{p_val_str}\n{pct_str}"
        
        bbox_props_gag = dict(
            boxstyle='round,pad=0.4',
            facecolor='white',
            edgecolor=colors['GAG']['trend_line'],
            linewidth=2.5,
            alpha=0.95
        )
        
        ax.text(0.04, 0.96, stats_text_gag,
                transform=ax.transAxes,
                fontsize=settings['stats_fontsize'],
                verticalalignment='top',
                horizontalalignment='left',
                bbox=bbox_props_gag,
                fontweight='bold',
                family='monospace',
                linespacing=1.2,
                zorder=100)
    
    # GRV statistics (top-right)
    if stats_grv is not None and stats_grv['trend_stats'] is not None:
        ts = stats_grv['trend_stats']
        r2_str = f"R²={ts['r_squared']:.2f}"
        p_val_str = format_pvalue(ts['p_value'])
        pct_change = ts['percent_change']
        
        if not np.isnan(pct_change):
            pct_sign = '+' if pct_change > 0 else ''
            pct_str = f"Δ{pct_sign}{pct_change:.1f}%"
        else:
            pct_str = "Δ=N/A"
        
        stats_text_grv = f"GRV\n{r2_str}\n{p_val_str}\n{pct_str}"
        
        bbox_props_grv = dict(
            boxstyle='round,pad=0.4',
            facecolor='white',
            edgecolor=colors['GRV']['trend_line'],
            linewidth=2.5,
            alpha=0.95
        )
        
        ax.text(0.96, 0.96, stats_text_grv,
                transform=ax.transAxes,
                fontsize=settings['stats_fontsize'],
                verticalalignment='top',
                horizontalalignment='right',
                bbox=bbox_props_grv,
                fontweight='bold',
                family='monospace',
                linespacing=1.2,
                zorder=100)
    
    # Formatting with LARGER FONTS
    ax.set_xlabel(year_label, fontsize=settings['axis_label_fontsize'], fontweight='bold')
    ax.set_ylabel(display_name, fontsize=settings['axis_label_fontsize'], fontweight='bold')
    ax.set_title(display_name, fontsize=settings['title_fontsize'], fontweight='bold', pad=10)
    
    # Minimal grid - Nature style
    ax.grid(True, alpha=0.2, linestyle=':', linewidth=0.6, zorder=1)
    ax.set_axisbelow(True)
    
    # Legend - only if requested
    if show_legend:
        ax.legend(loc='lower left', frameon=True, framealpha=0.95, 
                  edgecolor='black', fontsize=settings['legend_fontsize'], 
                  ncol=2, handletextpad=0.4, columnspacing=0.8,
                  borderpad=0.4, labelspacing=0.3)
    
    # Tick parameters - LARGER FONTS
    ax.tick_params(axis='both', which='major', labelsize=settings['tick_fontsize'], 
                   width=2, length=6, pad=5)
    
    # Year axis labels - prevent overlap
    all_years = []
    if stats_gag is not None:
        all_years.extend(stats_gag['years'])
    if stats_grv is not None:
        all_years.extend(stats_grv['years'])
    
    if len(all_years) > 0:
        all_tick_years, display_years = get_year_tick_labels(all_years, max_labels=4)
        ax.set_xticks(display_years)
        ax.set_xticklabels([int(y) for y in display_years], rotation=0, ha='center')
    
    # ★★★ L-SHAPED AXES - ONLY LEFT AND BOTTOM ★★★
    ax.spines['top'].set_visible(False)
    ax.spines['right'].set_visible(False)
    ax.spines['left'].set_linewidth(2)
    ax.spines['left'].set_color('black')
    ax.spines['bottom'].set_linewidth(2)
    ax.spines['bottom'].set_color('black')

# ============================================================================
# SAVE INDIVIDUAL PLOTS
# ============================================================================

def save_individual_plot(stats_gag, stats_grv, var_name, var_info, 
                        output_dir, colors, settings, language, dpi=600):
    """Save standalone plot with L-shaped axes"""
    plt.style.use('seaborn-v0_8-whitegrid')
    
    fig, ax = plt.subplots(1, 1, figsize=(8, 6.5))
    
    plot_dual_trends_nature(ax, stats_gag, stats_grv, var_name, var_info, 
                            colors, settings, language, show_legend=True)
    
    plt.tight_layout()
    
    safe_filename = var_name.replace('_', '-').replace('/', '-')
    
    png_path = os.path.join(output_dir, f'{safe_filename}_GAG-vs-GRV.png')
    plt.savefig(png_path, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    
    pdf_path = os.path.join(output_dir, f'{safe_filename}_GAG-vs-GRV.pdf')
    plt.savefig(pdf_path, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    
    plt.close()
    
    return png_path, pdf_path

# ============================================================================
# CREATE COMBINED PLOT (1 ROW × 3 COLUMNS)
# ============================================================================

def create_sugarcane_trends_plot(data_file, response_variables_all, output_dir_base, 
                                 language='en', dpi=600, colors=COLORS, 
                                 settings=NATURE_SETTINGS):
    """
    Create sugarcane trends plot with 1×3 layout
    Features: Yield, Saccharose, TPER only
    """
    
    lang_label = "ENGLISH" if language == 'en' else "FRANÇAIS"
    print("="*80)
    print(f"SUGARCANE 3 VARIABLES (1×3) - NATURE STYLE ({lang_label})")
    print("="*80)
    
    print(f"\nLoading data from: {data_file}")
    df = pd.read_csv(data_file)
    
    year_column = None
    for col in ['campaign_year', 'Harvest_Year', 'year', 'Campaign_Year']:
        if col in df.columns:
            year_column = col
            break
    
    if year_column is None:
        raise ValueError("No year column found!")
    
    if 'irrigation_type' not in df.columns:
        raise ValueError("irrigation_type column not found!")
    
    irrigation_types = df['irrigation_type'].unique()
    print(f"Irrigation types: {irrigation_types}")
    print(f"Year column: {year_column}")
    print(f"Years: {sorted(df[year_column].unique())}")
    
    os.makedirs(output_dir_base, exist_ok=True)
    
    # Get language-specific variables
    response_variables = response_variables_all[language]
    
    output_files = {
        'combined_plot': [],
        'individual_plots': [],
        'summaries': []
    }
    
    # Create individual plots directory
    individual_dir = os.path.join(output_dir_base, 'individual_plots')
    os.makedirs(individual_dir, exist_ok=True)
    
    print("\n" + "▓"*80)
    print("CALCULATING STATISTICS FOR 3 KEY VARIABLES")
    print("▓"*80)
    
    all_stats = {}
    
    for var_name, var_info in response_variables.items():
        if var_name not in df.columns:
            print(f"  ⚠ {var_name} not found, skipping...")
            continue
        
        stats_gag = calculate_temporal_statistics(df, var_name, year_column, 'GAG')
        stats_grv = calculate_temporal_statistics(df, var_name, year_column, 'GRV')
        
        all_stats[var_name] = {'GAG': stats_gag, 'GRV': stats_grv}
        
        print(f"\n  • {var_info['name']}:")
        
        if stats_gag is not None and stats_gag['trend_stats'] is not None:
            ts = stats_gag['trend_stats']
            sig = "✓" if ts['significant'] else "✗"
            print(f"      GAG: {sig} (R²={ts['r_squared']:.3f}, Δ={ts['percent_change']:+.1f}%)")
        else:
            print(f"      GAG: Insufficient data")
        
        if stats_grv is not None and stats_grv['trend_stats'] is not None:
            ts = stats_grv['trend_stats']
            sig = "✓" if ts['significant'] else "✗"
            print(f"      GRV: {sig} (R²={ts['r_squared']:.3f}, Δ={ts['percent_change']:+.1f}%)")
        else:
            print(f"      GRV: Insufficient data")
    
    print(f"\nCreating NATURE 1×3 grid plot...")
    print(f"  → Layout: 1 row × 3 columns")
    print(f"  → Variables: Yield, Saccharose, TPER")
    print(f"  → L-shaped axes (left + bottom only)")
    
    plt.style.use('seaborn-v0_8-whitegrid')
    
    # 1 row × 3 columns
    n_rows = 1
    n_cols = 3
    
    # Calculate figure size
    fig_width = n_cols * settings['subplot_width']
    fig_height = n_rows * settings['subplot_height']
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(fig_width, fig_height))
    
    # Plot each variable WITHOUT individual legends
    for idx, (var_name, var_info) in enumerate(response_variables.items()):
        if var_name not in all_stats:
            continue
        
        ax = axes[idx]
        stats_gag = all_stats[var_name]['GAG']
        stats_grv = all_stats[var_name]['GRV']
        
        # No legend in individual subplots
        plot_dual_trends_nature(ax, stats_gag, stats_grv, var_name, var_info, 
                               colors, settings, language, show_legend=False)
    
    plot_title = LABELS[language]['main_title']
    fig.suptitle(plot_title, fontsize=settings['suptitle_fontsize'], 
                 fontweight='bold', y=0.998)
    
    # Add single legend at the bottom
    from matplotlib.lines import Line2D
    legend_elements = [
        Line2D([0], [0], marker='o', color='w', 
               markerfacecolor=colors['GAG']['data_points'], 
               markersize=9, label='GAG',
               markeredgecolor=colors['GAG']['data_points'], markeredgewidth=2),
        Line2D([0], [0], marker='s', color='w', 
               markerfacecolor=colors['GRV']['data_points'], 
               markersize=9, label='GRV',
               markeredgecolor=colors['GRV']['data_points'], markeredgewidth=2)
    ]
    
    fig.legend(handles=legend_elements, 
               loc='lower center', 
               ncol=2, 
               fontsize=settings['legend_fontsize'],
               frameon=False,
               bbox_to_anchor=(0.5, -0.02),
               columnspacing=2.5,
               handletextpad=0.6)
    
    plt.tight_layout(rect=[0, 0.03, 1, 0.995], 
                     h_pad=settings['h_pad'],
                     w_pad=settings['w_pad'])
    
    # Save combined plot
    filename = 'Fig_Sugarcane_3vars_1x3.png' if language == 'en' else 'Fig_Canne_3vars_1x3.png'
    output_file = os.path.join(output_dir_base, filename)
    
    print(f"\nSaving combined plot: {os.path.basename(output_file)}")
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    print(f"✓ PNG saved at {dpi} DPI")
    
    pdf_file = output_file.replace('.png', '.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    print(f"✓ PDF saved")
    
    plt.close()
    
    output_files['combined_plot'] = [output_file, pdf_file]
    
    # Save individual plots
    print(f"\nSaving individual plots to: individual_plots/")
    
    for var_name, var_info in response_variables.items():
        if var_name not in all_stats:
            continue
        
        stats_gag = all_stats[var_name]['GAG']
        stats_grv = all_stats[var_name]['GRV']
        
        png_path, pdf_path = save_individual_plot(
            stats_gag, stats_grv, var_name, var_info, 
            individual_dir, colors, settings, language, dpi
        )
        
        output_files['individual_plots'].extend([png_path, pdf_path])
        print(f"  ✓ {var_info['name']}")
    
    # Save summary
    summary_data = []
    for var_name, var_info in response_variables.items():
        if var_name not in all_stats:
            continue
        
        for irr_type in ['GAG', 'GRV']:
            stats_dict = all_stats[var_name][irr_type]
            
            if stats_dict is not None and stats_dict['trend_stats'] is not None:
                trend_stats = stats_dict['trend_stats']
                
                summary_data.append({
                    'Variable': var_info['name'],
                    'Irrigation': irr_type,
                    'Slope': f"{trend_stats['slope']:.4f}",
                    'R²': f"{trend_stats['r_squared']:.3f}",
                    'P-value': f"{trend_stats['p_value']:.4e}",
                    'Significant': 'Yes' if trend_stats['significant'] else 'No',
                    'Change (%)': f"{trend_stats['percent_change']:+.1f}" if not np.isnan(trend_stats['percent_change']) else 'N/A',
                })
    
    if len(summary_data) > 0:
        summary_df = pd.DataFrame(summary_data)
        summary_file = output_file.replace('.png', '_summary.csv')
        summary_df.to_csv(summary_file, index=False)
        output_files['summaries'].append(summary_file)
        print(f"✓ Summary saved")
    
    print("\n" + "="*80)
    print(f"✓ {lang_label} VERSION COMPLETE!")
    print("="*80)
    
    return output_files

# ============================================================================
# MAIN EXECUTION
# ============================================================================

if __name__ == "__main__":
    
    
    output_files_en = create_sugarcane_trends_plot(
        data_file=DATA_FILE,
        response_variables_all=RESPONSE_VARIABLES,
        output_dir_base=OUTPUT_DIR_BASE_EN,
        language='en',
        dpi=OUTPUT_DPI,
        colors=COLORS,
        settings=NATURE_SETTINGS
    )
    
    output_files_fr = create_sugarcane_trends_plot(
        data_file=DATA_FILE,
        response_variables_all=RESPONSE_VARIABLES,
        output_dir_base=OUTPUT_DIR_BASE_FR,
        language='fr',
        dpi=OUTPUT_DPI,
        colors=COLORS,
        settings=NATURE_SETTINGS
    )
    
```

# COMPREHENSIVE SUGARCANE VARIABLES ANALYSIS

```python
"""
COMPREHENSIVE SUGARCANE VARIABLES ANALYSIS
================================================================================
"""

import pandas as pd
import numpy as np
from scipy import stats
import os

# ============================================================================
# CONFIGURATION
# ============================================================================

DATA_FILE = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/final_dataset_regrouped_categorical.csv'
OUTPUT_DIR = 'sugarcane_comprehensive_analysis'

VARIABLES = [
    'tc_per_ha_actual',
    'brix_percent',
    'hugot_sugar',
    'purity_percent',
    'estimated_recoverable_pol',
    'total_recoverable_sugar_t_per_ha'
]

VARIABLE_NAMES = {
    'tc_per_ha_actual': 'Cane Yield (t/ha)',
    'brix_percent': 'Brix (%)',
    'hugot_sugar': 'Saccharose (%)',
    'purity_percent': 'Purity (%)',
    'estimated_recoverable_pol': 'PER (%)',
    'total_recoverable_sugar_t_per_ha': 'TPER (t/ha)'
}

# ============================================================================
# TEMPORAL TREND FUNCTIONS
# ============================================================================

def calculate_temporal_trend(df, response_var, year_column, irrigation_type=None):
    """Calculate temporal trend statistics"""
    
    if irrigation_type is not None:
        df_filtered = df[df['irrigation_type'] == irrigation_type].copy()
    else:
        df_filtered = df.copy()
    
    df_clean = df_filtered[[year_column, response_var]].dropna()
    
    if len(df_clean) < 3:
        return None
    
    yearly_stats = df_clean.groupby(year_column)[response_var].agg([
        ('mean', 'mean'),
        ('std', 'std'),
        ('count', 'count')
    ]).reset_index()
    
    years = yearly_stats[year_column].values
    means = yearly_stats['mean'].values
    
    if len(years) >= 3:
        slope, intercept, r_value, p_value, std_err = stats.linregress(years, means)
        
        first_value = means[0]
        last_value = means[-1]
        
        if abs(first_value) > 1e-10:
            percent_change = ((last_value - first_value) / abs(first_value)) * 100
        else:
            percent_change = np.nan
        
        return {
            'slope': slope,
            'intercept': intercept,
            'r_squared': r_value**2,
            'p_value': p_value,
            'percent_change': percent_change,
            'first_value': first_value,
            'last_value': last_value,
            'n_years': len(years)
        }
    
    return None

def format_pvalue_category(p):
    """Categorize p-value"""
    if p < 0.001:
        return 'p<0.001'
    elif p < 0.01:
        return 'p<0.01'
    elif p < 0.05:
        return 'p<0.05'
    else:
        return 'ns'

def format_significance(p, percent_change):
    """Format significance with direction"""
    if p >= 0.05:
        return 'ns'
    
    sign = '+' if percent_change > 0 else ''
    p_cat = format_pvalue_category(p)
    
    return f"{sign}{percent_change:.1f}% ({p_cat})"

# ============================================================================
# ANALYSIS FUNCTIONS
# ============================================================================

def analyze_temporal_trends(df, variables, year_column):
    """Analyze temporal trends - overall and by irrigation type"""
    
    print("\n" + "="*80)
    print("PART 1: TEMPORAL TRENDS ANALYSIS")
    print("="*80)
    
    trends_data = []
    
    # Overall trends
    print("\n" + "-"*80)
    print("OVERALL TEMPORAL TRENDS (ALL DATA)")
    print("-"*80)
    
    for var in variables:
        if var not in df.columns:
            print(f"⚠ {var} not found, skipping...")
            continue
        
        trend = calculate_temporal_trend(df, var, year_column)
        
        if trend is not None:
            sig_str = format_significance(trend['p_value'], trend['percent_change'])
            
            print(f"{VARIABLE_NAMES[var]:<30} : R²={trend['r_squared']:.3f}, {sig_str}")
            
            trends_data.append({
                'Variable': VARIABLE_NAMES[var],
                'Irrigation_Type': 'ALL',
                'R_squared': trend['r_squared'],
                'P_value': trend['p_value'],
                'P_category': format_pvalue_category(trend['p_value']),
                'Percent_Change': trend['percent_change'],
                'Significance': sig_str,
                'N_years': trend['n_years']
            })
        else:
            print(f"{VARIABLE_NAMES[var]:<30} : Insufficient data")
    
    # By irrigation type
    if 'irrigation_type' in df.columns:
        irrigation_types = sorted(df['irrigation_type'].unique())
        
        for irr_type in irrigation_types:
            print(f"\n" + "-"*80)
            print(f"TEMPORAL TRENDS - {irr_type}")
            print("-"*80)
            
            for var in variables:
                if var not in df.columns:
                    continue
                
                trend = calculate_temporal_trend(df, var, year_column, irr_type)
                
                if trend is not None:
                    sig_str = format_significance(trend['p_value'], trend['percent_change'])
                    
                    print(f"{VARIABLE_NAMES[var]:<30} : R²={trend['r_squared']:.3f}, {sig_str}")
                    
                    trends_data.append({
                        'Variable': VARIABLE_NAMES[var],
                        'Irrigation_Type': irr_type,
                        'R_squared': trend['r_squared'],
                        'P_value': trend['p_value'],
                        'P_category': format_pvalue_category(trend['p_value']),
                        'Percent_Change': trend['percent_change'],
                        'Significance': sig_str,
                        'N_years': trend['n_years']
                    })
                else:
                    print(f"{VARIABLE_NAMES[var]:<30} : Insufficient data")
    
    return pd.DataFrame(trends_data)

def calculate_summary_statistics(df, variables):
    """Calculate mean ± std for all variables"""
    
    print("\n" + "="*80)
    print("PART 2: DESCRIPTIVE STATISTICS")
    print("="*80)
    
    summary_data = []
    
    # Overall statistics
    print("\n" + "-"*80)
    print("OVERALL STATISTICS (ALL DATA)")
    print("-"*80)
    
    for var in variables:
        if var not in df.columns:
            continue
        
        data = df[var].dropna()
        mean_val = data.mean()
        std_val = data.std()
        n_samples = len(data)
        
        summary_data.append({
            'Variable': VARIABLE_NAMES[var],
            'Irrigation_Type': 'ALL',
            'Mean': mean_val,
            'Std': std_val,
            'N': n_samples,
            'Mean_Std_Format': f"{mean_val:.2f} ± {std_val:.2f}"
        })
        
        print(f"{VARIABLE_NAMES[var]:<30} : {mean_val:>8.2f} ± {std_val:>6.2f}  (n={n_samples})")
    
    # By irrigation type
    if 'irrigation_type' in df.columns:
        irrigation_types = sorted(df['irrigation_type'].unique())
        
        for irr_type in irrigation_types:
            print(f"\n" + "-"*80)
            print(f"STATISTICS - {irr_type}")
            print("-"*80)
            
            df_filtered = df[df['irrigation_type'] == irr_type]
            
            for var in variables:
                if var not in df.columns:
                    continue
                
                data = df_filtered[var].dropna()
                mean_val = data.mean()
                std_val = data.std()
                n_samples = len(data)
                
                summary_data.append({
                    'Variable': VARIABLE_NAMES[var],
                    'Irrigation_Type': irr_type,
                    'Mean': mean_val,
                    'Std': std_val,
                    'N': n_samples,
                    'Mean_Std_Format': f"{mean_val:.2f} ± {std_val:.2f}"
                })
                
                print(f"{VARIABLE_NAMES[var]:<30} : {mean_val:>8.2f} ± {std_val:>6.2f}  (n={n_samples})")
    
    return pd.DataFrame(summary_data)

def calculate_percentage_differences(df):
    """Calculate absolute percentage point differences: Brix → Saccharose → PER"""
    
    print("\n" + "="*80)
    print("PART 3: PERCENTAGE POINT DIFFERENCES")
    print("="*80)
    
    results = []
    
    # Overall differences
    print("\n" + "-"*80)
    print("OVERALL DIFFERENCES")
    print("-"*80)
    print("\nBrix → Saccharose → PER")
    
    common_idx = df.index[df[['brix_percent', 'hugot_sugar', 'estimated_recoverable_pol']].notna().all(axis=1)]
    
    if len(common_idx) > 0:
        brix_mean = df.loc[common_idx, 'brix_percent'].mean()
        saccharose_mean = df.loc[common_idx, 'hugot_sugar'].mean()
        per_mean = df.loc[common_idx, 'estimated_recoverable_pol'].mean()
        
        diff_brix_to_saccharose = saccharose_mean - brix_mean
        diff_saccharose_to_per = per_mean - saccharose_mean
        diff_brix_to_per = per_mean - brix_mean
        
        print(f"Brix:       {brix_mean:>8.2f}%")
        print(f"Saccharose: {saccharose_mean:>8.2f}%  (Saccharose is {abs(diff_brix_to_saccharose):.2f} pp {'less' if diff_brix_to_saccharose < 0 else 'more'} than Brix)")
        print(f"PER:        {per_mean:>8.2f}%  (PER is {abs(diff_saccharose_to_per):.2f} pp {'less' if diff_saccharose_to_per < 0 else 'more'} than Saccharose)")
        print(f"                     (PER is {abs(diff_brix_to_per):.2f} pp {'less' if diff_brix_to_per < 0 else 'more'} than Brix)")
        
        results.append({
            'Irrigation_Type': 'ALL',
            'Brix_Mean': brix_mean,
            'Saccharose_Mean': saccharose_mean,
            'PER_Mean': per_mean,
            'Saccharose_minus_Brix_pp': diff_brix_to_saccharose,
            'PER_minus_Saccharose_pp': diff_saccharose_to_per,
            'PER_minus_Brix_pp': diff_brix_to_per,
            'N_samples': len(common_idx)
        })
    
    # By irrigation type
    if 'irrigation_type' in df.columns:
        irrigation_types = sorted(df['irrigation_type'].unique())
        
        for irr_type in irrigation_types:
            print(f"\n" + "-"*80)
            print(f"DIFFERENCES - {irr_type}")
            print("-"*80)
            print("\nBrix → Saccharose → PER")
            
            df_filtered = df[df['irrigation_type'] == irr_type]
            
            common_idx_irr = df_filtered.index[df_filtered[['brix_percent', 'hugot_sugar', 'estimated_recoverable_pol']].notna().all(axis=1)]
            
            if len(common_idx_irr) > 0:
                brix_mean = df_filtered.loc[common_idx_irr, 'brix_percent'].mean()
                saccharose_mean = df_filtered.loc[common_idx_irr, 'hugot_sugar'].mean()
                per_mean = df_filtered.loc[common_idx_irr, 'estimated_recoverable_pol'].mean()
                
                diff_brix_to_saccharose = saccharose_mean - brix_mean
                diff_saccharose_to_per = per_mean - saccharose_mean
                diff_brix_to_per = per_mean - brix_mean
                
                print(f"Brix:       {brix_mean:>8.2f}%")
                print(f"Saccharose: {saccharose_mean:>8.2f}%  (Saccharose is {abs(diff_brix_to_saccharose):.2f} pp {'less' if diff_brix_to_saccharose < 0 else 'more'} than Brix)")
                print(f"PER:        {per_mean:>8.2f}%  (PER is {abs(diff_saccharose_to_per):.2f} pp {'less' if diff_saccharose_to_per < 0 else 'more'} than Saccharose)")
                print(f"                     (PER is {abs(diff_brix_to_per):.2f} pp {'less' if diff_brix_to_per < 0 else 'more'} than Brix)")
                
                results.append({
                    'Irrigation_Type': irr_type,
                    'Brix_Mean': brix_mean,
                    'Saccharose_Mean': saccharose_mean,
                    'PER_Mean': per_mean,
                    'Saccharose_minus_Brix_pp': diff_brix_to_saccharose,
                    'PER_minus_Saccharose_pp': diff_saccharose_to_per,
                    'PER_minus_Brix_pp': diff_brix_to_per,
                    'N_samples': len(common_idx_irr)
                })
    
    return pd.DataFrame(results)

# ============================================================================
# MAIN EXECUTION
# ============================================================================

if __name__ == "__main__":
    
    print("="*80)
    print("COMPREHENSIVE SUGARCANE VARIABLES ANALYSIS")
    print("="*80)
    
    print(f"\nLoading data from: {DATA_FILE}")
    df = pd.read_csv(DATA_FILE)
    
    print(f"Total observations: {len(df)}")
    
    # Find year column
    year_column = None
    for col in ['campaign_year', 'Harvest_Year', 'year', 'Campaign_Year']:
        if col in df.columns:
            year_column = col
            break
    
    if year_column is None:
        raise ValueError("No year column found!")
    
    print(f"Year column: {year_column}")
    
    os.makedirs(OUTPUT_DIR, exist_ok=True)
    
    # PART 1: Temporal trends
    trends_df = analyze_temporal_trends(df, VARIABLES, year_column)
    
    # PART 2: Descriptive statistics
    summary_df = calculate_summary_statistics(df, VARIABLES)
    
    # PART 3: Percentage differences
    differences_df = calculate_percentage_differences(df)
    
    # Save all results
    print("\n" + "="*80)
    print("SAVING RESULTS")
    print("="*80)
    
    trends_file = os.path.join(OUTPUT_DIR, '1_temporal_trends.csv')
    trends_df.to_csv(trends_file, index=False)
    print(f"✓ Temporal trends saved: {trends_file}")
    
    summary_file = os.path.join(OUTPUT_DIR, '2_summary_statistics.csv')
    summary_df.to_csv(summary_file, index=False)
    print(f"✓ Summary statistics saved: {summary_file}")
    
    differences_file = os.path.join(OUTPUT_DIR, '3_percentage_differences.csv')
    differences_df.to_csv(differences_file, index=False)
    print(f"✓ Percentage differences saved: {differences_file}")
    
    # Create formatted summary tables
    print("\n" + "="*80)
    print("CREATING FORMATTED SUMMARY TABLES")
    print("="*80)
    
    # Formatted temporal trends
    print("\nTemporal Trends Summary:")
    pivot_trends = trends_df.pivot(index='Variable', columns='Irrigation_Type', values='Significance')
    if all(col in pivot_trends.columns for col in ['ALL', 'GAG', 'GRV']):
        pivot_trends = pivot_trends[['ALL', 'GAG', 'GRV']]
    print(pivot_trends.to_string())
    
    formatted_trends_file = os.path.join(OUTPUT_DIR, '1_temporal_trends_formatted.csv')
    pivot_trends.to_csv(formatted_trends_file)
    print(f"✓ Formatted trends saved: {formatted_trends_file}")
    
    # Formatted statistics
    print("\nDescriptive Statistics Summary:")
    pivot_summary = summary_df.pivot(index='Variable', columns='Irrigation_Type', values='Mean_Std_Format')
    if all(col in pivot_summary.columns for col in ['ALL', 'GAG', 'GRV']):
        pivot_summary = pivot_summary[['ALL', 'GAG', 'GRV']]
    print(pivot_summary.to_string())
    
    formatted_summary_file = os.path.join(OUTPUT_DIR, '2_summary_statistics_formatted.csv')
    pivot_summary.to_csv(formatted_summary_file)
    print(f"✓ Formatted statistics saved: {formatted_summary_file}")
    
```

# SUGARCANE VARIABLES BOXPLOTS 

```python
"""
SUGARCANE VARIABLES BOXPLOTS 
================================================================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
import os
warnings.filterwarnings('ignore')

plt.rcParams['text.usetex'] = False
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['mathtext.default'] = 'regular'

# ============================================================================
# CONFIGURATION
# ============================================================================

DATA_FILE = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/final_dataset_regrouped_categorical.csv'

OUTPUT_DIR = 'sugarcane_boxplots_nature'
OUTPUT_DPI = 600

# Variables in order
VARIABLES_ORDER = [
    'tc_per_ha_actual',
    'brix_percent',
    'hugot_sugar',
    'purity_percent',
    'estimated_recoverable_pol',
    'total_recoverable_sugar_t_per_ha'
]

VARIABLE_NAMES = {
    'tc_per_ha_actual': r'Cane Yield (t $ha^{-1}$)',
    'brix_percent': 'Brix (%)',
    'hugot_sugar': 'Saccharose (%)',
    'purity_percent': 'Purity (%)',
    'estimated_recoverable_pol': 'PER (%)',
    'total_recoverable_sugar_t_per_ha': r'TPER (t $ha^{-1}$)'
}

VARIABLE_YLABELS = {
    'tc_per_ha_actual': r'Cane Yield (t $ha^{-1}$)',
    'brix_percent': 'Brix (%)',
    'hugot_sugar': 'Saccharose (%)',
    'purity_percent': 'Purity (%)',
    'estimated_recoverable_pol': 'PER (%)',
    'total_recoverable_sugar_t_per_ha': r'TPER (t $ha^{-1}$)'
}

# Color scheme
COLORS = {
    'overall': '#2E86AB',
    'GAG': '#2E86AB',
    'GRV': '#E63946'
}

# Plot settings
NATURE_SETTINGS = {
    'subplot_width': 2.0,       # Equal width for each subplot
    'subplot_height': 5.0,      # Height for all subplots
    'title_fontsize': 13,
    'axis_label_fontsize': 12,
    'tick_fontsize': 11,
    'box_width': 0.35,          # Narrower boxes
    'linewidth': 1.5,
}

# Statistical significance thresholds
SIGNIFICANCE_LEVELS = {
    'p < 0.001': '***',
    'p < 0.01': '**',
    'p < 0.05': '*',
    'p >= 0.05': 'ns'
}

# ============================================================================
# STATISTICAL TESTING FUNCTIONS
# ============================================================================

def test_normality(data, alpha=0.05):
    """
    Test normality using Shapiro-Wilk test
    Returns: (is_normal, p_value)
    """
    if len(data) < 3:
        return False, np.nan
    
    try:
        stat, p_value = stats.shapiro(data)
        is_normal = p_value > alpha
        return is_normal, p_value
    except:
        return False, np.nan

def compare_groups(group1, group2, alpha=0.05):
    """
    Compare two groups using appropriate test:
    - T-test if both groups are normally distributed
    - Mann-Whitney U test otherwise
    
    Returns: (test_name, statistic, p_value, significance_marker)
    """
    # Remove NaN values
    group1 = group1[~np.isnan(group1)]
    group2 = group2[~np.isnan(group2)]
    
    if len(group1) < 3 or len(group2) < 3:
        return "Insufficient data", np.nan, np.nan, "ns"
    
    # Test normality for both groups
    normal1, p_norm1 = test_normality(group1)
    normal2, p_norm2 = test_normality(group2)
    
    # Choose appropriate test
    if normal1 and normal2:
        # Use independent samples t-test
        test_name = "T-test"
        statistic, p_value = stats.ttest_ind(group1, group2)
    else:
        # Use Mann-Whitney U test
        test_name = "Mann-Whitney U"
        statistic, p_value = stats.mannwhitneyu(group1, group2, alternative='two-sided')
    
    # Determine significance marker
    if p_value < 0.001:
        marker = '***'
    elif p_value < 0.01:
        marker = '**'
    elif p_value < 0.05:
        marker = '*'
    else:
        marker = 'ns'
    
    return test_name, statistic, p_value, marker

def get_statistical_results(df, variables):
    """
    Compute statistical tests for all variables comparing GAG vs GRV
    Returns: Dictionary with test results for each variable
    """
    results = {}
    
    print("\n" + "="*80)
    print("STATISTICAL TESTING: GAG vs GRV")
    print("="*80)
    
    for var in variables:
        if var not in df.columns or 'irrigation_type' not in df.columns:
            continue
        
        data_gag = df[df['irrigation_type'] == 'GAG'][var].dropna().values
        data_grv = df[df['irrigation_type'] == 'GRV'][var].dropna().values
        
        if len(data_gag) > 0 and len(data_grv) > 0:
            test_name, statistic, p_value, marker = compare_groups(data_gag, data_grv)
            
            results[var] = {
                'test': test_name,
                'statistic': statistic,
                'p_value': p_value,
                'marker': marker,
                'n_gag': len(data_gag),
                'n_grv': len(data_grv),
                'mean_gag': np.mean(data_gag),
                'mean_grv': np.mean(data_grv)
            }
            
            print(f"\n{VARIABLE_NAMES[var]}:")
            print(f"  Test: {test_name}")
            print(f"  GAG: n={len(data_gag)}, mean={np.mean(data_gag):.2f}")
            print(f"  GRV: n={len(data_grv)}, mean={np.mean(data_grv):.2f}")
            print(f"  Statistic: {statistic:.4f}")
            print(f"  p-value: {p_value:.4f}")
            print(f"  Significance: {marker}")
    
    return results

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def add_significance_marker(ax, marker, y_pos=None):
    """
    Add significance marker at the top center inside the plot
    """
    if y_pos is None:
        ylim = ax.get_ylim()
        y_pos = ylim[1] - (ylim[1] - ylim[0]) * 0.08
    
    ax.text(1.0, y_pos, marker, 
            ha='center', va='top',
            fontsize=16, fontweight='bold')

def create_single_boxplot(ax, data, color, box_width, settings, ylabel, title):
    """Create a single boxplot for one variable"""
    
    bp = ax.boxplot([data], positions=[1], widths=box_width,
                    patch_artist=True,
                    showmeans=True,
                    meanprops=dict(marker='D', markerfacecolor='white', 
                                 markeredgecolor='black', markersize=6),
                    medianprops=dict(color='black', linewidth=2),
                    boxprops=dict(facecolor=color, edgecolor='black', 
                                linewidth=settings['linewidth']),
                    whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                    capprops=dict(color='black', linewidth=settings['linewidth']),
                    flierprops=dict(marker='o', markerfacecolor='gray', 
                                  markersize=4, alpha=0.5))
    
    ax.set_xlim(0.5, 1.5)
    ax.set_xticks([1])
    ax.set_xticklabels([''], fontsize=settings['tick_fontsize'])
    ax.set_ylabel(ylabel, fontsize=settings['axis_label_fontsize'], fontweight='bold')
    ax.set_title(title, fontsize=settings['title_fontsize'], fontweight='bold', pad=8)
    
    ax.grid(True, alpha=0.25, linestyle=':', linewidth=0.6, axis='y', zorder=1)
    ax.set_axisbelow(True)
    
    ax.tick_params(axis='both', which='major', labelsize=settings['tick_fontsize'], 
                   width=1.5, length=5)
    
    for spine in ax.spines.values():
        spine.set_linewidth(1.5)
        spine.set_color('black')
    
    return bp

def create_split_boxplot(ax, data_gag, data_grv, colors, box_width, settings, 
                         ylabel, title, significance_marker=None):
    """Create boxplot split by irrigation type with optional significance marker"""
    
    # Adjust positions to center boxes better in the plot area
    positions = [0.7, 1.3]
    
    bp_gag = ax.boxplot([data_gag], positions=[positions[0]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['GAG'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    bp_grv = ax.boxplot([data_grv], positions=[positions[1]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['GRV'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    # Adjust x-axis limits to give more space on sides
    ax.set_xlim(0.4, 1.6)
    ax.set_xticks([1])
    ax.set_xticklabels([''], fontsize=settings['tick_fontsize'])
    ax.set_ylabel(ylabel, fontsize=settings['axis_label_fontsize'], fontweight='bold')
    ax.set_title(title, fontsize=settings['title_fontsize'], fontweight='bold', pad=8)
    
    ax.grid(True, alpha=0.25, linestyle=':', linewidth=0.6, axis='y', zorder=1)
    ax.set_axisbelow(True)
    
    ax.tick_params(axis='both', which='major', labelsize=settings['tick_fontsize'], 
                   width=1.5, length=5)
    
    for spine in ax.spines.values():
        spine.set_linewidth(1.5)
        spine.set_color('black')
    
    # Add significance marker if provided
    if significance_marker:
        add_significance_marker(ax, significance_marker)
    
    return bp_gag, bp_grv

# ============================================================================
# MAIN PLOTTING FUNCTIONS
# ============================================================================

def create_overall_boxplots(df, output_dir, settings, dpi=600):
    """Create overall boxplots (all data combined) - 6 subplots with equal width"""
    
    print("\n" + "="*80)
    print("CREATING OVERALL BOXPLOTS (ALL DATA)")
    print("="*80)
    
    plt.style.use('seaborn-v0_8-whitegrid')
    
    n_vars = len(VARIABLES_ORDER)
    fig_width = n_vars * settings['subplot_width']
    fig_height = settings['subplot_height']
    
    fig, axes = plt.subplots(1, n_vars, figsize=(fig_width, fig_height))
    
    for idx, var in enumerate(VARIABLES_ORDER):
        if var not in df.columns:
            print(f"⚠ {var} not found, skipping...")
            continue
        
        data = df[var].dropna().values
        
        if len(data) > 0:
            create_single_boxplot(
                axes[idx], 
                data, 
                COLORS['overall'], 
                settings['box_width'], 
                settings, 
                VARIABLE_YLABELS[var],
                VARIABLE_NAMES[var]
            )
            print(f"✓ {VARIABLE_NAMES[var]} created")
    
    fig.suptitle('Sugarcane Variables Distribution - Overall', 
                fontsize=settings['title_fontsize'] + 3, fontweight='bold', y=0.98)
    
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    
    output_file = os.path.join(output_dir, 'boxplots_overall.png')
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    print(f"\n✓ Saved: {output_file}")
    
    pdf_file = output_file.replace('.png', '.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    print(f"✓ Saved: {pdf_file}")
    
    plt.close()
    
    return [output_file, pdf_file]

def create_irrigation_boxplots(df, output_dir, settings, stat_results, dpi=600):
    """Create boxplots split by irrigation type with significance markers"""
    
    print("\n" + "="*80)
    print("CREATING BOXPLOTS BY IRRIGATION TYPE (GAG vs GRV) WITH STATS")
    print("="*80)
    
    if 'irrigation_type' not in df.columns:
        print("⚠ irrigation_type column not found!")
        return []
    
    plt.style.use('seaborn-v0_8-whitegrid')
    
    n_vars = len(VARIABLES_ORDER)
    fig_width = n_vars * settings['subplot_width']
    fig_height = settings['subplot_height']
    
    fig, axes = plt.subplots(1, n_vars, figsize=(fig_width, fig_height))
    
    for idx, var in enumerate(VARIABLES_ORDER):
        if var not in df.columns:
            print(f"⚠ {var} not found, skipping...")
            continue
        
        data_gag = df[df['irrigation_type'] == 'GAG'][var].dropna().values
        data_grv = df[df['irrigation_type'] == 'GRV'][var].dropna().values
        
        # Get significance marker for this variable
        marker = stat_results.get(var, {}).get('marker', None)
        
        if len(data_gag) > 0 and len(data_grv) > 0:
            create_split_boxplot(
                axes[idx], 
                data_gag, 
                data_grv, 
                COLORS, 
                settings['box_width'], 
                settings, 
                VARIABLE_YLABELS[var],
                VARIABLE_NAMES[var],
                significance_marker=marker
            )
            print(f"✓ {VARIABLE_NAMES[var]} created (significance: {marker})")
    
    from matplotlib.patches import Patch
    legend_elements = [
        Patch(facecolor=COLORS['GAG'], edgecolor='black', label='GAG'),
        Patch(facecolor=COLORS['GRV'], edgecolor='black', label='GRV')
    ]
    
    fig.legend(handles=legend_elements, 
              loc='lower center', 
              ncol=2, 
              fontsize=settings['axis_label_fontsize'],
              frameon=False,
              bbox_to_anchor=(0.5, -0.02))
    
    fig.suptitle('Sugarcane Variables Distribution - By Irrigation Type', 
                fontsize=settings['title_fontsize'] + 3, fontweight='bold', y=0.98)
    
    plt.tight_layout(rect=[0, 0.03, 1, 0.96])
    
    output_file = os.path.join(output_dir, 'boxplots_by_irrigation.png')
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    print(f"\n✓ Saved: {output_file}")
    
    pdf_file = output_file.replace('.png', '.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    print(f"✓ Saved: {pdf_file}")
    
    plt.close()
    
    return [output_file, pdf_file]

def save_statistical_results(stat_results, output_dir):
    """Save statistical results to CSV file"""
    
    results_list = []
    for var, results in stat_results.items():
        results_list.append({
            'Variable': VARIABLE_NAMES[var],
            'Test': results['test'],
            'Statistic': results['statistic'],
            'P-value': results['p_value'],
            'Significance': results['marker'],
            'n_GAG': results['n_gag'],
            'n_GRV': results['n_grv'],
            'Mean_GAG': results['mean_gag'],
            'Mean_GRV': results['mean_grv']
        })
    
    results_df = pd.DataFrame(results_list)
    output_file = os.path.join(output_dir, 'statistical_results.csv')
    results_df.to_csv(output_file, index=False)
    print(f"\n✓ Statistical results saved: {output_file}")
    
    return output_file

# ============================================================================
# MAIN EXECUTION
# ============================================================================

if __name__ == "__main__":
    
    print("="*80)
    print("SUGARCANE VARIABLES BOXPLOTS")
    print("="*80)
    
    print(f"\nLoading data from: {DATA_FILE}")
    df = pd.read_csv(DATA_FILE)
    
    print(f"Total observations: {len(df)}")
    
    os.makedirs(OUTPUT_DIR, exist_ok=True)
    
    # Perform statistical tests
    stat_results = get_statistical_results(df, VARIABLES_ORDER)
    
    # Save statistical results
    stats_file = save_statistical_results(stat_results, OUTPUT_DIR)
    
    # Create overall boxplots
    overall_files = create_overall_boxplots(df, OUTPUT_DIR, NATURE_SETTINGS, OUTPUT_DPI)
    
    # Create irrigation type boxplots with significance markers
    irrigation_files = create_irrigation_boxplots(df, OUTPUT_DIR, NATURE_SETTINGS, 
                                                  stat_results, OUTPUT_DPI)
    
```

```python
"""
SUGARCANE VARIABLES BOXPLOTS
=============================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from scipy import stats
import warnings
import os
warnings.filterwarnings('ignore')

plt.rcParams['text.usetex'] = False
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['mathtext.default'] = 'regular'

# ============================================================================
# CONFIGURATION
# ============================================================================

DATA_FILE = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/final_dataset_regrouped_categorical.csv'

OUTPUT_DIR = 'sugarcane_boxplots_nature'
OUTPUT_DPI = 600

# Variables in order
VARIABLES_ORDER = [
    'tc_per_ha_actual',
    'brix_percent',
    'hugot_sugar',
    'purity_percent',
    'estimated_recoverable_pol',
    'total_recoverable_sugar_t_per_ha'
]

VARIABLE_NAMES_EN = {
    'tc_per_ha_actual': r'Cane Yield (t $ha^{-1}$)',
    'brix_percent': 'Brix (%)',
    'hugot_sugar': 'Saccharose (%)',
    'purity_percent': 'Purity (%)',
    'estimated_recoverable_pol': 'PER (%)',
    'total_recoverable_sugar_t_per_ha': r'TPER (t $ha^{-1}$)'
}

VARIABLE_NAMES_FR = {
    'tc_per_ha_actual': r'Rendement en Canne (t $ha^{-1}$)',
    'brix_percent': 'Brix (%)',
    'hugot_sugar': 'Saccharose (%)',
    'purity_percent': 'Pureté (%)',
    'estimated_recoverable_pol': 'PER (%)',
    'total_recoverable_sugar_t_per_ha': r'TPER (t $ha^{-1}$)'
}

VARIABLE_YLABELS_EN = {
    'tc_per_ha_actual': r'Cane Yield (t $ha^{-1}$)',
    'brix_percent': 'Brix (%)',
    'hugot_sugar': 'Saccharose (%)',
    'purity_percent': 'Purity (%)',
    'estimated_recoverable_pol': 'PER (%)',
    'total_recoverable_sugar_t_per_ha': r'TPER (t $ha^{-1}$)'
}

VARIABLE_YLABELS_FR = {
    'tc_per_ha_actual': r'Rendement en Canne (t $ha^{-1}$)',
    'brix_percent': 'Brix (%)',
    'hugot_sugar': 'Saccharose (%)',
    'purity_percent': 'Pureté (%)',
    'estimated_recoverable_pol': 'PER (%)',
    'total_recoverable_sugar_t_per_ha': r'TPER (t $ha^{-1}$)'
}

# Text translations
TEXTS = {
    'en': {
        'title_overall': 'Sugarcane Variables Distribution - Overall',
        'title_2box': 'Sugarcane Variables Distribution - By Irrigation Type',
        'title_3box': 'Sugarcane Variables Distribution - Combined, GAG, and GRV',
        'combined': 'Combined',
        'gag': 'GAG',
        'grv': 'GRV'
    },
    'fr': {
        'title_overall': 'Distribution des Variables de la Canne à Sucre - Globale',
        'title_2box': 'Distribution des Variables de la Canne à Sucre - Par Type d\'Irrigation',
        'title_3box': 'Distribution des Variables de la Canne à Sucre - Combiné, GAG, et GRV',
        'combined': 'Combiné',
        'gag': 'GAG',
        'grv': 'GRV'
    }
}

# Color scheme
COLORS = {
    'overall': '#2E86AB',
    'GAG': '#2E86AB',
    'GRV': '#E63946',
    'Combined': '#A06CD5'  # Purple for combined data
}

# Plot settings
NATURE_SETTINGS = {
    'subplot_width': 2.0,       # Equal width for each subplot
    'subplot_height': 5.0,      # Height for all subplots
    'title_fontsize': 13,
    'axis_label_fontsize': 12,
    'tick_fontsize': 11,
    'box_width': 0.35,          # Narrower boxes
    'linewidth': 1.5,
}

# Statistical significance thresholds
SIGNIFICANCE_LEVELS = {
    'p < 0.001': '***',
    'p < 0.01': '**',
    'p < 0.05': '*',
    'p >= 0.05': 'ns'
}

# ============================================================================
# STATISTICAL TESTING FUNCTIONS
# ============================================================================

def test_normality(data, alpha=0.05):
    """
    Test normality using Shapiro-Wilk test
    Returns: (is_normal, p_value)
    """
    if len(data) < 3:
        return False, np.nan
    
    try:
        stat, p_value = stats.shapiro(data)
        is_normal = p_value > alpha
        return is_normal, p_value
    except:
        return False, np.nan

def compare_groups(group1, group2, alpha=0.05):
    """
    Compare two groups using appropriate test:
    - T-test if both groups are normally distributed
    - Mann-Whitney U test otherwise
    
    Returns: (test_name, statistic, p_value, significance_marker)
    """
    # Remove NaN values
    group1 = group1[~np.isnan(group1)]
    group2 = group2[~np.isnan(group2)]
    
    if len(group1) < 3 or len(group2) < 3:
        return "Insufficient data", np.nan, np.nan, "ns"
    
    # Test normality for both groups
    normal1, p_norm1 = test_normality(group1)
    normal2, p_norm2 = test_normality(group2)
    
    # Choose appropriate test
    if normal1 and normal2:
        # Use independent samples t-test
        test_name = "T-test"
        statistic, p_value = stats.ttest_ind(group1, group2)
    else:
        # Use Mann-Whitney U test
        test_name = "Mann-Whitney U"
        statistic, p_value = stats.mannwhitneyu(group1, group2, alternative='two-sided')
    
    # Determine significance marker
    if p_value < 0.001:
        marker = '***'
    elif p_value < 0.01:
        marker = '**'
    elif p_value < 0.05:
        marker = '*'
    else:
        marker = 'ns'
    
    return test_name, statistic, p_value, marker

def get_statistical_results(df, variables):
    """
    Compute statistical tests for all variables comparing GAG vs GRV
    Returns: Dictionary with test results for each variable
    """
    results = {}
    
    print("\n" + "="*80)
    print("STATISTICAL TESTING: GAG vs GRV")
    print("="*80)
    
    for var in variables:
        if var not in df.columns or 'irrigation_type' not in df.columns:
            continue
        
        data_gag = df[df['irrigation_type'] == 'GAG'][var].dropna().values
        data_grv = df[df['irrigation_type'] == 'GRV'][var].dropna().values
        
        if len(data_gag) > 0 and len(data_grv) > 0:
            test_name, statistic, p_value, marker = compare_groups(data_gag, data_grv)
            
            results[var] = {
                'test': test_name,
                'statistic': statistic,
                'p_value': p_value,
                'marker': marker,
                'n_gag': len(data_gag),
                'n_grv': len(data_grv),
                'mean_gag': np.mean(data_gag),
                'mean_grv': np.mean(data_grv)
            }
            
            print(f"\n{VARIABLE_NAMES_EN[var]}:")
            print(f"  Test: {test_name}")
            print(f"  GAG: n={len(data_gag)}, mean={np.mean(data_gag):.2f}")
            print(f"  GRV: n={len(data_grv)}, mean={np.mean(data_grv):.2f}")
            print(f"  Statistic: {statistic:.4f}")
            print(f"  p-value: {p_value:.4f}")
            print(f"  Significance: {marker}")
    
    return results

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def add_significance_marker(ax, marker, y_pos=None):
    """
    Add significance marker at the top center inside the plot
    """
    if y_pos is None:
        ylim = ax.get_ylim()
        y_pos = ylim[1] - (ylim[1] - ylim[0]) * 0.08
    
    ax.text(1.0, y_pos, marker, 
            ha='center', va='top',
            fontsize=16, fontweight='bold')

def create_single_boxplot(ax, data, color, box_width, settings, ylabel, title):
    """Create a single boxplot for one variable"""
    
    bp = ax.boxplot([data], positions=[1], widths=box_width,
                    patch_artist=True,
                    showmeans=True,
                    meanprops=dict(marker='D', markerfacecolor='white', 
                                 markeredgecolor='black', markersize=6),
                    medianprops=dict(color='black', linewidth=2),
                    boxprops=dict(facecolor=color, edgecolor='black', 
                                linewidth=settings['linewidth']),
                    whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                    capprops=dict(color='black', linewidth=settings['linewidth']),
                    flierprops=dict(marker='o', markerfacecolor='gray', 
                                  markersize=4, alpha=0.5))
    
    ax.set_xlim(0.5, 1.5)
    ax.set_xticks([1])
    ax.set_xticklabels([''], fontsize=settings['tick_fontsize'])
    ax.set_ylabel(ylabel, fontsize=settings['axis_label_fontsize'], fontweight='bold')
    ax.set_title(title, fontsize=settings['title_fontsize'], fontweight='bold', pad=8)
    
    ax.grid(True, alpha=0.25, linestyle=':', linewidth=0.6, axis='y', zorder=1)
    ax.set_axisbelow(True)
    
    ax.tick_params(axis='both', which='major', labelsize=settings['tick_fontsize'], 
                   width=1.5, length=5)
    
    for spine in ax.spines.values():
        spine.set_linewidth(1.5)
        spine.set_color('black')
    
    return bp

def create_split_boxplot(ax, data_gag, data_grv, colors, box_width, settings, 
                         ylabel, title, significance_marker=None):
    """Create boxplot with GAG and GRV only (2-box version)"""
    
    # Two positions: GAG (left), GRV (right)
    positions = [0.7, 1.3]
    
    # GAG boxplot
    bp_gag = ax.boxplot([data_gag], positions=[positions[0]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['GAG'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    # GRV boxplot
    bp_grv = ax.boxplot([data_grv], positions=[positions[1]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['GRV'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    # Adjust x-axis limits to give more space on sides
    ax.set_xlim(0.4, 1.6)
    ax.set_xticks([1])
    ax.set_xticklabels([''], fontsize=settings['tick_fontsize'])
    ax.set_ylabel(ylabel, fontsize=settings['axis_label_fontsize'], fontweight='bold')
    ax.set_title(title, fontsize=settings['title_fontsize'], fontweight='bold', pad=8)
    
    ax.grid(True, alpha=0.25, linestyle=':', linewidth=0.6, axis='y', zorder=1)
    ax.set_axisbelow(True)
    
    ax.tick_params(axis='both', which='major', labelsize=settings['tick_fontsize'], 
                   width=1.5, length=5)
    
    for spine in ax.spines.values():
        spine.set_linewidth(1.5)
        spine.set_color('black')
    
    # Add significance marker if provided
    if significance_marker:
        add_significance_marker(ax, significance_marker)
    
    return bp_gag, bp_grv

def create_triple_boxplot(ax, data_gag, data_grv, data_combined, colors, box_width, 
                          settings, ylabel, title, significance_marker=None):
    """Create boxplot with Combined, GAG, and GRV data (in that order)"""
    
    # Three positions: Combined (left), GAG (center), GRV (right)
    # Each centered in its own third of the plot
    positions = [0.5, 1.0, 1.5]  # Evenly spaced thirds
    
    # Combined boxplot (left third)
    bp_combined = ax.boxplot([data_combined], positions=[positions[0]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['Combined'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    # GAG boxplot (center third)
    bp_gag = ax.boxplot([data_gag], positions=[positions[1]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['GAG'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    # GRV boxplot (right third)
    bp_grv = ax.boxplot([data_grv], positions=[positions[2]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['GRV'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    # Set x-axis limits to show three equal thirds
    ax.set_xlim(0.25, 1.75)
    ax.set_xticks([1])
    ax.set_xticklabels([''], fontsize=settings['tick_fontsize'])
    ax.set_ylabel(ylabel, fontsize=settings['axis_label_fontsize'], fontweight='bold')
    ax.set_title(title, fontsize=settings['title_fontsize'], fontweight='bold', pad=8)
    
    ax.grid(True, alpha=0.25, linestyle=':', linewidth=0.6, axis='y', zorder=1)
    ax.set_axisbelow(True)
    
    ax.tick_params(axis='both', which='major', labelsize=settings['tick_fontsize'], 
                   width=1.5, length=5)
    
    for spine in ax.spines.values():
        spine.set_linewidth(1.5)
        spine.set_color('black')
    
    # Add significance marker between GAG (center) and GRV (right) only
    if significance_marker:
        ylim = ax.get_ylim()
        y_pos = ylim[1] - (ylim[1] - ylim[0]) * 0.08
        # Position between GAG (position 1.0) and GRV (position 1.5)
        x_pos = (positions[1] + positions[2]) / 2  # 1.25
        ax.text(x_pos, y_pos, significance_marker, 
                ha='center', va='top',
                fontsize=16, fontweight='bold')
    
    return bp_combined, bp_gag, bp_grv
    """Create boxplot split by irrigation type with optional significance marker"""
    
    # Adjust positions to center boxes better in the plot area
    positions = [0.7, 1.3]
    
    bp_gag = ax.boxplot([data_gag], positions=[positions[0]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['GAG'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    bp_grv = ax.boxplot([data_grv], positions=[positions[1]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['GRV'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    # Adjust x-axis limits to give more space on sides
    ax.set_xlim(0.4, 1.6)
    ax.set_xticks([1])
    ax.set_xticklabels([''], fontsize=settings['tick_fontsize'])
    ax.set_ylabel(ylabel, fontsize=settings['axis_label_fontsize'], fontweight='bold')
    ax.set_title(title, fontsize=settings['title_fontsize'], fontweight='bold', pad=8)
    
    ax.grid(True, alpha=0.25, linestyle=':', linewidth=0.6, axis='y', zorder=1)
    ax.set_axisbelow(True)
    
    ax.tick_params(axis='both', which='major', labelsize=settings['tick_fontsize'], 
                   width=1.5, length=5)
    
    for spine in ax.spines.values():
        spine.set_linewidth(1.5)
        spine.set_color('black')
    
    # Add significance marker if provided
    if significance_marker:
        add_significance_marker(ax, significance_marker)
    
    return bp_gag, bp_grv

# ============================================================================
# MAIN PLOTTING FUNCTIONS
# ============================================================================

def create_overall_boxplots(df, output_dir, settings, language='en', dpi=600):
    """Create overall boxplots (all data combined) - 6 subplots with equal width"""
    
    var_names = VARIABLE_NAMES_EN if language == 'en' else VARIABLE_NAMES_FR
    var_ylabels = VARIABLE_YLABELS_EN if language == 'en' else VARIABLE_YLABELS_FR
    texts = TEXTS[language]
    
    print("\n" + "="*80)
    print(f"CREATING OVERALL BOXPLOTS (ALL DATA) - {language.upper()}")
    print("="*80)
    
    plt.style.use('seaborn-v0_8-whitegrid')
    
    n_vars = len(VARIABLES_ORDER)
    fig_width = n_vars * settings['subplot_width']
    fig_height = settings['subplot_height']
    
    fig, axes = plt.subplots(1, n_vars, figsize=(fig_width, fig_height))
    
    for idx, var in enumerate(VARIABLES_ORDER):
        if var not in df.columns:
            print(f"⚠ {var} not found, skipping...")
            continue
        
        data = df[var].dropna().values
        
        if len(data) > 0:
            create_single_boxplot(
                axes[idx], 
                data, 
                COLORS['overall'], 
                settings['box_width'], 
                settings, 
                var_ylabels[var],
                var_names[var]
            )
            print(f"✓ {var_names[var]} created")
    
    fig.suptitle(texts['title_overall'], 
                fontsize=settings['title_fontsize'] + 3, fontweight='bold', y=0.98)
    
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    
    suffix = '_en' if language == 'en' else '_fr'
    output_file = os.path.join(output_dir, f'boxplots_overall{suffix}.png')
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    print(f"\n✓ Saved: {output_file}")
    
    pdf_file = output_file.replace('.png', '.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    print(f"✓ Saved: {pdf_file}")
    
    plt.close()
    
    return [output_file, pdf_file]

def create_irrigation_boxplots_2box(df, output_dir, settings, stat_results, language='en', dpi=600):
    """Create 2-box version: GAG vs GRV only with significance markers"""
    
    var_names = VARIABLE_NAMES_EN if language == 'en' else VARIABLE_NAMES_FR
    var_ylabels = VARIABLE_YLABELS_EN if language == 'en' else VARIABLE_YLABELS_FR
    texts = TEXTS[language]
    
    print("\n" + "="*80)
    print(f"CREATING 2-BOX BOXPLOTS: GAG vs GRV WITH STATS - {language.upper()}")
    print("="*80)
    
    if 'irrigation_type' not in df.columns:
        print("⚠ irrigation_type column not found!")
        return []
    
    plt.style.use('seaborn-v0_8-whitegrid')
    
    n_vars = len(VARIABLES_ORDER)
    fig_width = n_vars * settings['subplot_width']
    fig_height = settings['subplot_height']
    
    fig, axes = plt.subplots(1, n_vars, figsize=(fig_width, fig_height))
    
    for idx, var in enumerate(VARIABLES_ORDER):
        if var not in df.columns:
            print(f"⚠ {var} not found, skipping...")
            continue
        
        data_gag = df[df['irrigation_type'] == 'GAG'][var].dropna().values
        data_grv = df[df['irrigation_type'] == 'GRV'][var].dropna().values
        
        marker = stat_results.get(var, {}).get('marker', None)
        
        if len(data_gag) > 0 and len(data_grv) > 0:
            create_split_boxplot(
                axes[idx], 
                data_gag, 
                data_grv, 
                COLORS, 
                settings['box_width'], 
                settings, 
                var_ylabels[var],
                var_names[var],
                significance_marker=marker
            )
            print(f"✓ {var_names[var]} created (GAG vs GRV significance: {marker})")
    
    from matplotlib.patches import Patch
    legend_elements = [
        Patch(facecolor=COLORS['GAG'], edgecolor='black', label=texts['gag']),
        Patch(facecolor=COLORS['GRV'], edgecolor='black', label=texts['grv'])
    ]
    
    fig.legend(handles=legend_elements, 
              loc='lower center', 
              ncol=2, 
              fontsize=settings['axis_label_fontsize'],
              frameon=False,
              bbox_to_anchor=(0.5, -0.02))
    
    fig.suptitle(texts['title_2box'], 
                fontsize=settings['title_fontsize'] + 3, fontweight='bold', y=0.98)
    
    plt.tight_layout(rect=[0, 0.03, 1, 0.96])
    
    suffix = '_en' if language == 'en' else '_fr'
    output_file = os.path.join(output_dir, f'boxplots_2box{suffix}.png')
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    print(f"\n✓ Saved: {output_file}")
    
    pdf_file = output_file.replace('.png', '.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    print(f"✓ Saved: {pdf_file}")
    
    plt.close()
    
    return [output_file, pdf_file]

def create_irrigation_boxplots(df, output_dir, settings, stat_results, language='en', dpi=600):
    """Create 3-box version: Combined, GAG, and GRV (in that order) with significance markers"""
    
    var_names = VARIABLE_NAMES_EN if language == 'en' else VARIABLE_NAMES_FR
    var_ylabels = VARIABLE_YLABELS_EN if language == 'en' else VARIABLE_YLABELS_FR
    texts = TEXTS[language]
    
    print("\n" + "="*80)
    print(f"CREATING 3-BOX BOXPLOTS: COMBINED, GAG, GRV WITH STATS - {language.upper()}")
    print("="*80)
    
    if 'irrigation_type' not in df.columns:
        print("⚠ irrigation_type column not found!")
        return []
    
    plt.style.use('seaborn-v0_8-whitegrid')
    
    n_vars = len(VARIABLES_ORDER)
    fig_width = n_vars * settings['subplot_width']
    fig_height = settings['subplot_height']
    
    fig, axes = plt.subplots(1, n_vars, figsize=(fig_width, fig_height))
    
    # Reduce box width to fit three boxes
    triple_box_width = settings['box_width'] * 0.75
    
    for idx, var in enumerate(VARIABLES_ORDER):
        if var not in df.columns:
            print(f"⚠ {var} not found, skipping...")
            continue
        
        data_gag = df[df['irrigation_type'] == 'GAG'][var].dropna().values
        data_grv = df[df['irrigation_type'] == 'GRV'][var].dropna().values
        data_combined = df[var].dropna().values
        
        # Get significance marker for this variable (GAG vs GRV comparison)
        marker = stat_results.get(var, {}).get('marker', None)
        
        if len(data_gag) > 0 and len(data_grv) > 0 and len(data_combined) > 0:
            create_triple_boxplot(
                axes[idx], 
                data_gag, 
                data_grv,
                data_combined,
                COLORS, 
                triple_box_width, 
                settings, 
                var_ylabels[var],
                var_names[var],
                significance_marker=marker
            )
            print(f"✓ {var_names[var]} created (GAG vs GRV significance: {marker})")
    
    from matplotlib.patches import Patch
    legend_elements = [
        Patch(facecolor=COLORS['Combined'], edgecolor='black', label=texts['combined']),
        Patch(facecolor=COLORS['GAG'], edgecolor='black', label=texts['gag']),
        Patch(facecolor=COLORS['GRV'], edgecolor='black', label=texts['grv'])
    ]
    
    fig.legend(handles=legend_elements, 
              loc='lower center', 
              ncol=3, 
              fontsize=settings['axis_label_fontsize'],
              frameon=False,
              bbox_to_anchor=(0.5, -0.02))
    
    fig.suptitle(texts['title_3box'], 
                fontsize=settings['title_fontsize'] + 3, fontweight='bold', y=0.98)
    
    plt.tight_layout(rect=[0, 0.03, 1, 0.96])
    
    suffix = '_en' if language == 'en' else '_fr'
    output_file = os.path.join(output_dir, f'boxplots_3box{suffix}.png')
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    print(f"\n✓ Saved: {output_file}")
    
    pdf_file = output_file.replace('.png', '.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    print(f"✓ Saved: {pdf_file}")
    
    plt.close()
    
    return [output_file, pdf_file]

def save_statistical_results(stat_results, output_dir):
    """Save statistical results to CSV file"""
    
    results_list = []
    for var, results in stat_results.items():
        results_list.append({
            'Variable': VARIABLE_NAMES_EN[var],
            'Test': results['test'],
            'Statistic': results['statistic'],
            'P-value': results['p_value'],
            'Significance': results['marker'],
            'n_GAG': results['n_gag'],
            'n_GRV': results['n_grv'],
            'Mean_GAG': results['mean_gag'],
            'Mean_GRV': results['mean_grv']
        })
    
    results_df = pd.DataFrame(results_list)
    output_file = os.path.join(output_dir, 'statistical_results.csv')
    results_df.to_csv(output_file, index=False)
    print(f"\n✓ Statistical results saved: {output_file}")
    
    return output_file

# ============================================================================
# MAIN EXECUTION
# ============================================================================

if __name__ == "__main__":
    
    print("="*80)
    print("SUGARCANE VARIABLES BOXPLOTS - NATURE STYLE WITH STATISTICAL TESTING")
    print("="*80)
    
    print(f"\nLoading data from: {DATA_FILE}")
    df = pd.read_csv(DATA_FILE)
    
    print(f"Total observations: {len(df)}")
    
    os.makedirs(OUTPUT_DIR, exist_ok=True)
    
    # Perform statistical tests (once)
    stat_results = get_statistical_results(df, VARIABLES_ORDER)
    
    # Save statistical results
    stats_file = save_statistical_results(stat_results, OUTPUT_DIR)
    
    # Create all versions
    all_files = []
    
    # 1. Overall boxplots - English
    print("\n" + "="*80)
    print("CREATING OVERALL PLOTS")
    print("="*80)
    overall_files_en = create_overall_boxplots(df, OUTPUT_DIR, NATURE_SETTINGS, 'en', OUTPUT_DPI)
    all_files.extend(overall_files_en)
    
    # 2. Overall boxplots - French
    overall_files_fr = create_overall_boxplots(df, OUTPUT_DIR, NATURE_SETTINGS, 'fr', OUTPUT_DPI)
    all_files.extend(overall_files_fr)
    
    # 3. 2-box version (GAG vs GRV) - English
    print("\n" + "="*80)
    print("CREATING 2-BOX VERSIONS (GAG vs GRV)")
    print("="*80)
    box2_files_en = create_irrigation_boxplots_2box(df, OUTPUT_DIR, NATURE_SETTINGS, 
                                                     stat_results, 'en', OUTPUT_DPI)
    all_files.extend(box2_files_en)
    
    # 4. 2-box version (GAG vs GRV) - French
    box2_files_fr = create_irrigation_boxplots_2box(df, OUTPUT_DIR, NATURE_SETTINGS, 
                                                     stat_results, 'fr', OUTPUT_DPI)
    all_files.extend(box2_files_fr)
    
    # 5. 3-box version (Combined, GAG, GRV) - English
    print("\n" + "="*80)
    print("CREATING 3-BOX VERSIONS (COMBINED, GAG, GRV)")
    print("="*80)
    box3_files_en = create_irrigation_boxplots(df, OUTPUT_DIR, NATURE_SETTINGS, 
                                                stat_results, 'en', OUTPUT_DPI)
    all_files.extend(box3_files_en)
    
    # 6. 3-box version (Combined, GAG, GRV) - French
    box3_files_fr = create_irrigation_boxplots(df, OUTPUT_DIR, NATURE_SETTINGS, 
                                                stat_results, 'fr', OUTPUT_DPI)
    all_files.extend(box3_files_fr)
    
    print("\n" + "="*80)
    print("ALL BOXPLOTS COMPLETE WITH STATISTICAL TESTING")
    print("\nVariables in order:")
    for var in VARIABLES_ORDER:
        print(f"  - {VARIABLE_NAMES_EN[var]}")
    print("\nTotal files created: " + str(len(all_files)))
```

# SUGARCANE VARIABLES BOXPLOTS - 3 VARIABLES (NO STATISTICS)

```python
"""
SUGARCANE VARIABLES BOXPLOTS - 3 VARIABLES (NO STATISTICS)
================================================================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import warnings
import os
warnings.filterwarnings('ignore')

plt.rcParams['text.usetex'] = False
plt.rcParams['font.family'] = 'sans-serif'
plt.rcParams['mathtext.default'] = 'regular'

# ============================================================================
# CONFIGURATION
# ============================================================================

DATA_FILE = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/final_dataset_regrouped_categorical.csv'

OUTPUT_DIR = 'sugarcane_boxplots_3vars'
OUTPUT_DPI = 600

# Only 3 variables
VARIABLES_ORDER = [
    'tc_per_ha_actual',
    'hugot_sugar',
    'total_recoverable_sugar_t_per_ha'
]

VARIABLE_NAMES_EN = {
    'tc_per_ha_actual': r'Cane yield (t ha$^{-1}$)',
    'hugot_sugar': 'Saccharose (%)',
    'total_recoverable_sugar_t_per_ha': r'TPER (t ha$^{-1}$)'
}

VARIABLE_NAMES_FR = {
    'tc_per_ha_actual': r'Rendement canne (t ha$^{-1}$)',
    'hugot_sugar': 'Saccharose (%)',
    'total_recoverable_sugar_t_per_ha': r'TPER (t ha$^{-1}$)'
}

VARIABLE_YLABELS_EN = {
    'tc_per_ha_actual': r'Cane yield (t ha$^{-1}$)',
    'hugot_sugar': 'Saccharose (%)',
    'total_recoverable_sugar_t_per_ha': r'TPER (t ha$^{-1}$)'
}

VARIABLE_YLABELS_FR = {
    'tc_per_ha_actual': r'Rendement canne (t ha$^{-1}$)',
    'hugot_sugar': 'Saccharose (%)',
    'total_recoverable_sugar_t_per_ha': r'TPER (t ha$^{-1}$)'
}

# Text translations
TEXTS = {
    'en': {
        'title_overall': '',
        'title_2box': '',
        'title_3box': '',
        'combined': 'Combined',
        'gag': 'Drip',
        'grv': 'Surface gravity'
    },
    'fr': {
        'title_overall': '',
        'title_2box': '',
        'title_3box': '',
        'combined': 'Combiné',
        'gag': 'Goutte-à-goutte',
        'grv': 'Gravitaire'
    }
}

# Color scheme
COLORS = {
    'overall': '#2E86AB',
    'GAG': '#4A90E2',           # Blue for drip
    'GRV': '#E63946',           # Red for surface gravity
    'Combined': '#A06CD5'       # Purple for combined
}

# Plot settings
NATURE_SETTINGS = {
    'subplot_width': 2.5,       # Wider subplots for 3 variables
    'subplot_height': 5.0,
    'title_fontsize': 14,
    'axis_label_fontsize': 13,
    'tick_fontsize': 12,
    'box_width': 0.35,
    'linewidth': 1.5,
}

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def create_single_boxplot(ax, data, color, box_width, settings, ylabel, title):
    """Create a single boxplot for one variable"""
    
    bp = ax.boxplot([data], positions=[1], widths=box_width,
                    patch_artist=True,
                    showmeans=True,
                    meanprops=dict(marker='D', markerfacecolor='white', 
                                 markeredgecolor='black', markersize=6),
                    medianprops=dict(color='black', linewidth=2),
                    boxprops=dict(facecolor=color, edgecolor='black', 
                                linewidth=settings['linewidth']),
                    whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                    capprops=dict(color='black', linewidth=settings['linewidth']),
                    flierprops=dict(marker='o', markerfacecolor='gray', 
                                  markersize=4, alpha=0.5))
    
    ax.set_xlim(0.5, 1.5)
    ax.set_xticks([1])
    ax.set_xticklabels([''], fontsize=settings['tick_fontsize'])
    ax.set_ylabel(ylabel, fontsize=settings['axis_label_fontsize'], fontweight='600')
    ax.set_title(title, fontsize=settings['title_fontsize'], fontweight='600', pad=8, loc='left')
    
    ax.grid(True, alpha=0.25, linestyle=':', linewidth=0.7, axis='y', zorder=0)
    ax.set_axisbelow(True)
    
    ax.tick_params(axis='both', which='major', labelsize=settings['tick_fontsize'], 
                   width=1.2, length=5, direction='out')
    
    for spine in ['top', 'right']:
        ax.spines[spine].set_visible(False)
    for spine in ['left', 'bottom']:
        ax.spines[spine].set_linewidth(1.2)
        ax.spines[spine].set_color('#333333')
    
    return bp

def create_split_boxplot(ax, data_gag, data_grv, colors, box_width, settings, ylabel, title):
    """Create boxplot with GAG and GRV only (2-box version)"""
    
    positions = [0.7, 1.3]
    
    # Drip boxplot
    bp_gag = ax.boxplot([data_gag], positions=[positions[0]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['GAG'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    # Surface gravity boxplot
    bp_grv = ax.boxplot([data_grv], positions=[positions[1]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['GRV'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    ax.set_xlim(0.4, 1.6)
    ax.set_xticks([1])
    ax.set_xticklabels([''], fontsize=settings['tick_fontsize'])
    ax.set_ylabel(ylabel, fontsize=settings['axis_label_fontsize'], fontweight='600')
    ax.set_title(title, fontsize=settings['title_fontsize'], fontweight='600', pad=8, loc='left')
    
    ax.grid(True, alpha=0.25, linestyle=':', linewidth=0.7, axis='y', zorder=0)
    ax.set_axisbelow(True)
    
    ax.tick_params(axis='both', which='major', labelsize=settings['tick_fontsize'], 
                   width=1.2, length=5, direction='out')
    
    for spine in ['top', 'right']:
        ax.spines[spine].set_visible(False)
    for spine in ['left', 'bottom']:
        ax.spines[spine].set_linewidth(1.2)
        ax.spines[spine].set_color('#333333')
    
    return bp_gag, bp_grv

def create_triple_boxplot(ax, data_gag, data_grv, data_combined, colors, box_width, 
                          settings, ylabel, title):
    """Create boxplot with Combined, GAG, and GRV data (in that order)"""
    
    positions = [0.5, 1.0, 1.5]
    
    # Combined boxplot
    bp_combined = ax.boxplot([data_combined], positions=[positions[0]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['Combined'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    # Drip boxplot
    bp_gag = ax.boxplot([data_gag], positions=[positions[1]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['GAG'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    # Surface gravity boxplot
    bp_grv = ax.boxplot([data_grv], positions=[positions[2]], widths=box_width,
                        patch_artist=True,
                        showmeans=True,
                        meanprops=dict(marker='D', markerfacecolor='white', 
                                     markeredgecolor='black', markersize=5),
                        medianprops=dict(color='black', linewidth=1.5),
                        boxprops=dict(facecolor=colors['GRV'], edgecolor='black', 
                                    linewidth=settings['linewidth'], alpha=0.8),
                        whiskerprops=dict(color='black', linewidth=settings['linewidth']),
                        capprops=dict(color='black', linewidth=settings['linewidth']),
                        flierprops=dict(marker='o', markerfacecolor='gray', 
                                      markersize=3, alpha=0.4))
    
    ax.set_xlim(0.25, 1.75)
    ax.set_xticks([1])
    ax.set_xticklabels([''], fontsize=settings['tick_fontsize'])
    ax.set_ylabel(ylabel, fontsize=settings['axis_label_fontsize'], fontweight='600')
    ax.set_title(title, fontsize=settings['title_fontsize'], fontweight='600', pad=8, loc='left')
    
    ax.grid(True, alpha=0.25, linestyle=':', linewidth=0.7, axis='y', zorder=0)
    ax.set_axisbelow(True)
    
    ax.tick_params(axis='both', which='major', labelsize=settings['tick_fontsize'], 
                   width=1.2, length=5, direction='out')
    
    for spine in ['top', 'right']:
        ax.spines[spine].set_visible(False)
    for spine in ['left', 'bottom']:
        ax.spines[spine].set_linewidth(1.2)
        ax.spines[spine].set_color('#333333')
    
    return bp_combined, bp_gag, bp_grv

# ============================================================================
# MAIN PLOTTING FUNCTIONS
# ============================================================================

def create_overall_boxplots(df, output_dir, settings, language='en', dpi=600):
    """Create overall boxplots (all data combined) - 3 subplots"""
    
    var_names = VARIABLE_NAMES_EN if language == 'en' else VARIABLE_NAMES_FR
    var_ylabels = VARIABLE_YLABELS_EN if language == 'en' else VARIABLE_YLABELS_FR
    texts = TEXTS[language]
    
    print("\n" + "="*80)
    print(f"CREATING OVERALL BOXPLOTS - {language.upper()}")
    print("="*80)
    
    plt.style.use('seaborn-v0_8-whitegrid')
    
    n_vars = len(VARIABLES_ORDER)
    fig_width = n_vars * settings['subplot_width']
    fig_height = settings['subplot_height']
    
    fig, axes = plt.subplots(1, n_vars, figsize=(fig_width, fig_height), facecolor='white')
    
    for idx, var in enumerate(VARIABLES_ORDER):
        if var not in df.columns:
            print(f"⚠ {var} not found, skipping...")
            continue
        
        data = df[var].dropna().values
        
        if len(data) > 0:
            create_single_boxplot(
                axes[idx], 
                data, 
                COLORS['overall'], 
                settings['box_width'], 
                settings, 
                var_ylabels[var],
                var_names[var]
            )
            print(f"✓ {var_names[var]} created")
    
    fig.suptitle(texts['title_overall'], 
                fontsize=settings['title_fontsize'] + 2, fontweight='600', y=0.98)
    
    plt.tight_layout(rect=[0, 0, 1, 0.96])
    
    suffix = '_en' if language == 'en' else '_fr'
    output_file = os.path.join(output_dir, f'boxplots_overall{suffix}.png')
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', facecolor='white')
    print(f"\n✓ Saved: {output_file}")
    
    pdf_file = output_file.replace('.png', '.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight', facecolor='white')
    print(f"✓ Saved: {pdf_file}")
    
    plt.close()
    
    return [output_file, pdf_file]

def create_irrigation_boxplots_2box(df, output_dir, settings, language='en', dpi=600):
    """Create 2-box version: GAG vs GRV only"""
    
    var_names = VARIABLE_NAMES_EN if language == 'en' else VARIABLE_NAMES_FR
    var_ylabels = VARIABLE_YLABELS_EN if language == 'en' else VARIABLE_YLABELS_FR
    texts = TEXTS[language]
    
    print("\n" + "="*80)
    print(f"CREATING 2-BOX BOXPLOTS - {language.upper()}")
    print("="*80)
    
    if 'irrigation_type' not in df.columns:
        print("⚠ irrigation_type column not found!")
        return []
    
    plt.style.use('seaborn-v0_8-whitegrid')
    
    n_vars = len(VARIABLES_ORDER)
    fig_width = n_vars * settings['subplot_width']
    fig_height = settings['subplot_height']
    
    fig, axes = plt.subplots(1, n_vars, figsize=(fig_width, fig_height), facecolor='white')
    
    for idx, var in enumerate(VARIABLES_ORDER):
        if var not in df.columns:
            print(f"⚠ {var} not found, skipping...")
            continue
        
        data_gag = df[df['irrigation_type'] == 'GAG'][var].dropna().values
        data_grv = df[df['irrigation_type'] == 'GRV'][var].dropna().values
        
        if len(data_gag) > 0 and len(data_grv) > 0:
            create_split_boxplot(
                axes[idx], 
                data_gag, 
                data_grv, 
                COLORS, 
                settings['box_width'], 
                settings, 
                var_ylabels[var],
                var_names[var]
            )
            print(f"✓ {var_names[var]} created")
    
    from matplotlib.patches import Patch
    legend_elements = [
        Patch(facecolor=COLORS['GAG'], edgecolor='black', label=texts['gag']),
        Patch(facecolor=COLORS['GRV'], edgecolor='black', label=texts['grv'])
    ]
    
    fig.legend(handles=legend_elements, 
              loc='lower center', 
              ncol=2, 
              fontsize=settings['axis_label_fontsize'],
              frameon=False,
              bbox_to_anchor=(0.5, -0.02))
    
    fig.suptitle(texts['title_2box'], 
                fontsize=settings['title_fontsize'] + 2, fontweight='600', y=0.98)
    
    plt.tight_layout(rect=[0, 0.03, 1, 0.96])
    
    suffix = '_en' if language == 'en' else '_fr'
    output_file = os.path.join(output_dir, f'boxplots_2box{suffix}.png')
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', facecolor='white')
    print(f"\n✓ Saved: {output_file}")
    
    pdf_file = output_file.replace('.png', '.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight', facecolor='white')
    print(f"✓ Saved: {pdf_file}")
    
    plt.close()
    
    return [output_file, pdf_file]

def create_irrigation_boxplots_3box(df, output_dir, settings, language='en', dpi=600):
    """Create 3-box version: Combined, GAG, and GRV"""
    
    var_names = VARIABLE_NAMES_EN if language == 'en' else VARIABLE_NAMES_FR
    var_ylabels = VARIABLE_YLABELS_EN if language == 'en' else VARIABLE_YLABELS_FR
    texts = TEXTS[language]
    
    print("\n" + "="*80)
    print(f"CREATING 3-BOX BOXPLOTS - {language.upper()}")
    print("="*80)
    
    if 'irrigation_type' not in df.columns:
        print("⚠ irrigation_type column not found!")
        return []
    
    plt.style.use('seaborn-v0_8-whitegrid')
    
    n_vars = len(VARIABLES_ORDER)
    fig_width = n_vars * settings['subplot_width']
    fig_height = settings['subplot_height']
    
    fig, axes = plt.subplots(1, n_vars, figsize=(fig_width, fig_height), facecolor='white')
    
    triple_box_width = settings['box_width'] * 0.75
    
    for idx, var in enumerate(VARIABLES_ORDER):
        if var not in df.columns:
            print(f"⚠ {var} not found, skipping...")
            continue
        
        data_gag = df[df['irrigation_type'] == 'GAG'][var].dropna().values
        data_grv = df[df['irrigation_type'] == 'GRV'][var].dropna().values
        data_combined = df[var].dropna().values
        
        if len(data_gag) > 0 and len(data_grv) > 0 and len(data_combined) > 0:
            create_triple_boxplot(
                axes[idx], 
                data_gag, 
                data_grv,
                data_combined,
                COLORS, 
                triple_box_width, 
                settings, 
                var_ylabels[var],
                var_names[var]
            )
            print(f"✓ {var_names[var]} created")
    
    from matplotlib.patches import Patch
    legend_elements = [
        Patch(facecolor=COLORS['Combined'], edgecolor='black', label=texts['combined']),
        Patch(facecolor=COLORS['GAG'], edgecolor='black', label=texts['gag']),
        Patch(facecolor=COLORS['GRV'], edgecolor='black', label=texts['grv'])
    ]
    
    fig.legend(handles=legend_elements, 
              loc='lower center', 
              ncol=3, 
              fontsize=settings['axis_label_fontsize'],
              frameon=False,
              bbox_to_anchor=(0.5, -0.02))
    
    fig.suptitle(texts['title_3box'], 
                fontsize=settings['title_fontsize'] + 2, fontweight='600', y=0.98)
    
    plt.tight_layout(rect=[0, 0.03, 1, 0.96])
    
    suffix = '_en' if language == 'en' else '_fr'
    output_file = os.path.join(output_dir, f'boxplots_3box{suffix}.png')
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', facecolor='white')
    print(f"\n✓ Saved: {output_file}")
    
    pdf_file = output_file.replace('.png', '.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight', facecolor='white')
    print(f"✓ Saved: {pdf_file}")
    
    plt.close()
    
    return [output_file, pdf_file]

# ============================================================================
# MAIN EXECUTION
# ============================================================================

if __name__ == "__main__":
    
    print("="*80)
    print("SUGARCANE BOXPLOTS - 3 VARIABLES (YIELD, SACCHAROSE, TPER)")
    print("="*80)
    
    print(f"\nLoading data from: {DATA_FILE}")
    df = pd.read_csv(DATA_FILE)
    
    print(f"Total observations: {len(df)}")
    
    os.makedirs(OUTPUT_DIR, exist_ok=True)
    
    # Create all versions
    all_files = []
    
    # Overall boxplots
    overall_files_en = create_overall_boxplots(df, OUTPUT_DIR, NATURE_SETTINGS, 'en', OUTPUT_DPI)
    all_files.extend(overall_files_en)
    
    overall_files_fr = create_overall_boxplots(df, OUTPUT_DIR, NATURE_SETTINGS, 'fr', OUTPUT_DPI)
    all_files.extend(overall_files_fr)
    
    # 2-box version
    box2_files_en = create_irrigation_boxplots_2box(df, OUTPUT_DIR, NATURE_SETTINGS, 'en', OUTPUT_DPI)
    all_files.extend(box2_files_en)
    
    box2_files_fr = create_irrigation_boxplots_2box(df, OUTPUT_DIR, NATURE_SETTINGS, 'fr', OUTPUT_DPI)
    all_files.extend(box2_files_fr)
    
    # 3-box version
    box3_files_en = create_irrigation_boxplots_3box(df, OUTPUT_DIR, NATURE_SETTINGS, 'en', OUTPUT_DPI)
    all_files.extend(box3_files_en)
    
    box3_files_fr = create_irrigation_boxplots_3box(df, OUTPUT_DIR, NATURE_SETTINGS, 'fr', OUTPUT_DPI)
    all_files.extend(box3_files_fr)
    
    print("\n" + "="*80)
    print("ALL BOXPLOTS COMPLETE")
    print("="*80)
    print(f"\nOutput directory: {OUTPUT_DIR}")
    print("\nFiles generated:")
    print("  - boxplots_overall_en/fr.png/pdf")
    print("  - boxplots_2box_en/fr.png/pdf")
    print("  - boxplots_3box_en/fr.png/pdf")
    print("\nVariables:")
    for var in VARIABLES_ORDER:
        print(f"  - {VARIABLE_NAMES_EN[var]}")
    print(f"\nTotal files: {len(all_files)}")
```

# TEMPORAL TRENDS FOR SOIL PROPERTIES BY IRRIGATION TYPE

```python
"""
TEMPORAL TRENDS FOR SOIL PROPERTIES BY IRRIGATION TYPE
=======================================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
import warnings
import os
warnings.filterwarnings('ignore')

# ============================================================================
# CONFIGURATION
# ============================================================================

# Path to your data (ADJUST THIS)
DATA_FILE = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/final_dataset_regrouped_categorical.csv'

# Output configuration
OUTPUT_DIR_BASE = 'soil_properties_temporal_trends'
OUTPUT_DPI = 600

# Soil property groups with display names, units, and custom layouts
SOIL_PROPERTY_GROUPS = {
    'Granulometry': {
        'variables': {
            'Clay_percent_complete': {'name': 'Clay (%)', 'unit': '% year⁻¹'},
            'Fine_Silt_percent_complete': {'name': 'Fine Silt (%)', 'unit': '% year⁻¹'},
            'Coarse_Silt_percent_complete': {'name': 'Coarse Silt (%)', 'unit': '% year⁻¹'},
            'Fine_Sand_percent_complete': {'name': 'Fine Sand (%)', 'unit': '% year⁻¹'},
            'Coarse_Sand_percent_complete': {'name': 'Coarse Sand (%)', 'unit': '% year⁻¹'}
        },
        'output_file': 'granulometry_temporal_trends.png',
        'output_subdir': 'granulometry_individual',
        'layout': {'rows': 2, 'cols': 3}  # 2 rows x 3 columns
    },
    'Exchangeable_Elements': {
        'variables': {
            'Exchangeable_Calcium_meq_100g_complete': {'name': 'Exch. Ca (meq/100g)', 'unit': 'meq/100g year⁻¹'},
            'Exchangeable_Magnesium_meq_100g_complete': {'name': 'Exch. Mg (meq/100g)', 'unit': 'meq/100g year⁻¹'},
            'Exchangeable_Potassium_meq_100g_complete': {'name': 'Exch. K (meq/100g)', 'unit': 'meq/100g year⁻¹'},
            'Exchangeable_Sodium_meq_100g_complete': {'name': 'Exch. Na (meq/100g)', 'unit': 'meq/100g year⁻¹'}
        },
        'output_file': 'exchangeable_elements_temporal_trends.png',
        'output_subdir': 'exchangeable_elements_individual',
        'layout': {'rows': 2, 'cols': 2}  # 2 rows x 2 columns
    },
    'Other_Properties_Part1': {
        'variables': {
            'pH_complete': {'name': 'pH', 'unit': 'pH year⁻¹'},
            'Organic_Carbon_per_mille_complete': {'name': 'Organic C (‰)', 'unit': '‰ year⁻¹'},
            'Total_Nitrogen_per_mille_complete': {'name': 'Total N (‰)', 'unit': '‰ year⁻¹'},
            'Available_P2O5_mg_kg_complete': {'name': 'Available P₂O₅ (mg/kg)', 'unit': 'mg/kg year⁻¹'}
        },
        'output_file': 'other_properties_part1_temporal_trends.png',
        'output_subdir': 'other_properties_individual',
        'layout': {'rows': 2, 'cols': 2},  # 2 rows x 2 columns
        'group_title': 'Autres Propriétés du Sol - Partie 1'
    },
    'Other_Properties_Part2': {
        'variables': {
            'Electrical_Conductivity_mmho_cm_complete': {'name': 'EC (mmho/cm)', 'unit': 'mmho/cm year⁻¹'},
            'Mg_Ca_Ratio_complete': {'name': 'Mg/Ca Ratio', 'unit': 'year⁻¹'},
            'Mg_K_Ratio_complete': {'name': 'Mg/K Ratio', 'unit': 'year⁻¹'},
            'Compaction_Index_complete': {'name': 'Compaction Index', 'unit': 'year⁻¹'}
        },
        'output_file': 'other_properties_part2_temporal_trends.png',
        'output_subdir': 'other_properties_individual',
        'layout': {'rows': 2, 'cols': 2},  # 2 rows x 2 columns
        'group_title': 'Autres Propriétés du Sol - Partie 2'
    }
}

# Visual styling - GAG in blue, GRV in red
COLORS = {
    'GAG': {
        'data_points': '#2E86AB',  # Blue
        'trend_line': '#1E5F8C',
        'error_bars': '#2E86AB',
    },
    'GRV': {
        'data_points': '#E63946',  # Red
        'trend_line': '#C1121F',
        'error_bars': '#E63946',
    }
}

# Grid layout: 4 columns (for full combined plot only)
N_COLS = 4

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def calculate_temporal_statistics(df, response_var, year_column, irrigation_type):
    """
    Calculate temporal statistics for a response variable filtered by irrigation type
    
    Parameters:
    -----------
    df : DataFrame
    response_var : str
    year_column : str
    irrigation_type : str ('GAG' or 'GRV')
    
    Returns:
    --------
    dict with yearly_means, yearly_std, trend_stats, percent_change or None
    """
    # Filter by irrigation type
    df_filtered = df[df['irrigation_type'] == irrigation_type].copy()
    
    # Remove missing values
    df_clean = df_filtered[[year_column, response_var]].dropna()
    
    if len(df_clean) < 3:
        return None
    
    # Calculate yearly statistics
    yearly_stats = df_clean.groupby(year_column)[response_var].agg([
        ('mean', 'mean'),
        ('std', 'std'),
        ('count', 'count')
    ]).reset_index()
    
    years = yearly_stats[year_column].values
    means = yearly_stats['mean'].values
    stds = yearly_stats['std'].values
    
    # Linear regression
    if len(years) >= 3:
        slope, intercept, r_value, p_value, std_err = stats.linregress(years, means)
        
        # Calculate percentage change over entire period
        first_value = means[0]
        last_value = means[-1]
        
        if abs(first_value) > 1e-10:
            percent_change = ((last_value - first_value) / abs(first_value)) * 100
        else:
            percent_change = np.nan
        
        absolute_change = last_value - first_value
        
        trend_stats = {
            'slope': slope,
            'intercept': intercept,
            'r_squared': r_value**2,
            'p_value': p_value,
            'significant': p_value < 0.05,
            'percent_change': percent_change,
            'absolute_change': absolute_change,
            'years_span': years[-1] - years[0],
            'first_value': first_value,
            'last_value': last_value
        }
    else:
        trend_stats = None
    
    return {
        'years': years,
        'means': means,
        'stds': stds,
        'yearly_stats': yearly_stats,
        'trend_stats': trend_stats,
        'irrigation_type': irrigation_type
    }

def format_pvalue(p):
    """Format p-value for display"""
    if p < 0.001:
        return 'p < 0.001'
    elif p < 0.01:
        return f'p < 0.01'
    elif p < 0.05:
        return f'p = {p:.3f}'
    else:
        return f'p = {p:.2f}'

def plot_dual_trends(ax, stats_gag, stats_grv, var_name, var_info, colors):
    """
    Plot temporal trends for both GAG and GRV on the same axis
    
    Parameters:
    -----------
    ax : matplotlib axis
    stats_gag : dict from calculate_temporal_statistics for GAG
    stats_grv : dict from calculate_temporal_statistics for GRV
    var_name : str (variable key)
    var_info : dict with 'name' and 'unit'
    colors : dict
    """
    display_name = var_info['name']
    unit = var_info['unit']
    
    # Plot GAG data (Blue)
    if stats_gag is not None:
        years_gag = stats_gag['years']
        means_gag = stats_gag['means']
        stds_gag = stats_gag['stds']
        
        ax.errorbar(years_gag, means_gag, yerr=stds_gag, 
                    fmt='o', markersize=5, 
                    color=colors['GAG']['data_points'],
                    ecolor=colors['GAG']['error_bars'],
                    elinewidth=1.2,
                    capsize=3,
                    capthick=1.2,
                    alpha=0.8,
                    zorder=3,
                    label='GAG')
        
        ax.plot(years_gag, means_gag, '-', 
                color=colors['GAG']['data_points'], 
                alpha=0.3, 
                linewidth=1,
                zorder=2)
        
        # Plot GAG trend line
        if stats_gag['trend_stats'] is not None:
            trend_line_gag = stats_gag['trend_stats']['slope'] * years_gag + stats_gag['trend_stats']['intercept']
            ax.plot(years_gag, trend_line_gag, '--', 
                    color=colors['GAG']['trend_line'],
                    linewidth=2,
                    alpha=0.9,
                    zorder=4)
    
    # Plot GRV data (Red)
    if stats_grv is not None:
        years_grv = stats_grv['years']
        means_grv = stats_grv['means']
        stds_grv = stats_grv['stds']
        
        ax.errorbar(years_grv, means_grv, yerr=stds_grv, 
                    fmt='s', markersize=5, 
                    color=colors['GRV']['data_points'],
                    ecolor=colors['GRV']['error_bars'],
                    elinewidth=1.2,
                    capsize=3,
                    capthick=1.2,
                    alpha=0.8,
                    zorder=3,
                    label='GRV')
        
        ax.plot(years_grv, means_grv, '-', 
                color=colors['GRV']['data_points'], 
                alpha=0.3, 
                linewidth=1,
                zorder=2)
        
        # Plot GRV trend line
        if stats_grv['trend_stats'] is not None:
            trend_line_grv = stats_grv['trend_stats']['slope'] * years_grv + stats_grv['trend_stats']['intercept']
            ax.plot(years_grv, trend_line_grv, '--', 
                    color=colors['GRV']['trend_line'],
                    linewidth=2,
                    alpha=0.9,
                    zorder=4)
    
    # Add statistics text boxes
    # GAG statistics (top-left)
    if stats_gag is not None and stats_gag['trend_stats'] is not None:
        ts = stats_gag['trend_stats']
        r2_str = f"R² = {ts['r_squared']:.3f}"
        p_val_str = format_pvalue(ts['p_value'])
        pct_change = ts['percent_change']
        
        if not np.isnan(pct_change):
            pct_sign = '+' if pct_change > 0 else ''
            pct_str = f"Δ = {pct_sign}{pct_change:.1f}%"
        else:
            pct_str = "Δ = N/A"
        
        stats_text_gag = f"GAG\n{r2_str}\n{p_val_str}\n{pct_str}"
        
        bbox_props_gag = dict(
            boxstyle='round,pad=0.3',
            facecolor='white',
            edgecolor=colors['GAG']['trend_line'],
            linewidth=1.5,
            alpha=0.95
        )
        
        ax.text(0.05, 0.95, stats_text_gag,
                transform=ax.transAxes,
                fontsize=7.5,
                verticalalignment='top',
                horizontalalignment='left',
                bbox=bbox_props_gag,
                fontweight='bold',
                family='monospace')
    
    # GRV statistics (top-right)
    if stats_grv is not None and stats_grv['trend_stats'] is not None:
        ts = stats_grv['trend_stats']
        r2_str = f"R² = {ts['r_squared']:.3f}"
        p_val_str = format_pvalue(ts['p_value'])
        pct_change = ts['percent_change']
        
        if not np.isnan(pct_change):
            pct_sign = '+' if pct_change > 0 else ''
            pct_str = f"Δ = {pct_sign}{pct_change:.1f}%"
        else:
            pct_str = "Δ = N/A"
        
        stats_text_grv = f"GRV\n{r2_str}\n{p_val_str}\n{pct_str}"
        
        bbox_props_grv = dict(
            boxstyle='round,pad=0.3',
            facecolor='white',
            edgecolor=colors['GRV']['trend_line'],
            linewidth=1.5,
            alpha=0.95
        )
        
        ax.text(0.95, 0.95, stats_text_grv,
                transform=ax.transAxes,
                fontsize=7.5,
                verticalalignment='top',
                horizontalalignment='right',
                bbox=bbox_props_grv,
                fontweight='bold',
                family='monospace')
    
    # Formatting
    ax.set_xlabel('Année', fontsize=10, fontweight='bold')
    ax.set_ylabel(display_name, fontsize=10, fontweight='bold')
    ax.set_title(display_name, fontsize=11, fontweight='bold', pad=8)
    
    # Grid
    ax.grid(True, alpha=0.3, linestyle='--', linewidth=0.5, zorder=1)
    ax.set_axisbelow(True)
    
    # Legend
    ax.legend(loc='lower left', frameon=True, framealpha=0.9, 
              edgecolor='black', fontsize=8, ncol=2)
    
    # Tick parameters
    ax.tick_params(axis='both', which='major', labelsize=9, width=1.2, length=4)
    
    # Set integer years on x-axis
    all_years = []
    if stats_gag is not None:
        all_years.extend(stats_gag['years'])
    if stats_grv is not None:
        all_years.extend(stats_grv['years'])
    
    if len(all_years) > 0:
        unique_years = sorted(set(all_years))
        ax.set_xticks(unique_years)
        ax.set_xticklabels([int(y) for y in unique_years], rotation=45, ha='right')
    
    # Spine formatting
    for spine in ax.spines.values():
        spine.set_linewidth(1.2)
        spine.set_color('black')

# ============================================================================
# FUNCTION TO SAVE INDIVIDUAL PLOTS
# ============================================================================

def save_individual_plot(stats_gag, stats_grv, var_name, var_info, 
                        output_dir, colors, dpi=600):
    """
    Save a standalone plot for a single variable with both irrigation types
    """
    plt.style.use('seaborn-v0_8-white')
    
    # Create figure with single subplot
    fig, ax = plt.subplots(1, 1, figsize=(10, 9))
    
    # Plot the trends
    plot_dual_trends(ax, stats_gag, stats_grv, var_name, var_info, colors)
    
    # Adjust layout
    plt.tight_layout()
    
    # Create safe filename
    safe_filename = var_name.replace('_complete', '').replace('_', '-').replace('/', '-')
    
    # Save PNG
    png_path = os.path.join(output_dir, f'{safe_filename}_GAG_vs_GRV.png')
    plt.savefig(png_path, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    
    # Save PDF
    pdf_path = os.path.join(output_dir, f'{safe_filename}_GAG_vs_GRV.pdf')
    plt.savefig(pdf_path, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    
    plt.close()
    
    return png_path, pdf_path

# ============================================================================
# MAIN PLOTTING FUNCTION
# ============================================================================

def create_full_combined_plot(df, year_column, property_groups, output_dir_base, 
                             dpi=600, colors=COLORS, n_cols=N_COLS):
    """
    Create ONE full combined plot with ALL soil properties (17 variables)
    """
    print("\n" + "="*80)
    print("CREATING FULL COMBINED PLOT WITH ALL SOIL PROPERTIES")
    print("="*80)
    
    # Collect all variables from all groups
    all_variables = {}
    for group_name, group_info in property_groups.items():
        all_variables.update(group_info['variables'])
    
    print(f"\nTotal variables to plot: {len(all_variables)}")
    
    # Calculate statistics for all variables
    print("\nCalculating temporal statistics for all variables...")
    
    all_stats = {}
    
    for var_name, var_info in all_variables.items():
        display_name = var_info['name']
        
        if var_name not in df.columns:
            print(f"  ⚠ Warning: {var_name} not found in dataset, skipping...")
            continue
        
        # Calculate for GAG
        stats_gag = calculate_temporal_statistics(df, var_name, year_column, 'GAG')
        
        # Calculate for GRV
        stats_grv = calculate_temporal_statistics(df, var_name, year_column, 'GRV')
        
        all_stats[var_name] = {
            'GAG': stats_gag,
            'GRV': stats_grv
        }
        
        print(f"  ✓ {display_name}")
    
    # Create figure with subplots (4 columns)
    n_vars = len(all_stats)
    n_rows = int(np.ceil(n_vars / n_cols))
    
    print(f"\nCreating FULL COMBINED {n_rows}×{n_cols} grid plot...")
    
    plt.style.use('seaborn-v0_8-white')
    
    fig_width = 20
    fig_height = n_rows * 4
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(fig_width, fig_height))
    
    # Handle different subplot array shapes
    if n_rows == 1 and n_cols == 1:
        axes = np.array([[axes]])
    elif n_rows == 1:
        axes = axes.reshape(1, -1)
    elif n_cols == 1:
        axes = axes.reshape(-1, 1)
    
    axes_flat = axes.flatten()
    
    # Plot each variable
    for idx, (var_name, var_info) in enumerate(all_variables.items()):
        if var_name not in all_stats:
            continue
        
        ax = axes_flat[idx]
        stats_gag = all_stats[var_name]['GAG']
        stats_grv = all_stats[var_name]['GRV']
        
        plot_dual_trends(ax, stats_gag, stats_grv, var_name, var_info, colors)
    
    # Hide unused subplots
    for idx in range(len(all_stats), len(axes_flat)):
        axes_flat[idx].set_visible(False)
    
    # Overall title
    fig.suptitle('Propriétés du Sol - Évolution Temporelle (GAG vs GRV) - Toutes Variables',
                 fontsize=18, fontweight='bold', y=0.995)
    
    # Adjust layout
    plt.tight_layout(rect=[0, 0.02, 1, 0.99], h_pad=3.0, w_pad=2.5)
    
    # Save full combined figure
    output_file = os.path.join(output_dir_base, 'ALL_soil_properties_combined.png')
    print(f"\nSaving FULL COMBINED figure to: {output_file}")
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    print(f"✓ Combined PNG saved at {dpi} DPI")
    
    # Also save as PDF
    pdf_file = output_file.replace('.png', '.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    print(f"✓ Combined PDF saved: {pdf_file}")
    
    plt.close()
    
    # Summary statistics for ALL variables
    print(f"\n{'='*80}")
    print("SUMMARY STATISTICS - ALL SOIL PROPERTIES")
    print('='*80)
    
    summary_data = []
    for var_name, var_info in all_variables.items():
        if var_name not in all_stats:
            continue
        
        for irr_type in ['GAG', 'GRV']:
            stats_dict = all_stats[var_name][irr_type]
            
            if stats_dict is not None and stats_dict['trend_stats'] is not None:
                trend_stats = stats_dict['trend_stats']
                
                summary_data.append({
                    'Variable': var_info['name'],
                    'Irrigation': irr_type,
                    'Pente': f"{trend_stats['slope']:.4f}",
                    'R²': f"{trend_stats['r_squared']:.3f}",
                    'Valeur-P': f"{trend_stats['p_value']:.4e}",
                    'Significatif': 'Oui' if trend_stats['significant'] else 'Non',
                    'Δ (%)': f"{trend_stats['percent_change']:+.1f}" if not np.isnan(trend_stats['percent_change']) else 'N/A',
                })
    
    if len(summary_data) > 0:
        summary_df = pd.DataFrame(summary_data)
        print(summary_df.to_string(index=False))
        
        # Save summary
        summary_file = output_file.replace('.png', '_summary.csv')
        summary_df.to_csv(summary_file, index=False)
        print(f"\n✓ Full summary statistics saved: {summary_file}")
    
    return output_file, pdf_file, summary_file

def create_soil_properties_plots(data_file, property_groups, output_dir_base, 
                                 dpi=600, colors=COLORS, n_cols=N_COLS):
    """
    Create temporal trends plots for all soil property groups
    separated by irrigation type (GAG and GRV)
    
    Creates:
    1. ONE full combined plot with ALL 17 soil properties
    2. Individual group plots with custom layouts:
       - Granulometry: 2×3 grid
       - Exchangeable Elements: 2×2 grid
       - Other Properties Part 1: 2×2 grid
       - Other Properties Part 2: 2×2 grid
    3. Individual standalone plots for each variable
    """
    print("="*80)
    print("SOIL PROPERTIES TEMPORAL TRENDS (GAG vs GRV)")
    print("="*80)
    
    # Load data
    print(f"\nLoading data from: {data_file}")
    df = pd.read_csv(data_file)
    
    # Find year column
    year_column = None
    for col in ['campaign_year', 'Harvest_Year', 'year', 'Campaign_Year']:
        if col in df.columns:
            year_column = col
            break
    
    if year_column is None:
        raise ValueError("No year column found in dataset!")
    
    # Check irrigation_type column
    if 'irrigation_type' not in df.columns:
        raise ValueError("irrigation_type column not found in dataset!")
    
    irrigation_types = df['irrigation_type'].unique()
    print(f"\nIrrigation types found: {irrigation_types}")
    
    print(f"Year column: {year_column}")
    print(f"Years available: {sorted(df[year_column].unique())}")
    print(f"Total observations: {len(df)}")
    print(f"  GAG: {len(df[df['irrigation_type'] == 'GAG'])}")
    print(f"  GRV: {len(df[df['irrigation_type'] == 'GRV'])}")
    
    # Create base output directory
    os.makedirs(output_dir_base, exist_ok=True)
    print(f"\nOutput directory: {output_dir_base}")
    
    # ========================================================================
    # STEP 1: CREATE FULL COMBINED PLOT WITH ALL SOIL PROPERTIES
    # ========================================================================
    full_combined_files = create_full_combined_plot(
        df, year_column, property_groups, output_dir_base, dpi, colors, n_cols
    )
    
    # ========================================================================
    # STEP 2: CREATE GROUP-SPECIFIC PLOTS WITH CUSTOM LAYOUTS
    # ========================================================================
    
    # Process each property group
    for group_name, group_info in property_groups.items():
        print("\n" + "="*80)
        print(f"PROCESSING GROUP: {group_name}")
        print("="*80)
        
        variables = group_info['variables']
        output_file = os.path.join(output_dir_base, group_info['output_file'])
        output_subdir = os.path.join(output_dir_base, group_info['output_subdir'])
        layout = group_info['layout']
        n_rows = layout['rows']
        n_cols_group = layout['cols']
        
        print(f"Layout: {n_rows} rows × {n_cols_group} columns")
        
        # Create subdirectory for individual plots
        os.makedirs(output_subdir, exist_ok=True)
        
        # Calculate statistics for all variables in this group
        print("\nCalculating temporal statistics...")
        
        all_stats = {}
        
        for var_name, var_info in variables.items():
            display_name = var_info['name']
            
            if var_name not in df.columns:
                print(f"  ⚠ Warning: {var_name} not found in dataset, skipping...")
                continue
            
            # Calculate for GAG
            stats_gag = calculate_temporal_statistics(df, var_name, year_column, 'GAG')
            
            # Calculate for GRV
            stats_grv = calculate_temporal_statistics(df, var_name, year_column, 'GRV')
            
            all_stats[var_name] = {
                'GAG': stats_gag,
                'GRV': stats_grv
            }
            
            print(f"\n  • {display_name}:")
            
            # GAG stats
            if stats_gag is not None and stats_gag['trend_stats'] is not None:
                ts = stats_gag['trend_stats']
                sig = "✓ Sig" if ts['significant'] else "✗ NS"
                print(f"      GAG: {sig} (p={ts['p_value']:.4f}, R²={ts['r_squared']:.3f}, Δ={ts['percent_change']:+.1f}%)")
            else:
                print(f"      GAG: Insufficient data")
            
            # GRV stats
            if stats_grv is not None and stats_grv['trend_stats'] is not None:
                ts = stats_grv['trend_stats']
                sig = "✓ Sig" if ts['significant'] else "✗ NS"
                print(f"      GRV: {sig} (p={ts['p_value']:.4f}, R²={ts['r_squared']:.3f}, Δ={ts['percent_change']:+.1f}%)")
            else:
                print(f"      GRV: Insufficient data")
        
        # Create figure with custom layout
        print(f"\nCreating {n_rows}×{n_cols_group} grid plot...")
        
        plt.style.use('seaborn-v0_8-white')
        
        # Adjust figure size based on layout
        fig_width = n_cols_group * 6.5
        fig_height = n_rows * 4.5
        
        fig, axes = plt.subplots(n_rows, n_cols_group, figsize=(fig_width, fig_height))
        
        # Handle different subplot array shapes
        if n_rows == 1 and n_cols_group == 1:
            axes = np.array([[axes]])
        elif n_rows == 1:
            axes = axes.reshape(1, -1)
        elif n_cols_group == 1:
            axes = axes.reshape(-1, 1)
        
        axes_flat = axes.flatten()
        
        # Plot each variable
        for idx, (var_name, var_info) in enumerate(variables.items()):
            if var_name not in all_stats:
                continue
            
            ax = axes_flat[idx]
            stats_gag = all_stats[var_name]['GAG']
            stats_grv = all_stats[var_name]['GRV']
            
            plot_dual_trends(ax, stats_gag, stats_grv, var_name, var_info, colors)
        
        # Hide unused subplots
        for idx in range(len(all_stats), len(axes_flat)):
            axes_flat[idx].set_visible(False)
        
        # Overall title
        title_map = {
            'Granulometry': 'Granulométrie',
            'Exchangeable_Elements': 'Éléments Échangeables',
            'Other_Properties_Part1': 'Autres Propriétés du Sol - Partie 1',
            'Other_Properties_Part2': 'Autres Propriétés du Sol - Partie 2'
        }
        
        # Use custom title if provided, otherwise use default
        if 'group_title' in group_info:
            plot_title = f"{group_info['group_title']} - GAG vs GRV"
        else:
            plot_title = f'{title_map.get(group_name, group_name)} - GAG vs GRV'
        
        fig.suptitle(plot_title, fontsize=16, fontweight='bold', y=0.995)
        
        # Adjust layout
        plt.tight_layout(rect=[0, 0.02, 1, 0.99], h_pad=3.0, w_pad=2.5)
        
        # Save combined figure
        print(f"\nSaving combined figure to: {output_file}")
        plt.savefig(output_file, dpi=dpi, bbox_inches='tight', 
                    facecolor='white', edgecolor='none')
        print(f"✓ Combined PNG saved at {dpi} DPI")
        
        # Also save as PDF
        pdf_file = output_file.replace('.png', '.pdf')
        plt.savefig(pdf_file, format='pdf', bbox_inches='tight',
                    facecolor='white', edgecolor='none')
        print(f"✓ Combined PDF saved: {pdf_file}")
        
        plt.close()
        
        # Save individual plots
        print(f"\nSaving individual plots to: {output_subdir}")
        
        for var_name, var_info in variables.items():
            if var_name not in all_stats:
                continue
            
            stats_gag = all_stats[var_name]['GAG']
            stats_grv = all_stats[var_name]['GRV']
            display_name = var_info['name']
            
            png_path, pdf_path = save_individual_plot(
                stats_gag, stats_grv, var_name, var_info, 
                output_subdir, colors, dpi
            )
            
            print(f"  ✓ {display_name}: {os.path.basename(png_path)}")
        
        # Summary statistics for this group
        print(f"\n{'='*80}")
        print(f"SUMMARY STATISTICS - {group_name}")
        print('='*80)
        
        summary_data = []
        for var_name, var_info in variables.items():
            if var_name not in all_stats:
                continue
            
            for irr_type in ['GAG', 'GRV']:
                stats_dict = all_stats[var_name][irr_type]
                
                if stats_dict is not None and stats_dict['trend_stats'] is not None:
                    trend_stats = stats_dict['trend_stats']
                    
                    summary_data.append({
                        'Variable': var_info['name'],
                        'Irrigation': irr_type,
                        'Pente': f"{trend_stats['slope']:.4f}",
                        'R²': f"{trend_stats['r_squared']:.3f}",
                        'Valeur-P': f"{trend_stats['p_value']:.4e}",
                        'Significatif': 'Oui' if trend_stats['significant'] else 'Non',
                        'Δ (%)': f"{trend_stats['percent_change']:+.1f}" if not np.isnan(trend_stats['percent_change']) else 'N/A',
                    })
        
        if len(summary_data) > 0:
            summary_df = pd.DataFrame(summary_data)
            print(summary_df.to_string(index=False))
            
            # Save summary
            summary_file = output_file.replace('.png', '_summary.csv')
            summary_df.to_csv(summary_file, index=False)
            print(f"\n✓ Summary statistics saved: {summary_file}")
        else:
            print("No statistics available for this group")
    
    print("\n" + "="*80)
    print("ALL PLOTS CREATED SUCCESSFULLY!")
    print("="*80)
    print("\n✓ Full combined plot with ALL 17 soil properties")
    print("✓ Individual group plots with custom layouts:")
    print("    - Granulometry: 2×3 grid")
    print("    - Exchangeable Elements: 2×2 grid")
    print("    - Other Properties Part 1: 2×2 grid")
    print("    - Other Properties Part 2: 2×2 grid")
    print("✓ Individual standalone plots for each variable")

# ============================================================================
# MAIN EXECUTION
# ============================================================================

if __name__ == "__main__":
    
    create_soil_properties_plots(
        data_file=DATA_FILE,
        property_groups=SOIL_PROPERTY_GROUPS,
        output_dir_base=OUTPUT_DIR_BASE,
        dpi=OUTPUT_DPI,
        colors=COLORS,
        n_cols=N_COLS
    )
    
    print("\n" + "="*80)
    print("VISUALIZATION COMPLETE!")
    print("="*80)
    
    print(f"\nAll outputs saved to: {OUTPUT_DIR_BASE}")
    print("\nGenerated files:")
    
    print("\n📊 FULL COMBINED PLOT (ALL 17 soil properties):")
    print(f"  • ALL_soil_properties_combined.png")
    print(f"  • ALL_soil_properties_combined.pdf")
    print(f"  • ALL_soil_properties_combined_summary.csv")
    
    print("\n📊 GROUP-SPECIFIC COMBINED PLOTS WITH CUSTOM LAYOUTS:")
    for group_name, group_info in SOIL_PROPERTY_GROUPS.items():
        layout = group_info['layout']
        print(f"  • {group_name} ({layout['rows']}×{layout['cols']} grid):")
        print(f"      - {group_info['output_file']}")
        print(f"      - {group_info['output_file'].replace('.png', '.pdf')}")
        print(f"      - {group_info['output_file'].replace('.png', '_summary.csv')}")
    
    print(f"\n📈 INDIVIDUAL PLOTS:")
    # Count total variables
    total_vars = 0
    for group_name, group_info in SOIL_PROPERTY_GROUPS.items():
        n_vars = len(group_info['variables'])
        total_vars += n_vars
    
    print(f"  • Total: {total_vars * 2} files ({total_vars} PNG + {total_vars} PDF)")
    for group_name, group_info in SOIL_PROPERTY_GROUPS.items():
        n_vars = len(group_info['variables'])
        print(f"      {group_name}: {n_vars} variables in {OUTPUT_DIR_BASE}/{group_info['output_subdir']}")
    
    print("\n" + "="*80)
```

# TEMPORAL TRENDS BY BLOC AND IRRIGATION TYPE

```python
"""
TEMPORAL TRENDS BY BLOC AND IRRIGATION TYPE
============================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
from matplotlib.patches import Rectangle
import warnings
import os
from pathlib import Path
warnings.filterwarnings('ignore')

# ============================================================================
# CONFIGURATION
# ============================================================================

# Path to your data (ADJUST THIS)
DATA_FILE = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/final_dataset_regrouped_categorical_with_field_id.csv'

# Output configuration
OUTPUT_DIR = 'temporal_trends_by_bloc'
OUTPUT_DPI = 600

# Response variables with display names and units
RESPONSE_VARIABLES = {
    'brix_percent': {'name': 'Brix (%)', 'unit': '% year⁻¹'},
    'purity_percent': {'name': 'Purity (%)', 'unit': '% year⁻¹'},
    'hugot_sugar': {'name': 'Sucrose (%)', 'unit': '% year⁻¹'},
    'total_recoverable_sugar_t_per_ha': {'name': 'Total Recoverable Sugar (t ha⁻¹)', 'unit': 't ha⁻¹ year⁻¹'},
    'tc_per_ha_actual': {'name': 'Cane Yield (t ha⁻¹)', 'unit': 't ha⁻¹ year⁻¹'}
}

# Visual styling - GAG in blue, GRV in red
COLORS = {
    'GAG': {
        'data_points': '#2E86AB',  # Blue
        'trend_line': '#1E5F8C',
        'error_bars': '#2E86AB',
    },
    'GRV': {
        'data_points': '#E63946',  # Red
        'trend_line': '#C1121F',
        'error_bars': '#E63946',
    }
}

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def calculate_temporal_statistics(df, response_var, year_column, irrigation_type):
    """
    Calculate temporal statistics for a response variable filtered by irrigation type
    
    Parameters:
    -----------
    df : DataFrame
    response_var : str
    year_column : str
    irrigation_type : str ('GAG' or 'GRV')
    
    Returns:
    --------
    dict with yearly_means, yearly_std, trend_stats, percent_change
    """
    # Filter by irrigation type
    df_filtered = df[df['irrigation_type'] == irrigation_type].copy()
    
    # Remove missing values
    df_clean = df_filtered[[year_column, response_var]].dropna()
    
    if len(df_clean) < 3:
        return None
    
    # Calculate yearly statistics
    yearly_stats = df_clean.groupby(year_column)[response_var].agg([
        ('mean', 'mean'),
        ('std', 'std'),
        ('count', 'count')
    ]).reset_index()
    
    years = yearly_stats[year_column].values
    means = yearly_stats['mean'].values
    stds = yearly_stats['std'].values
    
    # Linear regression
    if len(years) >= 3:
        slope, intercept, r_value, p_value, std_err = stats.linregress(years, means)
        
        # Calculate percentage change over entire period
        first_value = means[0]
        last_value = means[-1]
        
        if abs(first_value) > 1e-10:
            percent_change = ((last_value - first_value) / abs(first_value)) * 100
        else:
            percent_change = np.nan
        
        absolute_change = last_value - first_value
        
        trend_stats = {
            'slope': slope,
            'intercept': intercept,
            'r_squared': r_value**2,
            'p_value': p_value,
            'significant': p_value < 0.05,
            'percent_change': percent_change,
            'absolute_change': absolute_change,
            'years_span': years[-1] - years[0],
            'first_value': first_value,
            'last_value': last_value
        }
    else:
        trend_stats = None
    
    return {
        'years': years,
        'means': means,
        'stds': stds,
        'yearly_stats': yearly_stats,
        'trend_stats': trend_stats,
        'irrigation_type': irrigation_type
    }

def format_pvalue(p):
    """Format p-value for display"""
    if p < 0.001:
        return 'p < 0.001'
    elif p < 0.01:
        return f'p < 0.01'
    elif p < 0.05:
        return f'p = {p:.3f}'
    else:
        return f'p = {p:.2f}'

def plot_dual_trends(ax, stats_gag, stats_grv, var_name, var_info, colors):
    """
    Plot temporal trends for both GAG and GRV on the same axis
    
    Parameters:
    -----------
    ax : matplotlib axis
    stats_gag : dict from calculate_temporal_statistics for GAG
    stats_grv : dict from calculate_temporal_statistics for GRV
    var_name : str (variable key)
    var_info : dict with 'name' and 'unit'
    colors : dict
    """
    display_name = var_info['name']
    unit = var_info['unit']
    
    # Plot GAG data (Blue)
    if stats_gag is not None:
        years_gag = stats_gag['years']
        means_gag = stats_gag['means']
        stds_gag = stats_gag['stds']
        
        ax.errorbar(years_gag, means_gag, yerr=stds_gag, 
                    fmt='o', markersize=6, 
                    color=colors['GAG']['data_points'],
                    ecolor=colors['GAG']['error_bars'],
                    elinewidth=1.5,
                    capsize=4,
                    capthick=1.5,
                    alpha=0.8,
                    zorder=3,
                    label='GAG')
        
        ax.plot(years_gag, means_gag, '-', 
                color=colors['GAG']['data_points'], 
                alpha=0.3, 
                linewidth=1,
                zorder=2)
        
        # Plot GAG trend line
        if stats_gag['trend_stats'] is not None:
            trend_line_gag = stats_gag['trend_stats']['slope'] * years_gag + stats_gag['trend_stats']['intercept']
            ax.plot(years_gag, trend_line_gag, '--', 
                    color=colors['GAG']['trend_line'],
                    linewidth=2.5,
                    alpha=0.9,
                    zorder=4)
    
    # Plot GRV data (Red)
    if stats_grv is not None:
        years_grv = stats_grv['years']
        means_grv = stats_grv['means']
        stds_grv = stats_grv['stds']
        
        ax.errorbar(years_grv, means_grv, yerr=stds_grv, 
                    fmt='s', markersize=6, 
                    color=colors['GRV']['data_points'],
                    ecolor=colors['GRV']['error_bars'],
                    elinewidth=1.5,
                    capsize=4,
                    capthick=1.5,
                    alpha=0.8,
                    zorder=3,
                    label='GRV')
        
        ax.plot(years_grv, means_grv, '-', 
                color=colors['GRV']['data_points'], 
                alpha=0.3, 
                linewidth=1,
                zorder=2)
        
        # Plot GRV trend line
        if stats_grv['trend_stats'] is not None:
            trend_line_grv = stats_grv['trend_stats']['slope'] * years_grv + stats_grv['trend_stats']['intercept']
            ax.plot(years_grv, trend_line_grv, '--', 
                    color=colors['GRV']['trend_line'],
                    linewidth=2.5,
                    alpha=0.9,
                    zorder=4)
    
    # Add statistics text boxes
    # GAG statistics (top-left)
    if stats_gag is not None and stats_gag['trend_stats'] is not None:
        ts = stats_gag['trend_stats']
        r2_str = f"R² = {ts['r_squared']:.3f}"
        p_val_str = format_pvalue(ts['p_value'])
        pct_change = ts['percent_change']
        
        if not np.isnan(pct_change):
            pct_sign = '+' if pct_change > 0 else ''
            pct_str = f"Δ = {pct_sign}{pct_change:.1f}%"
        else:
            pct_str = "Δ = N/A"
        
        stats_text_gag = f"GAG\n{r2_str}\n{p_val_str}\n{pct_str}"
        
        bbox_props_gag = dict(
            boxstyle='round,pad=0.4',
            facecolor='white',
            edgecolor=colors['GAG']['trend_line'],
            linewidth=2,
            alpha=0.95
        )
        
        ax.text(0.05, 0.95, stats_text_gag,
                transform=ax.transAxes,
                fontsize=8.5,
                verticalalignment='top',
                horizontalalignment='left',
                bbox=bbox_props_gag,
                fontweight='bold',
                family='monospace')
    
    # GRV statistics (top-right)
    if stats_grv is not None and stats_grv['trend_stats'] is not None:
        ts = stats_grv['trend_stats']
        r2_str = f"R² = {ts['r_squared']:.3f}"
        p_val_str = format_pvalue(ts['p_value'])
        pct_change = ts['percent_change']
        
        if not np.isnan(pct_change):
            pct_sign = '+' if pct_change > 0 else ''
            pct_str = f"Δ = {pct_sign}{pct_change:.1f}%"
        else:
            pct_str = "Δ = N/A"
        
        stats_text_grv = f"GRV\n{r2_str}\n{p_val_str}\n{pct_str}"
        
        bbox_props_grv = dict(
            boxstyle='round,pad=0.4',
            facecolor='white',
            edgecolor=colors['GRV']['trend_line'],
            linewidth=2,
            alpha=0.95
        )
        
        ax.text(0.95, 0.95, stats_text_grv,
                transform=ax.transAxes,
                fontsize=8.5,
                verticalalignment='top',
                horizontalalignment='right',
                bbox=bbox_props_grv,
                fontweight='bold',
                family='monospace')
    
    # Formatting
    ax.set_xlabel('Année', fontsize=12, fontweight='bold')
    ax.set_ylabel(display_name, fontsize=12, fontweight='bold')
    ax.set_title(display_name, fontsize=13, fontweight='bold', pad=10)
    
    # Grid
    ax.grid(True, alpha=0.3, linestyle='--', linewidth=0.5, zorder=1)
    ax.set_axisbelow(True)
    
    # Legend
    ax.legend(loc='lower left', frameon=True, framealpha=0.9, 
              edgecolor='black', fontsize=10, ncol=2)
    
    # Tick parameters
    ax.tick_params(axis='both', which='major', labelsize=10, width=1.5, length=5)
    
    # Set integer years on x-axis (use union of both year ranges)
    all_years = []
    if stats_gag is not None:
        all_years.extend(stats_gag['years'])
    if stats_grv is not None:
        all_years.extend(stats_grv['years'])
    
    if len(all_years) > 0:
        unique_years = sorted(set(all_years))
        ax.set_xticks(unique_years)
        ax.set_xticklabels([int(y) for y in unique_years], rotation=45, ha='right')
    
    # Spine formatting
    for spine in ax.spines.values():
        spine.set_linewidth(1.5)
        spine.set_color('black')

# ============================================================================
# MAIN PLOTTING FUNCTION FOR EACH BLOC
# ============================================================================

def create_bloc_temporal_plot(df_bloc, bloc_name, response_variables, 
                             output_dir, year_column, dpi=600, colors=COLORS):
    """
    Create combined temporal trends plot for all response variables
    for a specific Bloc, separated by irrigation type (GAG and GRV)
    """
    
    print(f"\n{'='*80}")
    print(f"PROCESSING BLOC: {bloc_name}")
    print('='*80)
    
    # Check data availability
    print(f"Total observations in Bloc {bloc_name}: {len(df_bloc)}")
    print(f"  GAG: {len(df_bloc[df_bloc['irrigation_type'] == 'GAG'])}")
    print(f"  GRV: {len(df_bloc[df_bloc['irrigation_type'] == 'GRV'])}")
    
    # Calculate statistics for all variables
    print(f"\nCalculating temporal statistics for Bloc {bloc_name}...")
    
    all_stats = {}
    
    for var_name, var_info in response_variables.items():
        display_name = var_info['name']
        
        if var_name not in df_bloc.columns:
            print(f"  ⚠ Warning: {var_name} not found in dataset, skipping...")
            continue
        
        # Calculate for GAG
        stats_gag = calculate_temporal_statistics(df_bloc, var_name, year_column, 'GAG')
        
        # Calculate for GRV
        stats_grv = calculate_temporal_statistics(df_bloc, var_name, year_column, 'GRV')
        
        all_stats[var_name] = {
            'GAG': stats_gag,
            'GRV': stats_grv
        }
        
        print(f"\n  • {display_name}:")
        
        # GAG stats
        if stats_gag is not None and stats_gag['trend_stats'] is not None:
            ts = stats_gag['trend_stats']
            sig = "✓ Sig" if ts['significant'] else "✗ NS"
            print(f"      GAG: {sig} (p={ts['p_value']:.4f}, R²={ts['r_squared']:.3f}, Δ={ts['percent_change']:+.1f}%)")
        else:
            print(f"      GAG: Insufficient data")
        
        # GRV stats
        if stats_grv is not None and stats_grv['trend_stats'] is not None:
            ts = stats_grv['trend_stats']
            sig = "✓ Sig" if ts['significant'] else "✗ NS"
            print(f"      GRV: {sig} (p={ts['p_value']:.4f}, R²={ts['r_squared']:.3f}, Δ={ts['percent_change']:+.1f}%)")
        else:
            print(f"      GRV: Insufficient data")
    
    # Skip if no valid data
    if len(all_stats) == 0:
        print(f"\n⚠ No valid data for Bloc {bloc_name}, skipping plot generation.")
        return None, None, None
    
    # Create figure with subplots
    n_vars = len(all_stats)
    n_cols = 3
    n_rows = int(np.ceil(n_vars / n_cols))
    
    print(f"\nCreating {n_rows}×{n_cols} grid plot for Bloc {bloc_name}...")
    
    plt.style.use('seaborn-v0_8-white')
    
    fig_width = 18
    fig_height = n_rows * 4.5
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(fig_width, fig_height))
    
    if n_rows == 1:
        axes = axes.reshape(1, -1)
    axes_flat = axes.flatten()
    
    # Plot each variable
    for idx, (var_name, var_info) in enumerate(response_variables.items()):
        if var_name not in all_stats:
            continue
        
        ax = axes_flat[idx]
        stats_gag = all_stats[var_name]['GAG']
        stats_grv = all_stats[var_name]['GRV']
        
        plot_dual_trends(ax, stats_gag, stats_grv, var_name, var_info, colors)
    
    # Hide unused subplots
    for idx in range(len(all_stats), len(axes_flat)):
        axes_flat[idx].set_visible(False)
    
    # Overall title
    fig.suptitle(f'Évolution Temporelle - Bloc {bloc_name} (GAG vs GRV)',
                 fontsize=18, fontweight='bold', y=0.995)
    
    # Adjust layout
    plt.tight_layout(rect=[0, 0.02, 1, 0.99], h_pad=3.5, w_pad=3.0)
    
    # Save figure
    output_file = os.path.join(output_dir, f'bloc_{bloc_name}_temporal_trends.png')
    print(f"\nSaving figure to: {output_file}")
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    print(f"✓ PNG saved at {dpi} DPI")
    
    # Also save as PDF
    pdf_file = os.path.join(output_dir, f'bloc_{bloc_name}_temporal_trends.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    print(f"✓ PDF saved: {pdf_file}")
    
    plt.close()
    
    # Summary statistics
    print(f"\nSUMMARY STATISTICS - Bloc {bloc_name}")
    print('='*80)
    
    summary_data = []
    for var_name, var_info in response_variables.items():
        if var_name not in all_stats:
            continue
        
        for irr_type in ['GAG', 'GRV']:
            stats_dict = all_stats[var_name][irr_type]
            
            if stats_dict is not None and stats_dict['trend_stats'] is not None:
                trend_stats = stats_dict['trend_stats']
                
                summary_data.append({
                    'Bloc': bloc_name,
                    'Variable': var_info['name'],
                    'Irrigation': irr_type,
                    'Pente': f"{trend_stats['slope']:.4f}",
                    'R²': f"{trend_stats['r_squared']:.3f}",
                    'Valeur-P': f"{trend_stats['p_value']:.4e}",
                    'Significatif': 'Oui' if trend_stats['significant'] else 'Non',
                    'Δ (%)': f"{trend_stats['percent_change']:+.1f}" if not np.isnan(trend_stats['percent_change']) else 'N/A',
                })
    
    summary_df = pd.DataFrame(summary_data)
    if len(summary_df) > 0:
        print(summary_df.to_string(index=False))
    
    # Save summary
    summary_file = os.path.join(output_dir, f'bloc_{bloc_name}_summary.csv')
    summary_df.to_csv(summary_file, index=False)
    print(f"\n✓ Summary statistics saved: {summary_file}")
    
    return fig, all_stats, summary_df

# ============================================================================
# MAIN EXECUTION FUNCTION
# ============================================================================

def create_all_bloc_plots(data_file, response_variables, output_dir, 
                         dpi=600, colors=COLORS):
    """
    Create temporal trend plots for all Blocs
    """
    print("="*80)
    print("TEMPORAL TRENDS BY BLOC AND IRRIGATION TYPE")
    print("="*80)
    
    # Load data
    print(f"\nLoading data from: {data_file}")
    df = pd.read_csv(data_file)
    
    # Find year column
    year_column = None
    for col in ['campaign_year', 'Harvest_Year', 'year', 'Campaign_Year']:
        if col in df.columns:
            year_column = col
            break
    
    if year_column is None:
        raise ValueError("No year column found in dataset!")
    
    # Check required columns
    if 'irrigation_type' not in df.columns:
        raise ValueError("irrigation_type column not found in dataset!")
    
    if 'Bloc' not in df.columns:
        raise ValueError("Bloc column not found in dataset!")
    
    print(f"\nYear column: {year_column}")
    print(f"Years available: {sorted(df[year_column].unique())}")
    print(f"Total observations: {len(df)}")
    
    # Get unique blocs
    blocs = sorted(df['Bloc'].unique())
    print(f"\nBlocs found: {blocs}")
    print(f"Number of Blocs: {len(blocs)}")
    
    # Print data distribution
    print("\nData distribution by Bloc and irrigation type:")
    distribution = df.groupby(['Bloc', 'irrigation_type']).size().unstack(fill_value=0)
    print(distribution)
    
    # Create output directory
    os.makedirs(output_dir, exist_ok=True)
    print(f"\nOutput directory: {output_dir}")
    
    # Process each Bloc
    all_summaries = []
    
    for bloc in blocs:
        df_bloc = df[df['Bloc'] == bloc].copy()
        
        fig, stats, summary_df = create_bloc_temporal_plot(
            df_bloc=df_bloc,
            bloc_name=bloc,
            response_variables=response_variables,
            output_dir=output_dir,
            year_column=year_column,
            dpi=dpi,
            colors=colors
        )
        
        if summary_df is not None and len(summary_df) > 0:
            all_summaries.append(summary_df)
    
    # Combine all summaries
    if len(all_summaries) > 0:
        combined_summary = pd.concat(all_summaries, ignore_index=True)
        combined_summary_file = os.path.join(output_dir, 'all_blocs_summary.csv')
        combined_summary.to_csv(combined_summary_file, index=False)
        print(f"\n{'='*80}")
        print("COMBINED SUMMARY FOR ALL BLOCS")
        print('='*80)
        print(combined_summary.to_string(index=False))
        print(f"\n✓ Combined summary saved: {combined_summary_file}")
    
    print("\n" + "="*80)
    print("VISUALIZATION COMPLETE!")
    print("="*80)
    print(f"\nGenerated files in: {output_dir}")
    print(f"  • {len(blocs)} combined PNG plots (one per Bloc)")
    print(f"  • {len(blocs)} PDF plots (one per Bloc)")
    print(f"  • {len(blocs)} CSV summaries (one per Bloc)")
    print(f"  • 1 combined summary CSV (all Blocs)")
    print(f"\nTotal files: {len(blocs) * 3 + 1}")

# ============================================================================
# MAIN EXECUTION
# ============================================================================

if __name__ == "__main__":
    
    create_all_bloc_plots(
        data_file=DATA_FILE,
        response_variables=RESPONSE_VARIABLES,
        output_dir=OUTPUT_DIR,
        dpi=OUTPUT_DPI,
        colors=COLORS
    )
    
    print("\n" + "="*80)
```

# TEMPORAL TRENDS FOR SOIL PROPERTIES BY IRRIGATION TYPE

```python
"""
TEMPORAL TRENDS FOR SOIL PROPERTIES BY IRRIGATION TYPE
=======================================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
import warnings
import os
warnings.filterwarnings('ignore')

# ============================================================================
# CONFIGURATION
# ============================================================================

# Path to your data (ADJUST THIS)
DATA_FILE = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/final_dataset_regrouped_categorical.csv'

# Output configuration
OUTPUT_DIR_BASE = 'soil_properties_temporal_trends'
OUTPUT_DPI = 600

# Soil property groups with display names, units, and custom layouts
SOIL_PROPERTY_GROUPS = {
    'Granulometry': {
        'variables': {
            'Clay_percent_complete': {'name': 'Clay (%)', 'unit': '% year⁻¹'},
            'Fine_Silt_percent_complete': {'name': 'Fine Silt (%)', 'unit': '% year⁻¹'},
            'Coarse_Silt_percent_complete': {'name': 'Coarse Silt (%)', 'unit': '% year⁻¹'},
            'Fine_Sand_percent_complete': {'name': 'Fine Sand (%)', 'unit': '% year⁻¹'},
            'Coarse_Sand_percent_complete': {'name': 'Coarse Sand (%)', 'unit': '% year⁻¹'}
        },
        'output_file': 'granulometry_temporal_trends.png',
        'output_subdir': 'granulometry_individual',
        'layout': {'rows': 2, 'cols': 3}  # 2 rows x 3 columns
    },
    'Exchangeable_Elements': {
        'variables': {
            'Exchangeable_Calcium_meq_100g_complete': {'name': 'Exch. Ca (meq/100g)', 'unit': 'meq/100g year⁻¹'},
            'Exchangeable_Magnesium_meq_100g_complete': {'name': 'Exch. Mg (meq/100g)', 'unit': 'meq/100g year⁻¹'},
            'Exchangeable_Potassium_meq_100g_complete': {'name': 'Exch. K (meq/100g)', 'unit': 'meq/100g year⁻¹'},
            'Exchangeable_Sodium_meq_100g_complete': {'name': 'Exch. Na (meq/100g)', 'unit': 'meq/100g year⁻¹'}
        },
        'output_file': 'exchangeable_elements_temporal_trends.png',
        'output_subdir': 'exchangeable_elements_individual',
        'layout': {'rows': 2, 'cols': 2}  # 2 rows x 2 columns
    },
    'Other_Properties_Part1': {
        'variables': {
            'pH_complete': {'name': 'pH', 'unit': 'pH year⁻¹'},
            'Organic_Carbon_per_mille_complete': {'name': 'Organic C (‰)', 'unit': '‰ year⁻¹'},
            'Total_Nitrogen_per_mille_complete': {'name': 'Total N (‰)', 'unit': '‰ year⁻¹'},
            'Available_P2O5_mg_kg_complete': {'name': 'Available P₂O₅ (mg/kg)', 'unit': 'mg/kg year⁻¹'}
        },
        'output_file': 'other_properties_part1_temporal_trends.png',
        'output_subdir': 'other_properties_individual',
        'layout': {'rows': 2, 'cols': 2},  # 2 rows x 2 columns
        'group_title': 'Autres Propriétés du Sol - Partie 1'
    },
    'Other_Properties_Part2': {
        'variables': {
            'Electrical_Conductivity_mmho_cm_complete': {'name': 'EC (mmho/cm)', 'unit': 'mmho/cm year⁻¹'},
            'Mg_Ca_Ratio_complete': {'name': 'Mg/Ca Ratio', 'unit': 'year⁻¹'},
            'Mg_K_Ratio_complete': {'name': 'Mg/K Ratio', 'unit': 'year⁻¹'},
            'Compaction_Index_complete': {'name': 'Compaction Index', 'unit': 'year⁻¹'}
        },
        'output_file': 'other_properties_part2_temporal_trends.png',
        'output_subdir': 'other_properties_individual',
        'layout': {'rows': 2, 'cols': 2},  # 2 rows x 2 columns
        'group_title': 'Autres Propriétés du Sol - Partie 2'
    }
}

# Visual styling - GAG in blue, GRV in red
COLORS = {
    'GAG': {
        'data_points': '#2E86AB',  # Blue
        'trend_line': '#1E5F8C',
        'error_bars': '#2E86AB',
    },
    'GRV': {
        'data_points': '#E63946',  # Red
        'trend_line': '#C1121F',
        'error_bars': '#E63946',
    }
}

# Grid layout: 4 columns (for full combined plot only)
N_COLS = 4

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def calculate_temporal_statistics(df, response_var, year_column, irrigation_type):
    """
    Calculate temporal statistics for a response variable filtered by irrigation type
    
    Parameters:
    -----------
    df : DataFrame
    response_var : str
    year_column : str
    irrigation_type : str ('GAG' or 'GRV')
    
    Returns:
    --------
    dict with yearly_means, yearly_std, trend_stats, percent_change or None
    """
    # Filter by irrigation type
    df_filtered = df[df['irrigation_type'] == irrigation_type].copy()
    
    # Remove missing values
    df_clean = df_filtered[[year_column, response_var]].dropna()
    
    if len(df_clean) < 3:
        return None
    
    # Calculate yearly statistics
    yearly_stats = df_clean.groupby(year_column)[response_var].agg([
        ('mean', 'mean'),
        ('std', 'std'),
        ('count', 'count')
    ]).reset_index()
    
    years = yearly_stats[year_column].values
    means = yearly_stats['mean'].values
    stds = yearly_stats['std'].values
    
    # Linear regression
    if len(years) >= 3:
        slope, intercept, r_value, p_value, std_err = stats.linregress(years, means)
        
        # Calculate percentage change over entire period
        first_value = means[0]
        last_value = means[-1]
        
        if abs(first_value) > 1e-10:
            percent_change = ((last_value - first_value) / abs(first_value)) * 100
        else:
            percent_change = np.nan
        
        absolute_change = last_value - first_value
        
        trend_stats = {
            'slope': slope,
            'intercept': intercept,
            'r_squared': r_value**2,
            'p_value': p_value,
            'significant': p_value < 0.05,
            'percent_change': percent_change,
            'absolute_change': absolute_change,
            'years_span': years[-1] - years[0],
            'first_value': first_value,
            'last_value': last_value
        }
    else:
        trend_stats = None
    
    return {
        'years': years,
        'means': means,
        'stds': stds,
        'yearly_stats': yearly_stats,
        'trend_stats': trend_stats,
        'irrigation_type': irrigation_type
    }

def format_pvalue(p):
    """Format p-value for display"""
    if p < 0.001:
        return 'p < 0.001'
    elif p < 0.01:
        return f'p < 0.01'
    elif p < 0.05:
        return f'p = {p:.3f}'
    else:
        return f'p = {p:.2f}'

def plot_dual_trends(ax, stats_gag, stats_grv, var_name, var_info, colors):
    """
    Plot temporal trends for both GAG and GRV on the same axis
    
    Parameters:
    -----------
    ax : matplotlib axis
    stats_gag : dict from calculate_temporal_statistics for GAG
    stats_grv : dict from calculate_temporal_statistics for GRV
    var_name : str (variable key)
    var_info : dict with 'name' and 'unit'
    colors : dict
    """
    display_name = var_info['name']
    unit = var_info['unit']
    
    # Plot GAG data (Blue)
    if stats_gag is not None:
        years_gag = stats_gag['years']
        means_gag = stats_gag['means']
        stds_gag = stats_gag['stds']
        
        ax.errorbar(years_gag, means_gag, yerr=stds_gag, 
                    fmt='o', markersize=5, 
                    color=colors['GAG']['data_points'],
                    ecolor=colors['GAG']['error_bars'],
                    elinewidth=1.2,
                    capsize=3,
                    capthick=1.2,
                    alpha=0.8,
                    zorder=3,
                    label='GAG')
        
        ax.plot(years_gag, means_gag, '-', 
                color=colors['GAG']['data_points'], 
                alpha=0.3, 
                linewidth=1,
                zorder=2)
        
        # Plot GAG trend line
        if stats_gag['trend_stats'] is not None:
            trend_line_gag = stats_gag['trend_stats']['slope'] * years_gag + stats_gag['trend_stats']['intercept']
            ax.plot(years_gag, trend_line_gag, '--', 
                    color=colors['GAG']['trend_line'],
                    linewidth=2,
                    alpha=0.9,
                    zorder=4)
    
    # Plot GRV data (Red)
    if stats_grv is not None:
        years_grv = stats_grv['years']
        means_grv = stats_grv['means']
        stds_grv = stats_grv['stds']
        
        ax.errorbar(years_grv, means_grv, yerr=stds_grv, 
                    fmt='s', markersize=5, 
                    color=colors['GRV']['data_points'],
                    ecolor=colors['GRV']['error_bars'],
                    elinewidth=1.2,
                    capsize=3,
                    capthick=1.2,
                    alpha=0.8,
                    zorder=3,
                    label='GRV')
        
        ax.plot(years_grv, means_grv, '-', 
                color=colors['GRV']['data_points'], 
                alpha=0.3, 
                linewidth=1,
                zorder=2)
        
        # Plot GRV trend line
        if stats_grv['trend_stats'] is not None:
            trend_line_grv = stats_grv['trend_stats']['slope'] * years_grv + stats_grv['trend_stats']['intercept']
            ax.plot(years_grv, trend_line_grv, '--', 
                    color=colors['GRV']['trend_line'],
                    linewidth=2,
                    alpha=0.9,
                    zorder=4)
    
    # Add statistics text boxes
    # GAG statistics (top-left)
    if stats_gag is not None and stats_gag['trend_stats'] is not None:
        ts = stats_gag['trend_stats']
        r2_str = f"R² = {ts['r_squared']:.3f}"
        p_val_str = format_pvalue(ts['p_value'])
        pct_change = ts['percent_change']
        
        if not np.isnan(pct_change):
            pct_sign = '+' if pct_change > 0 else ''
            pct_str = f"Δ = {pct_sign}{pct_change:.1f}%"
        else:
            pct_str = "Δ = N/A"
        
        stats_text_gag = f"GAG\n{r2_str}\n{p_val_str}\n{pct_str}"
        
        bbox_props_gag = dict(
            boxstyle='round,pad=0.3',
            facecolor='white',
            edgecolor=colors['GAG']['trend_line'],
            linewidth=1.5,
            alpha=0.95
        )
        
        ax.text(0.05, 0.95, stats_text_gag,
                transform=ax.transAxes,
                fontsize=7.5,
                verticalalignment='top',
                horizontalalignment='left',
                bbox=bbox_props_gag,
                fontweight='bold',
                family='monospace')
    
    # GRV statistics (top-right)
    if stats_grv is not None and stats_grv['trend_stats'] is not None:
        ts = stats_grv['trend_stats']
        r2_str = f"R² = {ts['r_squared']:.3f}"
        p_val_str = format_pvalue(ts['p_value'])
        pct_change = ts['percent_change']
        
        if not np.isnan(pct_change):
            pct_sign = '+' if pct_change > 0 else ''
            pct_str = f"Δ = {pct_sign}{pct_change:.1f}%"
        else:
            pct_str = "Δ = N/A"
        
        stats_text_grv = f"GRV\n{r2_str}\n{p_val_str}\n{pct_str}"
        
        bbox_props_grv = dict(
            boxstyle='round,pad=0.3',
            facecolor='white',
            edgecolor=colors['GRV']['trend_line'],
            linewidth=1.5,
            alpha=0.95
        )
        
        ax.text(0.95, 0.95, stats_text_grv,
                transform=ax.transAxes,
                fontsize=7.5,
                verticalalignment='top',
                horizontalalignment='right',
                bbox=bbox_props_grv,
                fontweight='bold',
                family='monospace')
    
    # Formatting
    ax.set_xlabel('Année', fontsize=10, fontweight='bold')
    ax.set_ylabel(display_name, fontsize=10, fontweight='bold')
    ax.set_title(display_name, fontsize=11, fontweight='bold', pad=8)
    
    # Grid
    ax.grid(True, alpha=0.3, linestyle='--', linewidth=0.5, zorder=1)
    ax.set_axisbelow(True)
    
    # Legend
    ax.legend(loc='lower left', frameon=True, framealpha=0.9, 
              edgecolor='black', fontsize=8, ncol=2)
    
    # Tick parameters
    ax.tick_params(axis='both', which='major', labelsize=9, width=1.2, length=4)
    
    # Set integer years on x-axis
    all_years = []
    if stats_gag is not None:
        all_years.extend(stats_gag['years'])
    if stats_grv is not None:
        all_years.extend(stats_grv['years'])
    
    if len(all_years) > 0:
        unique_years = sorted(set(all_years))
        ax.set_xticks(unique_years)
        ax.set_xticklabels([int(y) for y in unique_years], rotation=45, ha='right')
    
    # Spine formatting
    for spine in ax.spines.values():
        spine.set_linewidth(1.2)
        spine.set_color('black')

# ============================================================================
# FUNCTION TO SAVE INDIVIDUAL PLOTS
# ============================================================================

def save_individual_plot(stats_gag, stats_grv, var_name, var_info, 
                        output_dir, colors, dpi=600):
    """
    Save a standalone plot for a single variable with both irrigation types
    """
    plt.style.use('seaborn-v0_8-white')
    
    # Create figure with single subplot
    fig, ax = plt.subplots(1, 1, figsize=(10, 9))
    
    # Plot the trends
    plot_dual_trends(ax, stats_gag, stats_grv, var_name, var_info, colors)
    
    # Adjust layout
    plt.tight_layout()
    
    # Create safe filename
    safe_filename = var_name.replace('_complete', '').replace('_', '-').replace('/', '-')
    
    # Save PNG
    png_path = os.path.join(output_dir, f'{safe_filename}_GAG_vs_GRV.png')
    plt.savefig(png_path, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    
    # Save PDF
    pdf_path = os.path.join(output_dir, f'{safe_filename}_GAG_vs_GRV.pdf')
    plt.savefig(pdf_path, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    
    plt.close()
    
    return png_path, pdf_path

# ============================================================================
# MAIN PLOTTING FUNCTION
# ============================================================================

def create_full_combined_plot(df, year_column, property_groups, output_dir_base, 
                             dpi=600, colors=COLORS, n_cols=N_COLS):
    """
    Create ONE full combined plot with ALL soil properties (17 variables)
    """
    print("\n" + "="*80)
    print("CREATING FULL COMBINED PLOT WITH ALL SOIL PROPERTIES")
    print("="*80)
    
    # Collect all variables from all groups
    all_variables = {}
    for group_name, group_info in property_groups.items():
        all_variables.update(group_info['variables'])
    
    print(f"\nTotal variables to plot: {len(all_variables)}")
    
    # Calculate statistics for all variables
    print("\nCalculating temporal statistics for all variables...")
    
    all_stats = {}
    
    for var_name, var_info in all_variables.items():
        display_name = var_info['name']
        
        if var_name not in df.columns:
            print(f"  ⚠ Warning: {var_name} not found in dataset, skipping...")
            continue
        
        # Calculate for GAG
        stats_gag = calculate_temporal_statistics(df, var_name, year_column, 'GAG')
        
        # Calculate for GRV
        stats_grv = calculate_temporal_statistics(df, var_name, year_column, 'GRV')
        
        all_stats[var_name] = {
            'GAG': stats_gag,
            'GRV': stats_grv
        }
        
        print(f"  ✓ {display_name}")
    
    # Create figure with subplots (4 columns)
    n_vars = len(all_stats)
    n_rows = int(np.ceil(n_vars / n_cols))
    
    print(f"\nCreating FULL COMBINED {n_rows}×{n_cols} grid plot...")
    
    plt.style.use('seaborn-v0_8-white')
    
    fig_width = 20
    fig_height = n_rows * 4
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(fig_width, fig_height))
    
    # Handle different subplot array shapes
    if n_rows == 1 and n_cols == 1:
        axes = np.array([[axes]])
    elif n_rows == 1:
        axes = axes.reshape(1, -1)
    elif n_cols == 1:
        axes = axes.reshape(-1, 1)
    
    axes_flat = axes.flatten()
    
    # Plot each variable
    for idx, (var_name, var_info) in enumerate(all_variables.items()):
        if var_name not in all_stats:
            continue
        
        ax = axes_flat[idx]
        stats_gag = all_stats[var_name]['GAG']
        stats_grv = all_stats[var_name]['GRV']
        
        plot_dual_trends(ax, stats_gag, stats_grv, var_name, var_info, colors)
    
    # Hide unused subplots
    for idx in range(len(all_stats), len(axes_flat)):
        axes_flat[idx].set_visible(False)
    
    # Overall title
    fig.suptitle('Propriétés du Sol - Évolution Temporelle (GAG vs GRV) - Toutes Variables',
                 fontsize=18, fontweight='bold', y=0.995)
    
    # Adjust layout
    plt.tight_layout(rect=[0, 0.02, 1, 0.99], h_pad=3.0, w_pad=2.5)
    
    # Save full combined figure
    output_file = os.path.join(output_dir_base, 'ALL_soil_properties_combined.png')
    print(f"\nSaving FULL COMBINED figure to: {output_file}")
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    print(f"✓ Combined PNG saved at {dpi} DPI")
    
    # Also save as PDF
    pdf_file = output_file.replace('.png', '.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    print(f"✓ Combined PDF saved: {pdf_file}")
    
    plt.close()
    
    # Summary statistics for ALL variables
    print(f"\n{'='*80}")
    print("SUMMARY STATISTICS - ALL SOIL PROPERTIES")
    print('='*80)
    
    summary_data = []
    for var_name, var_info in all_variables.items():
        if var_name not in all_stats:
            continue
        
        for irr_type in ['GAG', 'GRV']:
            stats_dict = all_stats[var_name][irr_type]
            
            if stats_dict is not None and stats_dict['trend_stats'] is not None:
                trend_stats = stats_dict['trend_stats']
                
                summary_data.append({
                    'Variable': var_info['name'],
                    'Irrigation': irr_type,
                    'Pente': f"{trend_stats['slope']:.4f}",
                    'R²': f"{trend_stats['r_squared']:.3f}",
                    'Valeur-P': f"{trend_stats['p_value']:.4e}",
                    'Significatif': 'Oui' if trend_stats['significant'] else 'Non',
                    'Δ (%)': f"{trend_stats['percent_change']:+.1f}" if not np.isnan(trend_stats['percent_change']) else 'N/A',
                })
    
    if len(summary_data) > 0:
        summary_df = pd.DataFrame(summary_data)
        print(summary_df.to_string(index=False))
        
        # Save summary
        summary_file = output_file.replace('.png', '_summary.csv')
        summary_df.to_csv(summary_file, index=False)
        print(f"\n✓ Full summary statistics saved: {summary_file}")
    
    return output_file, pdf_file, summary_file

def create_soil_properties_plots(data_file, property_groups, output_dir_base, 
                                 dpi=600, colors=COLORS, n_cols=N_COLS):
    """
    Create temporal trends plots for all soil property groups
    separated by irrigation type (GAG and GRV)
    
    Creates:
    1. ONE full combined plot with ALL 17 soil properties
    2. Individual group plots with custom layouts:
       - Granulometry: 2×3 grid
       - Exchangeable Elements: 2×2 grid
       - Other Properties Part 1: 2×2 grid
       - Other Properties Part 2: 2×2 grid
    3. Individual standalone plots for each variable
    """
    print("="*80)
    print("SOIL PROPERTIES TEMPORAL TRENDS (GAG vs GRV)")
    print("="*80)
    
    # Load data
    print(f"\nLoading data from: {data_file}")
    df = pd.read_csv(data_file)
    
    # Find year column
    year_column = None
    for col in ['campaign_year', 'Harvest_Year', 'year', 'Campaign_Year']:
        if col in df.columns:
            year_column = col
            break
    
    if year_column is None:
        raise ValueError("No year column found in dataset!")
    
    # Check irrigation_type column
    if 'irrigation_type' not in df.columns:
        raise ValueError("irrigation_type column not found in dataset!")
    
    irrigation_types = df['irrigation_type'].unique()
    print(f"\nIrrigation types found: {irrigation_types}")
    
    print(f"Year column: {year_column}")
    print(f"Years available: {sorted(df[year_column].unique())}")
    print(f"Total observations: {len(df)}")
    print(f"  GAG: {len(df[df['irrigation_type'] == 'GAG'])}")
    print(f"  GRV: {len(df[df['irrigation_type'] == 'GRV'])}")
    
    # Create base output directory
    os.makedirs(output_dir_base, exist_ok=True)
    print(f"\nOutput directory: {output_dir_base}")
    
    # ========================================================================
    # STEP 1: CREATE FULL COMBINED PLOT WITH ALL SOIL PROPERTIES
    # ========================================================================
    full_combined_files = create_full_combined_plot(
        df, year_column, property_groups, output_dir_base, dpi, colors, n_cols
    )
    
    # ========================================================================
    # STEP 2: CREATE GROUP-SPECIFIC PLOTS WITH CUSTOM LAYOUTS
    # ========================================================================
    
    # Process each property group
    for group_name, group_info in property_groups.items():
        print("\n" + "="*80)
        print(f"PROCESSING GROUP: {group_name}")
        print("="*80)
        
        variables = group_info['variables']
        output_file = os.path.join(output_dir_base, group_info['output_file'])
        output_subdir = os.path.join(output_dir_base, group_info['output_subdir'])
        layout = group_info['layout']
        n_rows = layout['rows']
        n_cols_group = layout['cols']
        
        print(f"Layout: {n_rows} rows × {n_cols_group} columns")
        
        # Create subdirectory for individual plots
        os.makedirs(output_subdir, exist_ok=True)
        
        # Calculate statistics for all variables in this group
        print("\nCalculating temporal statistics...")
        
        all_stats = {}
        
        for var_name, var_info in variables.items():
            display_name = var_info['name']
            
            if var_name not in df.columns:
                print(f"  ⚠ Warning: {var_name} not found in dataset, skipping...")
                continue
            
            # Calculate for GAG
            stats_gag = calculate_temporal_statistics(df, var_name, year_column, 'GAG')
            
            # Calculate for GRV
            stats_grv = calculate_temporal_statistics(df, var_name, year_column, 'GRV')
            
            all_stats[var_name] = {
                'GAG': stats_gag,
                'GRV': stats_grv
            }
            
            print(f"\n  • {display_name}:")
            
            # GAG stats
            if stats_gag is not None and stats_gag['trend_stats'] is not None:
                ts = stats_gag['trend_stats']
                sig = "✓ Sig" if ts['significant'] else "✗ NS"
                print(f"      GAG: {sig} (p={ts['p_value']:.4f}, R²={ts['r_squared']:.3f}, Δ={ts['percent_change']:+.1f}%)")
            else:
                print(f"      GAG: Insufficient data")
            
            # GRV stats
            if stats_grv is not None and stats_grv['trend_stats'] is not None:
                ts = stats_grv['trend_stats']
                sig = "✓ Sig" if ts['significant'] else "✗ NS"
                print(f"      GRV: {sig} (p={ts['p_value']:.4f}, R²={ts['r_squared']:.3f}, Δ={ts['percent_change']:+.1f}%)")
            else:
                print(f"      GRV: Insufficient data")
        
        # Create figure with custom layout
        print(f"\nCreating {n_rows}×{n_cols_group} grid plot...")
        
        plt.style.use('seaborn-v0_8-white')
        
        # Adjust figure size based on layout
        fig_width = n_cols_group * 6.5
        fig_height = n_rows * 4.5
        
        fig, axes = plt.subplots(n_rows, n_cols_group, figsize=(fig_width, fig_height))
        
        # Handle different subplot array shapes
        if n_rows == 1 and n_cols_group == 1:
            axes = np.array([[axes]])
        elif n_rows == 1:
            axes = axes.reshape(1, -1)
        elif n_cols_group == 1:
            axes = axes.reshape(-1, 1)
        
        axes_flat = axes.flatten()
        
        # Plot each variable
        for idx, (var_name, var_info) in enumerate(variables.items()):
            if var_name not in all_stats:
                continue
            
            ax = axes_flat[idx]
            stats_gag = all_stats[var_name]['GAG']
            stats_grv = all_stats[var_name]['GRV']
            
            plot_dual_trends(ax, stats_gag, stats_grv, var_name, var_info, colors)
        
        # Hide unused subplots
        for idx in range(len(all_stats), len(axes_flat)):
            axes_flat[idx].set_visible(False)
        
        # Overall title
        title_map = {
            'Granulometry': 'Granulométrie',
            'Exchangeable_Elements': 'Éléments Échangeables',
            'Other_Properties_Part1': 'Autres Propriétés du Sol - Partie 1',
            'Other_Properties_Part2': 'Autres Propriétés du Sol - Partie 2'
        }
        
        # Use custom title if provided, otherwise use default
        if 'group_title' in group_info:
            plot_title = f"{group_info['group_title']} - GAG vs GRV"
        else:
            plot_title = f'{title_map.get(group_name, group_name)} - GAG vs GRV'
        
        fig.suptitle(plot_title, fontsize=16, fontweight='bold', y=0.995)
        
        # Adjust layout
        plt.tight_layout(rect=[0, 0.02, 1, 0.99], h_pad=3.0, w_pad=2.5)
        
        # Save combined figure
        print(f"\nSaving combined figure to: {output_file}")
        plt.savefig(output_file, dpi=dpi, bbox_inches='tight', 
                    facecolor='white', edgecolor='none')
        print(f"✓ Combined PNG saved at {dpi} DPI")
        
        # Also save as PDF
        pdf_file = output_file.replace('.png', '.pdf')
        plt.savefig(pdf_file, format='pdf', bbox_inches='tight',
                    facecolor='white', edgecolor='none')
        print(f"✓ Combined PDF saved: {pdf_file}")
        
        plt.close()
        
        # Save individual plots
        print(f"\nSaving individual plots to: {output_subdir}")
        
        for var_name, var_info in variables.items():
            if var_name not in all_stats:
                continue
            
            stats_gag = all_stats[var_name]['GAG']
            stats_grv = all_stats[var_name]['GRV']
            display_name = var_info['name']
            
            png_path, pdf_path = save_individual_plot(
                stats_gag, stats_grv, var_name, var_info, 
                output_subdir, colors, dpi
            )
            
            print(f"  ✓ {display_name}: {os.path.basename(png_path)}")
        
        # Summary statistics for this group
        print(f"\n{'='*80}")
        print(f"SUMMARY STATISTICS - {group_name}")
        print('='*80)
        
        summary_data = []
        for var_name, var_info in variables.items():
            if var_name not in all_stats:
                continue
            
            for irr_type in ['GAG', 'GRV']:
                stats_dict = all_stats[var_name][irr_type]
                
                if stats_dict is not None and stats_dict['trend_stats'] is not None:
                    trend_stats = stats_dict['trend_stats']
                    
                    summary_data.append({
                        'Variable': var_info['name'],
                        'Irrigation': irr_type,
                        'Pente': f"{trend_stats['slope']:.4f}",
                        'R²': f"{trend_stats['r_squared']:.3f}",
                        'Valeur-P': f"{trend_stats['p_value']:.4e}",
                        'Significatif': 'Oui' if trend_stats['significant'] else 'Non',
                        'Δ (%)': f"{trend_stats['percent_change']:+.1f}" if not np.isnan(trend_stats['percent_change']) else 'N/A',
                    })
        
        if len(summary_data) > 0:
            summary_df = pd.DataFrame(summary_data)
            print(summary_df.to_string(index=False))
            
            # Save summary
            summary_file = output_file.replace('.png', '_summary.csv')
            summary_df.to_csv(summary_file, index=False)
            print(f"\n✓ Summary statistics saved: {summary_file}")
        else:
            print("No statistics available for this group")
    

# ============================================================================
# MAIN EXECUTION
# ============================================================================

if __name__ == "__main__":
    
    create_soil_properties_plots(
        data_file=DATA_FILE,
        property_groups=SOIL_PROPERTY_GROUPS,
        output_dir_base=OUTPUT_DIR_BASE,
        dpi=OUTPUT_DPI,
        colors=COLORS,
        n_cols=N_COLS
    )
    
    
    print("\n📊 GROUP-SPECIFIC COMBINED PLOTS WITH CUSTOM LAYOUTS:")
    for group_name, group_info in SOIL_PROPERTY_GROUPS.items():
        layout = group_info['layout']
        print(f"  • {group_name} ({layout['rows']}×{layout['cols']} grid):")
        print(f"      - {group_info['output_file']}")
        print(f"      - {group_info['output_file'].replace('.png', '.pdf')}")
        print(f"      - {group_info['output_file'].replace('.png', '_summary.csv')}")
    
    print(f"\n📈 INDIVIDUAL PLOTS:")
    # Count total variables
    total_vars = 0
    for group_name, group_info in SOIL_PROPERTY_GROUPS.items():
        n_vars = len(group_info['variables'])
        total_vars += n_vars
    
    print(f"  • Total: {total_vars * 2} files ({total_vars} PNG + {total_vars} PDF)")
    for group_name, group_info in SOIL_PROPERTY_GROUPS.items():
        n_vars = len(group_info['variables'])
        print(f"      {group_name}: {n_vars} variables in {OUTPUT_DIR_BASE}/{group_info['output_subdir']}")
    
    print("\n" + "="*80)
```

# SOIL PROPERTIES TEMPORAL TRENDS BY BLOC AND IRRIGATION TYPE

```python
"""
SOIL PROPERTIES TEMPORAL TRENDS BY BLOC AND IRRIGATION TYPE
===========================================================
"""

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats
import seaborn as sns
import warnings
import os
warnings.filterwarnings('ignore')

# ============================================================================
# CONFIGURATION
# ============================================================================

# Path to your data (ADJUST THIS)
DATA_FILE = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/final_dataset_regrouped_categorical_with_field_id.csv'

# Output configuration
OUTPUT_DIR = 'soil_properties_by_bloc'
OUTPUT_DPI = 600

# Soil property groups with display names and units
SOIL_PROPERTY_GROUPS = {
    'Granulometry': {
        'variables': {
            'Clay_percent_complete': {'name': 'Clay (%)', 'unit': '% year⁻¹'},
            'Fine_Silt_percent_complete': {'name': 'Fine Silt (%)', 'unit': '% year⁻¹'},
            'Coarse_Silt_percent_complete': {'name': 'Coarse Silt (%)', 'unit': '% year⁻¹'},
            'Fine_Sand_percent_complete': {'name': 'Fine Sand (%)', 'unit': '% year⁻¹'},
            'Coarse_Sand_percent_complete': {'name': 'Coarse Sand (%)', 'unit': '% year⁻¹'}
        }
    },
    'Exchangeable_Elements': {
        'variables': {
            'Exchangeable_Calcium_meq_100g_complete': {'name': 'Exch. Ca (meq/100g)', 'unit': 'meq/100g year⁻¹'},
            'Exchangeable_Magnesium_meq_100g_complete': {'name': 'Exch. Mg (meq/100g)', 'unit': 'meq/100g year⁻¹'},
            'Exchangeable_Potassium_meq_100g_complete': {'name': 'Exch. K (meq/100g)', 'unit': 'meq/100g year⁻¹'},
            'Exchangeable_Sodium_meq_100g_complete': {'name': 'Exch. Na (meq/100g)', 'unit': 'meq/100g year⁻¹'}
        }
    },
    'Other_Properties': {
        'variables': {
            'pH_complete': {'name': 'pH', 'unit': 'pH year⁻¹'},
            'Organic_Carbon_per_mille_complete': {'name': 'Organic C (‰)', 'unit': '‰ year⁻¹'},
            'Total_Nitrogen_per_mille_complete': {'name': 'Total N (‰)', 'unit': '‰ year⁻¹'},
            'Available_P2O5_mg_kg_complete': {'name': 'Available P₂O₅ (mg/kg)', 'unit': 'mg/kg year⁻¹'},
            'Electrical_Conductivity_mmho_cm_complete': {'name': 'EC (mmho/cm)', 'unit': 'mmho/cm year⁻¹'},
            'Mg_Ca_Ratio_complete': {'name': 'Mg/Ca Ratio', 'unit': 'year⁻¹'},
            'Mg_K_Ratio_complete': {'name': 'Mg/K Ratio', 'unit': 'year⁻¹'},
            'Compaction_Index_complete': {'name': 'Compaction Index', 'unit': 'year⁻¹'}
        }
    }
}

# Visual styling - GAG in blue, GRV in red
COLORS = {
    'GAG': {
        'data_points': '#2E86AB',  # Blue
        'trend_line': '#1E5F8C',
        'error_bars': '#2E86AB',
    },
    'GRV': {
        'data_points': '#E63946',  # Red
        'trend_line': '#C1121F',
        'error_bars': '#E63946',
    }
}

# Grid layout: 4 columns
N_COLS = 4

# ============================================================================
# HELPER FUNCTIONS
# ============================================================================

def calculate_temporal_statistics(df, response_var, year_column, irrigation_type):
    """
    Calculate temporal statistics for a response variable filtered by irrigation type
    
    Parameters:
    -----------
    df : DataFrame
    response_var : str
    year_column : str
    irrigation_type : str ('GAG' or 'GRV')
    
    Returns:
    --------
    dict with yearly_means, yearly_std, trend_stats, percent_change or None
    """
    # Filter by irrigation type
    df_filtered = df[df['irrigation_type'] == irrigation_type].copy()
    
    # Remove missing values
    df_clean = df_filtered[[year_column, response_var]].dropna()
    
    if len(df_clean) < 3:
        return None
    
    # Calculate yearly statistics
    yearly_stats = df_clean.groupby(year_column)[response_var].agg([
        ('mean', 'mean'),
        ('std', 'std'),
        ('count', 'count')
    ]).reset_index()
    
    years = yearly_stats[year_column].values
    means = yearly_stats['mean'].values
    stds = yearly_stats['std'].values
    
    # Linear regression
    if len(years) >= 3:
        slope, intercept, r_value, p_value, std_err = stats.linregress(years, means)
        
        # Calculate percentage change over entire period
        first_value = means[0]
        last_value = means[-1]
        
        if abs(first_value) > 1e-10:
            percent_change = ((last_value - first_value) / abs(first_value)) * 100
        else:
            percent_change = np.nan
        
        absolute_change = last_value - first_value
        
        trend_stats = {
            'slope': slope,
            'intercept': intercept,
            'r_squared': r_value**2,
            'p_value': p_value,
            'significant': p_value < 0.05,
            'percent_change': percent_change,
            'absolute_change': absolute_change,
            'years_span': years[-1] - years[0],
            'first_value': first_value,
            'last_value': last_value
        }
    else:
        trend_stats = None
    
    return {
        'years': years,
        'means': means,
        'stds': stds,
        'yearly_stats': yearly_stats,
        'trend_stats': trend_stats,
        'irrigation_type': irrigation_type
    }

def format_pvalue(p):
    """Format p-value for display"""
    if p < 0.001:
        return 'p < 0.001'
    elif p < 0.01:
        return f'p < 0.01'
    elif p < 0.05:
        return f'p = {p:.3f}'
    else:
        return f'p = {p:.2f}'

def plot_dual_trends(ax, stats_gag, stats_grv, var_name, var_info, colors):
    """
    Plot temporal trends for both GAG and GRV on the same axis
    
    Parameters:
    -----------
    ax : matplotlib axis
    stats_gag : dict from calculate_temporal_statistics for GAG
    stats_grv : dict from calculate_temporal_statistics for GRV
    var_name : str (variable key)
    var_info : dict with 'name' and 'unit'
    colors : dict
    """
    display_name = var_info['name']
    unit = var_info['unit']
    
    # Plot GAG data (Blue)
    if stats_gag is not None:
        years_gag = stats_gag['years']
        means_gag = stats_gag['means']
        stds_gag = stats_gag['stds']
        
        ax.errorbar(years_gag, means_gag, yerr=stds_gag, 
                    fmt='o', markersize=5, 
                    color=colors['GAG']['data_points'],
                    ecolor=colors['GAG']['error_bars'],
                    elinewidth=1.2,
                    capsize=3,
                    capthick=1.2,
                    alpha=0.8,
                    zorder=3,
                    label='GAG')
        
        ax.plot(years_gag, means_gag, '-', 
                color=colors['GAG']['data_points'], 
                alpha=0.3, 
                linewidth=1,
                zorder=2)
        
        # Plot GAG trend line
        if stats_gag['trend_stats'] is not None:
            trend_line_gag = stats_gag['trend_stats']['slope'] * years_gag + stats_gag['trend_stats']['intercept']
            ax.plot(years_gag, trend_line_gag, '--', 
                    color=colors['GAG']['trend_line'],
                    linewidth=2,
                    alpha=0.9,
                    zorder=4)
    
    # Plot GRV data (Red)
    if stats_grv is not None:
        years_grv = stats_grv['years']
        means_grv = stats_grv['means']
        stds_grv = stats_grv['stds']
        
        ax.errorbar(years_grv, means_grv, yerr=stds_grv, 
                    fmt='s', markersize=5, 
                    color=colors['GRV']['data_points'],
                    ecolor=colors['GRV']['error_bars'],
                    elinewidth=1.2,
                    capsize=3,
                    capthick=1.2,
                    alpha=0.8,
                    zorder=3,
                    label='GRV')
        
        ax.plot(years_grv, means_grv, '-', 
                color=colors['GRV']['data_points'], 
                alpha=0.3, 
                linewidth=1,
                zorder=2)
        
        # Plot GRV trend line
        if stats_grv['trend_stats'] is not None:
            trend_line_grv = stats_grv['trend_stats']['slope'] * years_grv + stats_grv['trend_stats']['intercept']
            ax.plot(years_grv, trend_line_grv, '--', 
                    color=colors['GRV']['trend_line'],
                    linewidth=2,
                    alpha=0.9,
                    zorder=4)
    
    # Add statistics text boxes
    # GAG statistics (top-left)
    if stats_gag is not None and stats_gag['trend_stats'] is not None:
        ts = stats_gag['trend_stats']
        r2_str = f"R² = {ts['r_squared']:.3f}"
        p_val_str = format_pvalue(ts['p_value'])
        pct_change = ts['percent_change']
        
        if not np.isnan(pct_change):
            pct_sign = '+' if pct_change > 0 else ''
            pct_str = f"Δ = {pct_sign}{pct_change:.1f}%"
        else:
            pct_str = "Δ = N/A"
        
        stats_text_gag = f"GAG\n{r2_str}\n{p_val_str}\n{pct_str}"
        
        bbox_props_gag = dict(
            boxstyle='round,pad=0.3',
            facecolor='white',
            edgecolor=colors['GAG']['trend_line'],
            linewidth=1.5,
            alpha=0.95
        )
        
        ax.text(0.05, 0.95, stats_text_gag,
                transform=ax.transAxes,
                fontsize=7.5,
                verticalalignment='top',
                horizontalalignment='left',
                bbox=bbox_props_gag,
                fontweight='bold',
                family='monospace')
    
    # GRV statistics (top-right)
    if stats_grv is not None and stats_grv['trend_stats'] is not None:
        ts = stats_grv['trend_stats']
        r2_str = f"R² = {ts['r_squared']:.3f}"
        p_val_str = format_pvalue(ts['p_value'])
        pct_change = ts['percent_change']
        
        if not np.isnan(pct_change):
            pct_sign = '+' if pct_change > 0 else ''
            pct_str = f"Δ = {pct_sign}{pct_change:.1f}%"
        else:
            pct_str = "Δ = N/A"
        
        stats_text_grv = f"GRV\n{r2_str}\n{p_val_str}\n{pct_str}"
        
        bbox_props_grv = dict(
            boxstyle='round,pad=0.3',
            facecolor='white',
            edgecolor=colors['GRV']['trend_line'],
            linewidth=1.5,
            alpha=0.95
        )
        
        ax.text(0.95, 0.95, stats_text_grv,
                transform=ax.transAxes,
                fontsize=7.5,
                verticalalignment='top',
                horizontalalignment='right',
                bbox=bbox_props_grv,
                fontweight='bold',
                family='monospace')
    
    # Formatting
    ax.set_xlabel('Année', fontsize=10, fontweight='bold')
    ax.set_ylabel(display_name, fontsize=10, fontweight='bold')
    ax.set_title(display_name, fontsize=11, fontweight='bold', pad=8)
    
    # Grid
    ax.grid(True, alpha=0.3, linestyle='--', linewidth=0.5, zorder=1)
    ax.set_axisbelow(True)
    
    # Legend
    ax.legend(loc='lower left', frameon=True, framealpha=0.9, 
              edgecolor='black', fontsize=8, ncol=2)
    
    # Tick parameters
    ax.tick_params(axis='both', which='major', labelsize=9, width=1.2, length=4)
    
    # Set integer years on x-axis
    all_years = []
    if stats_gag is not None:
        all_years.extend(stats_gag['years'])
    if stats_grv is not None:
        all_years.extend(stats_grv['years'])
    
    if len(all_years) > 0:
        unique_years = sorted(set(all_years))
        ax.set_xticks(unique_years)
        ax.set_xticklabels([int(y) for y in unique_years], rotation=45, ha='right')
    
    # Spine formatting
    for spine in ax.spines.values():
        spine.set_linewidth(1.2)
        spine.set_color('black')

# ============================================================================
# MAIN PLOTTING FUNCTION FOR EACH BLOC
# ============================================================================

def create_bloc_soil_properties_plot(df_bloc, bloc_name, property_groups, 
                                    output_dir, year_column, dpi=600, 
                                    colors=COLORS, n_cols=N_COLS):
    """
    Create combined soil properties temporal trends plot for a specific Bloc
    with ALL 17 soil properties in one figure (GAG vs GRV)
    """
    
    print(f"\n{'='*80}")
    print(f"PROCESSING BLOC: {bloc_name}")
    print('='*80)
    
    # Check data availability
    print(f"Total observations in Bloc {bloc_name}: {len(df_bloc)}")
    print(f"  GAG: {len(df_bloc[df_bloc['irrigation_type'] == 'GAG'])}")
    print(f"  GRV: {len(df_bloc[df_bloc['irrigation_type'] == 'GRV'])}")
    
    # Collect all variables from all groups
    all_variables = {}
    for group_name, group_info in property_groups.items():
        all_variables.update(group_info['variables'])
    
    print(f"\nTotal soil properties to plot: {len(all_variables)}")
    
    # Calculate statistics for all variables
    print(f"\nCalculating temporal statistics for Bloc {bloc_name}...")
    
    all_stats = {}
    
    for var_name, var_info in all_variables.items():
        display_name = var_info['name']
        
        if var_name not in df_bloc.columns:
            print(f"  ⚠ Warning: {var_name} not found in dataset, skipping...")
            continue
        
        # Calculate for GAG
        stats_gag = calculate_temporal_statistics(df_bloc, var_name, year_column, 'GAG')
        
        # Calculate for GRV
        stats_grv = calculate_temporal_statistics(df_bloc, var_name, year_column, 'GRV')
        
        all_stats[var_name] = {
            'GAG': stats_gag,
            'GRV': stats_grv
        }
        
        print(f"\n  • {display_name}:")
        
        # GAG stats
        if stats_gag is not None and stats_gag['trend_stats'] is not None:
            ts = stats_gag['trend_stats']
            sig = "✓ Sig" if ts['significant'] else "✗ NS"
            print(f"      GAG: {sig} (p={ts['p_value']:.4f}, R²={ts['r_squared']:.3f}, Δ={ts['percent_change']:+.1f}%)")
        else:
            print(f"      GAG: Insufficient data")
        
        # GRV stats
        if stats_grv is not None and stats_grv['trend_stats'] is not None:
            ts = stats_grv['trend_stats']
            sig = "✓ Sig" if ts['significant'] else "✗ NS"
            print(f"      GRV: {sig} (p={ts['p_value']:.4f}, R²={ts['r_squared']:.3f}, Δ={ts['percent_change']:+.1f}%)")
        else:
            print(f"      GRV: Insufficient data")
    
    # Skip if no valid data
    if len(all_stats) == 0:
        print(f"\n⚠ No valid data for Bloc {bloc_name}, skipping plot generation.")
        return None, None, None
    
    # Create figure with subplots (4 columns)
    n_vars = len(all_stats)
    n_rows = int(np.ceil(n_vars / n_cols))
    
    print(f"\nCreating {n_rows}×{n_cols} grid plot for Bloc {bloc_name}...")
    
    plt.style.use('seaborn-v0_8-white')
    
    fig_width = 20
    fig_height = n_rows * 4
    
    fig, axes = plt.subplots(n_rows, n_cols, figsize=(fig_width, fig_height))
    
    # Handle different subplot array shapes
    if n_rows == 1 and n_cols == 1:
        axes = np.array([[axes]])
    elif n_rows == 1:
        axes = axes.reshape(1, -1)
    elif n_cols == 1:
        axes = axes.reshape(-1, 1)
    
    axes_flat = axes.flatten()
    
    # Plot each variable
    for idx, (var_name, var_info) in enumerate(all_variables.items()):
        if var_name not in all_stats:
            continue
        
        ax = axes_flat[idx]
        stats_gag = all_stats[var_name]['GAG']
        stats_grv = all_stats[var_name]['GRV']
        
        plot_dual_trends(ax, stats_gag, stats_grv, var_name, var_info, colors)
    
    # Hide unused subplots
    for idx in range(len(all_stats), len(axes_flat)):
        axes_flat[idx].set_visible(False)
    
    # Overall title
    fig.suptitle(f'Propriétés du Sol - Bloc {bloc_name} (GAG vs GRV)',
                 fontsize=18, fontweight='bold', y=0.995)
    
    # Adjust layout
    plt.tight_layout(rect=[0, 0.02, 1, 0.99], h_pad=3.0, w_pad=2.5)
    
    # Save figure
    output_file = os.path.join(output_dir, f'bloc_{bloc_name}_soil_properties.png')
    print(f"\nSaving figure to: {output_file}")
    plt.savefig(output_file, dpi=dpi, bbox_inches='tight', 
                facecolor='white', edgecolor='none')
    print(f"✓ PNG saved at {dpi} DPI")
    
    # Also save as PDF
    pdf_file = os.path.join(output_dir, f'bloc_{bloc_name}_soil_properties.pdf')
    plt.savefig(pdf_file, format='pdf', bbox_inches='tight',
                facecolor='white', edgecolor='none')
    print(f"✓ PDF saved: {pdf_file}")
    
    plt.close()
    
    # Summary statistics
    print(f"\nSUMMARY STATISTICS - Bloc {bloc_name}")
    print('='*80)
    
    summary_data = []
    for var_name, var_info in all_variables.items():
        if var_name not in all_stats:
            continue
        
        for irr_type in ['GAG', 'GRV']:
            stats_dict = all_stats[var_name][irr_type]
            
            if stats_dict is not None and stats_dict['trend_stats'] is not None:
                trend_stats = stats_dict['trend_stats']
                
                summary_data.append({
                    'Bloc': bloc_name,
                    'Variable': var_info['name'],
                    'Irrigation': irr_type,
                    'Pente': f"{trend_stats['slope']:.4f}",
                    'R²': f"{trend_stats['r_squared']:.3f}",
                    'Valeur-P': f"{trend_stats['p_value']:.4e}",
                    'Significatif': 'Oui' if trend_stats['significant'] else 'Non',
                    'Δ (%)': f"{trend_stats['percent_change']:+.1f}" if not np.isnan(trend_stats['percent_change']) else 'N/A',
                })
    
    summary_df = pd.DataFrame(summary_data)
    if len(summary_df) > 0:
        print(summary_df.to_string(index=False))
    
    # Save summary
    summary_file = os.path.join(output_dir, f'bloc_{bloc_name}_soil_properties_summary.csv')
    summary_df.to_csv(summary_file, index=False)
    print(f"\n✓ Summary statistics saved: {summary_file}")
    
    return fig, all_stats, summary_df

# ============================================================================
# MAIN EXECUTION FUNCTION
# ============================================================================

def create_all_bloc_soil_properties_plots(data_file, property_groups, output_dir, 
                                         dpi=600, colors=COLORS, n_cols=N_COLS):
    """
    Create soil properties temporal trend plots for all Blocs
    """
    print("="*80)
    print("SOIL PROPERTIES TEMPORAL TRENDS BY BLOC AND IRRIGATION TYPE")
    print("="*80)
    
    # Load data
    print(f"\nLoading data from: {data_file}")
    df = pd.read_csv(data_file)
    
    # Find year column
    year_column = None
    for col in ['campaign_year', 'Harvest_Year', 'year', 'Campaign_Year']:
        if col in df.columns:
            year_column = col
            break
    
    if year_column is None:
        raise ValueError("No year column found in dataset!")
    
    # Check required columns
    if 'irrigation_type' not in df.columns:
        raise ValueError("irrigation_type column not found in dataset!")
    
    if 'Bloc' not in df.columns:
        raise ValueError("Bloc column not found in dataset!")
    
    print(f"\nYear column: {year_column}")
    print(f"Years available: {sorted(df[year_column].unique())}")
    print(f"Total observations: {len(df)}")
    
    # Get unique blocs
    blocs = sorted(df['Bloc'].unique())
    print(f"\nBlocs found: {blocs}")
    print(f"Number of Blocs: {len(blocs)}")
    
    # Print data distribution
    print("\nData distribution by Bloc and irrigation type:")
    distribution = df.groupby(['Bloc', 'irrigation_type']).size().unstack(fill_value=0)
    print(distribution)
    
    # Create output directory
    os.makedirs(output_dir, exist_ok=True)
    print(f"\nOutput directory: {output_dir}")
    
    # Count total soil properties
    total_properties = sum(len(group['variables']) for group in property_groups.values())
    print(f"\nTotal soil properties: {total_properties}")
    
    # Process each Bloc
    all_summaries = []
    
    for bloc in blocs:
        df_bloc = df[df['Bloc'] == bloc].copy()
        
        fig, stats, summary_df = create_bloc_soil_properties_plot(
            df_bloc=df_bloc,
            bloc_name=bloc,
            property_groups=property_groups,
            output_dir=output_dir,
            year_column=year_column,
            dpi=dpi,
            colors=colors,
            n_cols=n_cols
        )
        
        if summary_df is not None and len(summary_df) > 0:
            all_summaries.append(summary_df)
    
    # Combine all summaries
    if len(all_summaries) > 0:
        combined_summary = pd.concat(all_summaries, ignore_index=True)
        combined_summary_file = os.path.join(output_dir, 'all_blocs_soil_properties_summary.csv')
        combined_summary.to_csv(combined_summary_file, index=False)
        print(f"\n{'='*80}")
        print("COMBINED SUMMARY FOR ALL BLOCS - SOIL PROPERTIES")
        print('='*80)
        print(combined_summary.to_string(index=False))
        print(f"\n✓ Combined summary saved: {combined_summary_file}")
    
    print("\n" + "="*80)
    print("VISUALIZATION COMPLETE!")
    print("="*80)
    print(f"\nGenerated files in: {output_dir}")
    print(f"  • {len(blocs)} combined PNG plots (one per Bloc, {total_properties} variables each)")
    print(f"  • {len(blocs)} PDF plots (one per Bloc)")
    print(f"  • {len(blocs)} CSV summaries (one per Bloc)")
    print(f"  • 1 combined summary CSV (all Blocs)")
    print(f"\nTotal files: {len(blocs) * 3 + 1}")

# ============================================================================
# MAIN EXECUTION
# ============================================================================

if __name__ == "__main__":
    
    create_all_bloc_soil_properties_plots(
        data_file=DATA_FILE,
        property_groups=SOIL_PROPERTY_GROUPS,
        output_dir=OUTPUT_DIR,
        dpi=OUTPUT_DPI,
        colors=COLORS,
        n_cols=N_COLS
    )
    
    print("\n" + "="*80)
```

# DESCRIPTIVE STATISTICS FOR SUGARCANE DATASET

```python
"""
DESCRIPTIVE STATISTICS FOR SUGARCANE DATASET
================================================================================
"""

import pandas as pd
import numpy as np

# ============================================================================
# CONFIGURATION
# ============================================================================

DATA_FILE = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/final_dataset_regrouped_categorical.csv'

# All features for analysis
features = [
    'tc_per_ha_actual', 'brix_percent',
    'pH_complete', 'Exchangeable_Calcium_meq_100g_complete', 'Exchangeable_Magnesium_meq_100g_complete',
    'Exchangeable_Potassium_meq_100g_complete', 'Clay_percent_complete', 'Fine_Silt_percent_complete', 
    'Coarse_Silt_percent_complete', 'Fine_Sand_percent_complete', 'Coarse_Sand_percent_complete', 
    'Total_Nitrogen_per_mille_complete', 'Organic_Carbon_per_mille_complete', 'Mg_K_Ratio_complete', 
    'Mg_Ca_Ratio_complete', 'SC_drying_duration_actual_days', 'growing_season_rainfall_mm', 'temp_mini_mean', 
    'temp_maxi_mean', 'temp_moy_mean', 'ampl_mean', 'h_mini_mean', 'h_maxi_mean', 'h_moy_mean', 'h_6h_mean', 
    'h_12h_mean', 'h_16h_mean', 'h_24h_mean', 'soil_temp_6h_mean', 'soil_temp_16h_mean', 'radiation_total', 
    'radiation_ERA_total', 'evaporation_total', 'growing_period_days_weather', 'CWD', 'CDD', 'SDII', 'R10', 
    'R20', 'Rx1', 'Rx5', 'R95p', 'R99p', 'HD', 'TNn', 'TNx', 'TXn', 'TXx', 'tc_per_ha_per_week', 
    'tc_per_ha_52_weeks', 'pol_juice_percent', 'purity_percent', 'fiber_percent', 'pol_cane_percent', 
    'glucose_cane_percent', 'hugot_sugar', 'estimated_recoverable_pol', 'CN_Ratio_complete', 
    'Available_P2O5_mg_kg_complete', 'Compaction_Index_complete', 'Electrical_Conductivity_mmho_cm_complete', 
    'Exchangeable_Sodium_meq_100g_complete', 'total_recoverable_sugar_t_per_ha'
]

# ============================================================================
# LOAD DATA AND CREATE DERIVED VARIABLES
# ============================================================================

print("="*100)
print("DESCRIPTIVE STATISTICS FOR SUGARCANE DATASET")
print("="*100)

print(f"\nLoading data from: {DATA_FILE}")
df = pd.read_csv(DATA_FILE)

print(f"Total observations: {len(df)}")

# Create Total Sand and Total Silt
print("\n" + "="*100)
print("CREATING DERIVED VARIABLES")
print("="*100)

if 'Fine_Sand_percent_complete' in df.columns and 'Coarse_Sand_percent_complete' in df.columns:
    df['Total_Sand_percent'] = df['Fine_Sand_percent_complete'] + df['Coarse_Sand_percent_complete']
    print("✓ Total_Sand_percent created (Fine_Sand + Coarse_Sand)")
    # Add to features list
    features.append('Total_Sand_percent')
else:
    print("⚠ Could not create Total_Sand_percent - missing columns")

if 'Fine_Silt_percent_complete' in df.columns and 'Coarse_Silt_percent_complete' in df.columns:
    df['Total_Silt_percent'] = df['Fine_Silt_percent_complete'] + df['Coarse_Silt_percent_complete']
    print("✓ Total_Silt_percent created (Fine_Silt + Coarse_Silt)")
    # Add to features list
    features.append('Total_Silt_percent')
else:
    print("⚠ Could not create Total_Silt_percent - missing columns")

# ============================================================================
# COMPUTE DESCRIPTIVE STATISTICS
# ============================================================================

print("\n" + "="*100)
print("COMPUTING DESCRIPTIVE STATISTICS")
print("="*100)

# Filter to only features that exist in the dataset
available_features = [f for f in features if f in df.columns]
missing_features = [f for f in features if f not in df.columns]

if missing_features:
    print(f"\n⚠ Warning: {len(missing_features)} features not found in dataset:")
    for feat in missing_features[:10]:  # Show first 10
        print(f"  - {feat}")
    if len(missing_features) > 10:
        print(f"  ... and {len(missing_features) - 10} more")

print(f"\nAnalyzing {len(available_features)} features...")

# Create statistics dataframe
stats_data = []

for feat in available_features:
    data = df[feat].dropna()
    
    if len(data) > 0:
        stats_data.append({
            'Feature': feat,
            'N': len(data),
            'Mean': data.mean(),
            'Std': data.std(),
            'Min': data.min(),
            'Median': data.median(),
            'Max': data.max(),
            'Missing': df[feat].isna().sum(),
            'Missing_%': (df[feat].isna().sum() / len(df)) * 100
        })

stats_df = pd.DataFrame(stats_data)

# ============================================================================
# DISPLAY RESULTS
# ============================================================================

print("\n" + "="*100)
print("DESCRIPTIVE STATISTICS SUMMARY")
print("="*100)
print(f"\nTotal Features: {len(stats_df)}")
print(f"Total Observations: {len(df)}")

print("\n" + "-"*100)
print(f"{'Feature':<50} {'N':>8} {'Mean ± Std':>20} {'Min':>12} {'Median':>12} {'Max':>12} {'Missing':>10}")
print("-"*100)

for idx, row in stats_df.iterrows():
    # Format Mean ± Std
    mean_std = f"{row['Mean']:.2f} ± {row['Std']:.2f}"
    
    # Format feature name (truncate if too long)
    feat_name = row['Feature'][:48] if len(row['Feature']) > 48 else row['Feature']
    
    print(f"{feat_name:<50} {row['N']:>8.0f} {mean_std:>20} {row['Min']:>12.2f} {row['Median']:>12.2f} {row['Max']:>12.2f} {row['Missing']:>10.0f}")

print("-"*100)

# ============================================================================
# HIGHLIGHT DERIVED VARIABLES
# ============================================================================

print("\n" + "="*100)
print("DERIVED VARIABLES SUMMARY")
print("="*100)

derived_vars = ['Total_Sand_percent', 'Total_Silt_percent']
derived_stats = stats_df[stats_df['Feature'].isin(derived_vars)]

if not derived_stats.empty:
    print("\n" + "-"*100)
    print(f"{'Variable':<30} {'N':>8} {'Mean ± Std':>25} {'Min':>12} {'Median':>12} {'Max':>12}")
    print("-"*100)
    
    for idx, row in derived_stats.iterrows():
        mean_std = f"{row['Mean']:.2f} ± {row['Std']:.2f}"
        print(f"{row['Feature']:<30} {row['N']:>8.0f} {mean_std:>25} {row['Min']:>12.2f} {row['Median']:>12.2f} {row['Max']:>12.2f}")
    
    print("-"*100)
else:
    print("\n⚠ No derived variables found in dataset")

# ============================================================================
# CATEGORY BREAKDOWN (if needed)
# ============================================================================

print("\n" + "="*100)
print("KEY VARIABLES BREAKDOWN")
print("="*100)

# Show breakdown for main output variables
key_vars = ['tc_per_ha_actual', 'hugot_sugar', 'total_recoverable_sugar_t_per_ha']

for var in key_vars:
    if var in stats_df['Feature'].values:
        row = stats_df[stats_df['Feature'] == var].iloc[0]
        
        print(f"\n{var}:")
        print(f"  N observations: {row['N']:.0f}")
        print(f"  Mean ± Std: {row['Mean']:.2f} ± {row['Std']:.2f}")
        print(f"  Range: [{row['Min']:.2f}, {row['Max']:.2f}]")
        print(f"  Median: {row['Median']:.2f}")
        print(f"  Missing: {row['Missing']:.0f} ({row['Missing_%']:.1f}%)")

# ============================================================================
# SOIL TEXTURE SUMMARY
# ============================================================================

print("\n" + "="*100)
print("SOIL TEXTURE COMPONENTS SUMMARY")
print("="*100)

soil_vars = ['Clay_percent_complete', 'Total_Silt_percent', 'Total_Sand_percent']
soil_stats = stats_df[stats_df['Feature'].isin(soil_vars)]

if not soil_stats.empty:
    print("\n" + "-"*100)
    print(f"{'Component':<30} {'N':>8} {'Mean ± Std':>25} {'Min':>12} {'Median':>12} {'Max':>12}")
    print("-"*100)
    
    for var in soil_vars:
        if var in soil_stats['Feature'].values:
            row = soil_stats[soil_stats['Feature'] == var].iloc[0]
            mean_std = f"{row['Mean']:.2f} ± {row['Std']:.2f}"
            print(f"{var:<30} {row['N']:>8.0f} {mean_std:>25} {row['Min']:>12.2f} {row['Median']:>12.2f} {row['Max']:>12.2f}")
    
    print("-"*100)
    
    # Check if soil components sum to ~100%
    if all(var in df.columns for var in soil_vars):
        df_soil = df[soil_vars].dropna()
        if len(df_soil) > 0:
            soil_sum = df_soil.sum(axis=1)
            print(f"\nSoil texture sum (Clay + Silt + Sand):")
            print(f"  Mean: {soil_sum.mean():.2f}%")
            print(f"  Range: [{soil_sum.min():.2f}%, {soil_sum.max():.2f}%]")

# ============================================================================
# SAVE RESULTS TO CSV (OPTIONAL)
# ============================================================================

output_file = 'descriptive_statistics_summary.csv'
stats_df.to_csv(output_file, index=False)
print("\n" + "="*100)
print(f"✓ Descriptive statistics saved to: {output_file}")
print("="*100)

# ============================================================================
# SUMMARY BY CATEGORIES
# ============================================================================

print("\n" + "="*100)
print("MISSING DATA SUMMARY")
print("="*100)

# Sort by missing percentage
stats_df_sorted = stats_df.sort_values('Missing_%', ascending=False)

print("\nTop 10 features with most missing data:")
print("-"*100)
print(f"{'Feature':<50} {'Missing':>10} {'Missing_%':>12}")
print("-"*100)

for idx, row in stats_df_sorted.head(10).iterrows():
    feat_name = row['Feature'][:48] if len(row['Feature']) > 48 else row['Feature']
    print(f"{feat_name:<50} {row['Missing']:>10.0f} {row['Missing_%']:>12.1f}%")

print("-"*100)

# Overall missing data summary
total_values = len(df) * len(available_features)
total_missing = stats_df['Missing'].sum()
missing_pct = (total_missing / total_values) * 100

print(f"\nOverall missing data:")
print(f"  Total data points: {total_values:,}")
print(f"  Total missing: {total_missing:,.0f}")
print(f"  Missing percentage: {missing_pct:.2f}%")

print("\n" + "="*100)
print("ANALYSIS COMPLETE")
print("="*100)
```

# DESCRIPTIVE STATISTICS BY IRRIGATION TYPE

```python
"""
DESCRIPTIVE STATISTICS BY IRRIGATION TYPE
================================================================================

Creates comprehensive descriptive statistics for all features by irrigation type:
- GAG (Drip irrigation)
- GRV (Surface gravity irrigation)
"""

import pandas as pd
import numpy as np

# ============================================================================
# CONFIGURATION
# ============================================================================

DATA_FILE = 'C:/Users/U078-N800/Desktop/Ml_Internship/Copy/Ml_Training/Regrouped_Analysis/final_dataset_regrouped_categorical.csv'

# All features for analysis
features = [
    'tc_per_ha_actual', 'brix_percent',
    'pH_complete', 'Exchangeable_Calcium_meq_100g_complete', 'Exchangeable_Magnesium_meq_100g_complete',
    'Exchangeable_Potassium_meq_100g_complete', 'Clay_percent_complete', 'Fine_Silt_percent_complete', 
    'Coarse_Silt_percent_complete', 'Fine_Sand_percent_complete', 'Coarse_Sand_percent_complete', 
    'Total_Nitrogen_per_mille_complete', 'Organic_Carbon_per_mille_complete', 'Mg_K_Ratio_complete', 
    'Mg_Ca_Ratio_complete', 'SC_drying_duration_actual_days', 'growing_season_rainfall_mm', 'temp_mini_mean', 
    'temp_maxi_mean', 'temp_moy_mean', 'ampl_mean', 'h_mini_mean', 'h_maxi_mean', 'h_moy_mean', 'h_6h_mean', 
    'h_12h_mean', 'h_16h_mean', 'h_24h_mean', 'soil_temp_6h_mean', 'soil_temp_16h_mean', 'radiation_total', 
    'radiation_ERA_total', 'evaporation_total', 'growing_period_days_weather', 'CWD', 'CDD', 'SDII', 'R10', 
    'R20', 'Rx1', 'Rx5', 'R95p', 'R99p', 'HD', 'TNn', 'TNx', 'TXn', 'TXx', 'tc_per_ha_per_week', 
    'tc_per_ha_52_weeks', 'pol_juice_percent', 'purity_percent', 'fiber_percent', 'pol_cane_percent', 
    'glucose_cane_percent', 'hugot_sugar', 'estimated_recoverable_pol', 'CN_Ratio_complete', 
    'Available_P2O5_mg_kg_complete', 'Compaction_Index_complete', 'Electrical_Conductivity_mmho_cm_complete', 
    'Exchangeable_Sodium_meq_100g_complete', 'total_recoverable_sugar_t_per_ha'
]

# ============================================================================
# LOAD DATA AND CREATE DERIVED VARIABLES
# ============================================================================

print("="*120)
print("DESCRIPTIVE STATISTICS BY IRRIGATION TYPE")
print("="*120)

print(f"\nLoading data from: {DATA_FILE}")
df = pd.read_csv(DATA_FILE)

print(f"Total observations: {len(df)}")

# Check irrigation_type column
if 'irrigation_type' not in df.columns:
    print("\n⚠ ERROR: 'irrigation_type' column not found in dataset!")
    exit()

print(f"\nIrrigation type distribution:")
print(df['irrigation_type'].value_counts())

# Create Total Sand and Total Silt
print("\n" + "="*120)
print("CREATING DERIVED VARIABLES")
print("="*120)

if 'Fine_Sand_percent_complete' in df.columns and 'Coarse_Sand_percent_complete' in df.columns:
    df['Total_Sand_percent'] = df['Fine_Sand_percent_complete'] + df['Coarse_Sand_percent_complete']
    print("✓ Total_Sand_percent created (Fine_Sand + Coarse_Sand)")
    features.append('Total_Sand_percent')
else:
    print("⚠ Could not create Total_Sand_percent - missing columns")

if 'Fine_Silt_percent_complete' in df.columns and 'Coarse_Silt_percent_complete' in df.columns:
    df['Total_Silt_percent'] = df['Fine_Silt_percent_complete'] + df['Coarse_Silt_percent_complete']
    print("✓ Total_Silt_percent created (Fine_Silt + Coarse_Silt)")
    features.append('Total_Silt_percent')
else:
    print("⚠ Could not create Total_Silt_percent - missing columns")

# ============================================================================
# COMPUTE DESCRIPTIVE STATISTICS BY IRRIGATION TYPE
# ============================================================================

print("\n" + "="*120)
print("COMPUTING DESCRIPTIVE STATISTICS BY IRRIGATION TYPE")
print("="*120)

# Filter to only features that exist in the dataset
available_features = [f for f in features if f in df.columns]
missing_features = [f for f in features if f not in df.columns]

if missing_features:
    print(f"\n⚠ Warning: {len(missing_features)} features not found in dataset")

print(f"\nAnalyzing {len(available_features)} features...")

# Split data by irrigation type
df_gag = df[df['irrigation_type'] == 'GAG']
df_grv = df[df['irrigation_type'] == 'GRV']

print(f"\nGAG (Drip): {len(df_gag)} observations")
print(f"GRV (Surface gravity): {len(df_grv)} observations")

# Create statistics for each irrigation type
stats_gag = []
stats_grv = []

for feat in available_features:
    # GAG statistics
    data_gag = df_gag[feat].dropna()
    if len(data_gag) > 0:
        stats_gag.append({
            'Feature': feat,
            'N': len(data_gag),
            'Mean': data_gag.mean(),
            'Std': data_gag.std(),
            'Min': data_gag.min(),
            'Median': data_gag.median(),
            'Max': data_gag.max(),
            'Missing': df_gag[feat].isna().sum(),
            'Missing_%': (df_gag[feat].isna().sum() / len(df_gag)) * 100
        })
    
    # GRV statistics
    data_grv = df_grv[feat].dropna()
    if len(data_grv) > 0:
        stats_grv.append({
            'Feature': feat,
            'N': len(data_grv),
            'Mean': data_grv.mean(),
            'Std': data_grv.std(),
            'Min': data_grv.min(),
            'Median': data_grv.median(),
            'Max': data_grv.max(),
            'Missing': df_grv[feat].isna().sum(),
            'Missing_%': (df_grv[feat].isna().sum() / len(df_grv)) * 100
        })

stats_df_gag = pd.DataFrame(stats_gag)
stats_df_grv = pd.DataFrame(stats_grv)

# ============================================================================
# DISPLAY RESULTS - GAG (DRIP IRRIGATION)
# ============================================================================

print("\n" + "="*120)
print("GAG - DRIP IRRIGATION")
print("="*120)
print(f"Total Observations: {len(df_gag)}")

print("\n" + "-"*120)
print(f"{'Feature':<50} {'N':>8} {'Mean ± Std':>20} {'Min':>12} {'Median':>12} {'Max':>12} {'Missing':>10}")
print("-"*120)

for idx, row in stats_df_gag.iterrows():
    mean_std = f"{row['Mean']:.2f} ± {row['Std']:.2f}"
    feat_name = row['Feature'][:48] if len(row['Feature']) > 48 else row['Feature']
    print(f"{feat_name:<50} {row['N']:>8.0f} {mean_std:>20} {row['Min']:>12.2f} {row['Median']:>12.2f} {row['Max']:>12.2f} {row['Missing']:>10.0f}")

print("-"*120)

# ============================================================================
# DISPLAY RESULTS - GRV (SURFACE GRAVITY IRRIGATION)
# ============================================================================

print("\n" + "="*120)
print("GRV - SURFACE GRAVITY IRRIGATION")
print("="*120)
print(f"Total Observations: {len(df_grv)}")

print("\n" + "-"*120)
print(f"{'Feature':<50} {'N':>8} {'Mean ± Std':>20} {'Min':>12} {'Median':>12} {'Max':>12} {'Missing':>10}")
print("-"*120)

for idx, row in stats_df_grv.iterrows():
    mean_std = f"{row['Mean']:.2f} ± {row['Std']:.2f}"
    feat_name = row['Feature'][:48] if len(row['Feature']) > 48 else row['Feature']
    print(f"{feat_name:<50} {row['N']:>8.0f} {mean_std:>20} {row['Min']:>12.2f} {row['Median']:>12.2f} {row['Max']:>12.2f} {row['Missing']:>10.0f}")

print("-"*120)

# ============================================================================
# SIDE-BY-SIDE COMPARISON
# ============================================================================

print("\n" + "="*120)
print("SIDE-BY-SIDE COMPARISON: GAG vs GRV")
print("="*120)

print("\n" + "-"*120)
print(f"{'Feature':<40} {'GAG Mean ± Std':>25} {'GRV Mean ± Std':>25} {'Difference':>15}")
print("-"*120)

# Merge the two dataframes for comparison
comparison_data = []

for feat in available_features:
    gag_row = stats_df_gag[stats_df_gag['Feature'] == feat]
    grv_row = stats_df_grv[stats_df_grv['Feature'] == feat]
    
    if not gag_row.empty and not grv_row.empty:
        gag_mean = gag_row.iloc[0]['Mean']
        gag_std = gag_row.iloc[0]['Std']
        grv_mean = grv_row.iloc[0]['Mean']
        grv_std = grv_row.iloc[0]['Std']
        
        diff = gag_mean - grv_mean
        diff_pct = (diff / grv_mean * 100) if grv_mean != 0 else 0
        
        feat_name = feat[:38] if len(feat) > 38 else feat
        gag_str = f"{gag_mean:.2f} ± {gag_std:.2f}"
        grv_str = f"{grv_mean:.2f} ± {grv_std:.2f}"
        diff_str = f"{diff:+.2f} ({diff_pct:+.1f}%)"
        
        print(f"{feat_name:<40} {gag_str:>25} {grv_str:>25} {diff_str:>15}")
        
        comparison_data.append({
            'Feature': feat,
            'GAG_Mean': gag_mean,
            'GAG_Std': gag_std,
            'GRV_Mean': grv_mean,
            'GRV_Std': grv_std,
            'Difference': diff,
            'Difference_%': diff_pct
        })

print("-"*120)

comparison_df = pd.DataFrame(comparison_data)

# ============================================================================
# KEY VARIABLES COMPARISON
# ============================================================================

print("\n" + "="*120)
print("KEY VARIABLES DETAILED COMPARISON")
print("="*120)

key_vars = ['tc_per_ha_actual', 'hugot_sugar', 'total_recoverable_sugar_t_per_ha']

for var in key_vars:
    if var in available_features:
        print(f"\n{var}:")
        print("-" * 80)
        
        gag_row = stats_df_gag[stats_df_gag['Feature'] == var].iloc[0]
        grv_row = stats_df_grv[stats_df_grv['Feature'] == var].iloc[0]
        
        print(f"  GAG (Drip irrigation):")
        print(f"    N: {gag_row['N']:.0f}")
        print(f"    Mean ± Std: {gag_row['Mean']:.2f} ± {gag_row['Std']:.2f}")
        print(f"    Range: [{gag_row['Min']:.2f}, {gag_row['Max']:.2f}]")
        print(f"    Median: {gag_row['Median']:.2f}")
        
        print(f"\n  GRV (Surface gravity irrigation):")
        print(f"    N: {grv_row['N']:.0f}")
        print(f"    Mean ± Std: {grv_row['Mean']:.2f} ± {grv_row['Std']:.2f}")
        print(f"    Range: [{grv_row['Min']:.2f}, {grv_row['Max']:.2f}]")
        print(f"    Median: {grv_row['Median']:.2f}")
        
        diff = gag_row['Mean'] - grv_row['Mean']
        diff_pct = (diff / grv_row['Mean'] * 100) if grv_row['Mean'] != 0 else 0
        
        print(f"\n  Difference (GAG - GRV): {diff:+.2f} ({diff_pct:+.1f}%)")

# ============================================================================
# DERIVED VARIABLES COMPARISON
# ============================================================================

print("\n" + "="*120)
print("DERIVED VARIABLES COMPARISON")
print("="*120)

derived_vars = ['Total_Sand_percent', 'Total_Silt_percent']

print("\n" + "-"*120)
print(f"{'Variable':<30} {'GAG Mean ± Std':>25} {'GRV Mean ± Std':>25} {'Difference':>15}")
print("-"*120)

for var in derived_vars:
    if var in available_features:
        gag_row = stats_df_gag[stats_df_gag['Feature'] == var]
        grv_row = stats_df_grv[stats_df_grv['Feature'] == var]
        
        if not gag_row.empty and not grv_row.empty:
            gag_row = gag_row.iloc[0]
            grv_row = grv_row.iloc[0]
            
            gag_str = f"{gag_row['Mean']:.2f} ± {gag_row['Std']:.2f}"
            grv_str = f"{grv_row['Mean']:.2f} ± {grv_row['Std']:.2f}"
            
            diff = gag_row['Mean'] - grv_row['Mean']
            diff_pct = (diff / grv_row['Mean'] * 100) if grv_row['Mean'] != 0 else 0
            diff_str = f"{diff:+.2f} ({diff_pct:+.1f}%)"
            
            print(f"{var:<30} {gag_str:>25} {grv_str:>25} {diff_str:>15}")

print("-"*120)

# ============================================================================
# SOIL TEXTURE COMPARISON
# ============================================================================

print("\n" + "="*120)
print("SOIL TEXTURE COMPONENTS COMPARISON")
print("="*120)

soil_vars = ['Clay_percent_complete', 'Total_Silt_percent', 'Total_Sand_percent']

print("\n" + "-"*120)
print(f"{'Component':<30} {'GAG Mean ± Std':>25} {'GRV Mean ± Std':>25} {'Difference':>15}")
print("-"*120)

for var in soil_vars:
    if var in available_features:
        gag_row = stats_df_gag[stats_df_gag['Feature'] == var]
        grv_row = stats_df_grv[stats_df_grv['Feature'] == var]
        
        if not gag_row.empty and not grv_row.empty:
            gag_row = gag_row.iloc[0]
            grv_row = grv_row.iloc[0]
            
            gag_str = f"{gag_row['Mean']:.2f} ± {gag_row['Std']:.2f}"
            grv_str = f"{grv_row['Mean']:.2f} ± {grv_row['Std']:.2f}"
            
            diff = gag_row['Mean'] - grv_row['Mean']
            diff_pct = (diff / grv_row['Mean'] * 100) if grv_row['Mean'] != 0 else 0
            diff_str = f"{diff:+.2f} ({diff_pct:+.1f}%)"
            
            print(f"{var:<30} {gag_str:>25} {grv_str:>25} {diff_str:>15}")

print("-"*120)

# ============================================================================
# LARGEST DIFFERENCES
# ============================================================================

print("\n" + "="*120)
print("TOP 10 LARGEST DIFFERENCES (GAG vs GRV)")
print("="*120)

# Sort by absolute difference percentage
comparison_df_sorted = comparison_df.copy()
comparison_df_sorted['Abs_Diff_%'] = comparison_df_sorted['Difference_%'].abs()
comparison_df_sorted = comparison_df_sorted.sort_values('Abs_Diff_%', ascending=False)

print("\n" + "-"*120)
print(f"{'Feature':<50} {'GAG Mean':>12} {'GRV Mean':>12} {'Difference':>15} {'Diff %':>12}")
print("-"*120)

for idx, row in comparison_df_sorted.head(10).iterrows():
    feat_name = row['Feature'][:48] if len(row['Feature']) > 48 else row['Feature']
    print(f"{feat_name:<50} {row['GAG_Mean']:>12.2f} {row['GRV_Mean']:>12.2f} {row['Difference']:>15.2f} {row['Difference_%']:>11.1f}%")

print("-"*120)

# ============================================================================
# SAVE RESULTS TO CSV
# ============================================================================

print("\n" + "="*120)
print("SAVING RESULTS TO FILES")
print("="*120)

# Save individual irrigation type statistics
stats_df_gag.to_csv('descriptive_stats_GAG.csv', index=False)
print("✓ GAG statistics saved to: descriptive_stats_GAG.csv")

stats_df_grv.to_csv('descriptive_stats_GRV.csv', index=False)
print("✓ GRV statistics saved to: descriptive_stats_GRV.csv")

# Save comparison
comparison_df.to_csv('descriptive_stats_comparison_GAG_vs_GRV.csv', index=False)
print("✓ Comparison saved to: descriptive_stats_comparison_GAG_vs_GRV.csv")

# ============================================================================
# MISSING DATA SUMMARY BY IRRIGATION TYPE
# ============================================================================

print("\n" + "="*120)
print("MISSING DATA SUMMARY BY IRRIGATION TYPE")
print("="*120)

print("\nGAG (Drip irrigation):")
print(f"  Total features: {len(stats_df_gag)}")
print(f"  Total possible values: {len(df_gag) * len(available_features):,}")
print(f"  Total missing: {stats_df_gag['Missing'].sum():,.0f}")
print(f"  Missing percentage: {(stats_df_gag['Missing'].sum() / (len(df_gag) * len(available_features)) * 100):.2f}%")

print("\nGRV (Surface gravity irrigation):")
print(f"  Total features: {len(stats_df_grv)}")
print(f"  Total possible values: {len(df_grv) * len(available_features):,}")
print(f"  Total missing: {stats_df_grv['Missing'].sum():,.0f}")
print(f"  Missing percentage: {(stats_df_grv['Missing'].sum() / (len(df_grv) * len(available_features)) * 100):.2f}%")

print("\n" + "="*120)
print("ANALYSIS COMPLETE")
print("="*120)
```
